# 09 GHSL Delhi Metropolitan Production
Uses the frozen Delhi-validated baseline. Run one cell at a time. Cells 3-6 are gates; do not launch Cell 7 unless all pass.

70 km × 70 km = 4,900 km²
1 km × 1 km cells = 4,900 canonical cells per epoch
10 epochs = 49,000 city-grid-year rows per city


```python
# CELL 1: BOOTSTRAP
from google.colab import drive
from pathlib import Path
import os, shutil, subprocess
drive.mount('/content/drive', force_remount=False)
D=Path('/content/drive/MyDrive'); R=D/'remote_exec.sh'; K=D/'hadoop-hdfs.pem'; L=Path('/content/hadoop-hdfs.pem')
if not R.is_file(): raise FileNotFoundError(R)
os.chmod(R,0o700)
if not L.is_file(): shutil.copy2(K,L)
os.chmod(L,0o600)
def run_remote(command,timeout=300,show_output=True):
    r=subprocess.run(['bash',str(R),command],text=True,capture_output=True,timeout=timeout)
    if show_output and r.stdout: print(r.stdout)
    if show_output and r.stderr: print('REMOTE STDERR:'); print(r.stderr)
    if r.returncode: raise RuntimeError(f'Remote command failed with return code {r.returncode}.')
    return r
print('CELL 1: SUCCESS')
```

```python
# CELL 2: FROZEN DELHI NCR CONFIGURATION

ACTIVE_CITY_SLUG = "delhi_ncr"

ACTIVE_CITY = {
    "city_name": "Delhi NCR",
    "city_slug": "delhi_ncr",
    "seed_lon": 77.2090,
    "seed_lat": 28.6139,
    "target_crs": "ESRI:54009",
    "expected_grid_cells": 4900,
    "expected_complete_cells": 4900,
    "expected_partial_cells": 0,
    "extent_mollweide": {
        "min_x": 7103000.0,
        "min_y": 3446000.0,
        "max_x": 7173000.0,
        "max_y": 3516000.0,
    },
    "hdfs_output_base": (
        "hdfs://localhost:9000"
        "/data/ghsl/curated/cities/"
        "city=delhi_ncr"
    ),
    "remote_work_dir": (
        "/home/ubuntu/ghsl_multicity/delhi_ncr"
    ),
}

assert (
    ACTIVE_CITY["extent_mollweide"]["max_x"]
    - ACTIVE_CITY["extent_mollweide"]["min_x"]
    == 70000
)

print(ACTIVE_CITY)
print("CELL 2: SUCCESS")
```

```python
# ============================================================
# CELL 3: BASELINE AND HDFS PREFLIGHT
#
# Corrected behavior:
#   - Verifies the frozen SHA-256 inventory
#   - Compiles baseline Python files into /tmp
#   - Does not create __pycache__ in the baseline
#   - Removes bytecode created by the previous Cell 3 attempt
#   - Verifies all permanent baseline files are read-only
#
# Does not:
#   - modify baseline source files
#   - launch Spark
#   - modify HDFS
# ============================================================

cmd = r"""
set -e

source /home/ubuntu/ghsl-sedona-env.sh

export HADOOP_HOME=/usr/local/hadoop
export PATH="${HADOOP_HOME}/bin:${HADOOP_HOME}/sbin:${PATH}"

BASE="/home/ubuntu/ghsl_multicity_baseline"
COMPILE_ROOT="/tmp/ghsl_baseline_compile_check"

echo "============================================================"
echo "DELHI NCR BASELINE AND HDFS PREFLIGHT"
echo "============================================================"


# ------------------------------------------------------------
# HDFS service validation.
# ------------------------------------------------------------

for daemon in \
    NameNode \
    DataNode \
    SecondaryNameNode

do
    jps \
        | awk '{print $2}' \
        | grep -qx "$daemon"
done

hdfs dfsadmin -safemode get \
    | grep -qi "Safe mode is OFF"

echo "HDFS services: PASS"


# ------------------------------------------------------------
# Required baseline files.
# ------------------------------------------------------------

required_files=(
    "$BASE/baseline_manifest.json"
    "$BASE/SHA256SUMS"
    "$BASE/README.txt"
    "$BASE/source_city_config.json"
    "$BASE/batch/run_city_epochs.py"
    "$BASE/templates/population_epoch_template.py"
    "$BASE/templates/smod_epoch_template.py"
    "$BASE/templates/built_surface_epoch_template.py"
)

for file in "${required_files[@]}"; do
    if [ ! -f "$file" ]; then
        echo "ERROR: Required baseline file missing:"
        echo "$file"
        exit 20
    fi
done

echo "Required baseline files: PASS"


# ------------------------------------------------------------
# Verify the permanent baseline inventory before doing any
# compilation.
# ------------------------------------------------------------

cd "$BASE"

sha256sum -c SHA256SUMS

echo "Baseline SHA-256 inventory: PASS"


# ------------------------------------------------------------
# Remove only generated Python bytecode created by the earlier
# preflight attempt.
#
# No source, manifest, checksum, config, or README is removed.
# ------------------------------------------------------------

find "$BASE" \
    -type d \
    -name "__pycache__" \
    -print \
    -exec rm -rf {} + \
    2>/dev/null \
    || true

find "$BASE" \
    -type f \
    \( -name "*.pyc" -o -name "*.pyo" \) \
    -print \
    -delete \
    2>/dev/null \
    || true

echo "Generated baseline bytecode removed: PASS"


# ------------------------------------------------------------
# Compile outside the frozen baseline.
# ------------------------------------------------------------

rm -rf "$COMPILE_ROOT"

mkdir -p \
    "$COMPILE_ROOT/batch" \
    "$COMPILE_ROOT/templates"

PYTHONPYCACHEPREFIX="$COMPILE_ROOT" \
python3 -m py_compile \
    "$BASE/batch/run_city_epochs.py"

for file in \
    "$BASE/templates/population_epoch_template.py" \
    "$BASE/templates/smod_epoch_template.py" \
    "$BASE/templates/built_surface_epoch_template.py"

do
    PYTHONPYCACHEPREFIX="$COMPILE_ROOT" \
    python3 -m py_compile "$file"
done

echo "Baseline Python compilation: PASS"


# ------------------------------------------------------------
# Confirm compilation created nothing inside the baseline.
# ------------------------------------------------------------

if find "$BASE" \
    -type d \
    -name "__pycache__" \
    | grep -q .
then
    echo "ERROR: Compilation created __pycache__ in baseline."
    exit 21
fi

if find "$BASE" \
    -type f \
    \( -name "*.pyc" -o -name "*.pyo" \) \
    | grep -q .
then
    echo "ERROR: Compilation created bytecode in baseline."
    exit 22
fi

echo "Baseline remains bytecode-free: PASS"


# ------------------------------------------------------------
# Verify all permanent baseline files are read-only.
#
# The baseline directories remain owner-writable so future
# city workspaces can be created elsewhere. Only files need
# to remain frozen.
# ------------------------------------------------------------

writable_files="$(
    find "$BASE" \
        -type f \
        -perm -u=w \
        -print
)"

if [ -n "$writable_files" ]; then
    echo "ERROR: Writable permanent baseline files found:"
    echo "$writable_files"
    exit 23
fi

echo "Baseline permanent files read-only: PASS"


# ------------------------------------------------------------
# Recheck the inventory after cleanup and compilation.
# ------------------------------------------------------------

cd "$BASE"

sha256sum -c SHA256SUMS

echo "Post-check SHA-256 inventory: PASS"


# ------------------------------------------------------------
# Clean temporary compilation output.
# ------------------------------------------------------------

rm -rf "$COMPILE_ROOT"

echo
echo "BASELINE_PREFLIGHT_SUCCESS"
"""

preflight_result = run_remote(
    cmd,
    timeout=300,
    show_output=True,
)

print()
print("=" * 80)
print("CELL 3 RESULT")
print("=" * 80)

print(
    "Return code:",
    preflight_result.returncode,
)

print("Spark launched: NO")
print("HDFS modified: NO")
print("Baseline source files modified: NO")
print("CELL 3: SUCCESS")
```

```python
# ============================================================
# CELL 4: CREATE DELHI NCR WORKSPACE FROM FROZEN BASELINE
#
# Source:
#   /home/ubuntu/ghsl_multicity_baseline
#
# Destination:
#   /home/ubuntu/ghsl_multicity/delhi_ncr
#
# Safety:
#   - Refuses to overwrite an existing partial workspace
#   - Removes its own partial workspace if this build fails
#   - Does not change the frozen baseline
#   - Does not change Hyderabad or Bengaluru
#   - Does not launch Spark
#   - Does not modify HDFS
# ============================================================

import base64
import hashlib
import json
import textwrap


# ============================================================
# VALIDATE NOTEBOOK STATE
# ============================================================

for required_name in [
    "ACTIVE_CITY",
    "ACTIVE_CITY_SLUG",
    "run_remote",
]:
    if required_name not in globals():
        raise RuntimeError(
            f"{required_name} is unavailable. "
            "Rerun Cells 1 and 2."
        )


if ACTIVE_CITY_SLUG != "delhi_ncr":
    raise RuntimeError(
        "ACTIVE_CITY_SLUG must be delhi_ncr."
    )


expected_workspace = (
    "/home/ubuntu/ghsl_multicity/delhi_ncr"
)

expected_hdfs_root = (
    "hdfs://localhost:9000"
    "/data/ghsl/curated/cities/"
    "city=delhi_ncr"
)

expected_extent = {
    "min_x": 7103000.0,
    "min_y": 3446000.0,
    "max_x": 7173000.0,
    "max_y": 3516000.0,
}


if (
    ACTIVE_CITY["remote_work_dir"]
    != expected_workspace
):
    raise RuntimeError(
        "Unexpected Delhi NCR workspace: "
        + ACTIVE_CITY["remote_work_dir"]
    )


if (
    ACTIVE_CITY["hdfs_output_base"]
    != expected_hdfs_root
):
    raise RuntimeError(
        "Unexpected Delhi NCR HDFS root: "
        + ACTIVE_CITY["hdfs_output_base"]
    )


if (
    ACTIVE_CITY["city_name"]
    != "Delhi NCR"
):
    raise RuntimeError(
        "Unexpected Delhi NCR city name."
    )


if (
    ACTIVE_CITY["extent_mollweide"]
    != expected_extent
):
    raise RuntimeError(
        "Unexpected Delhi NCR extent.\n"
        f"Expected: {expected_extent}\n"
        f"Actual:   "
        f"{ACTIVE_CITY['extent_mollweide']}"
    )


if (
    ACTIVE_CITY["expected_grid_cells"]
    != 4900
):
    raise RuntimeError(
        "Delhi NCR expected_grid_cells must be 4900."
    )


# ============================================================
# SERIALIZE DELHI NCR CONFIGURATION
# ============================================================

config_json = (
    json.dumps(
        ACTIVE_CITY,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)

config_b64 = base64.b64encode(
    config_json.encode("utf-8")
).decode("ascii")


# ============================================================
# BUILD REMOTE WORKSPACE CREATOR
# ============================================================

remote_builder = textwrap.dedent(
    r'''
    from pathlib import Path
    import ast
    import hashlib
    import json
    import py_compile
    import shutil
    import sys
    import time


    BASELINE = Path(
        "/home/ubuntu/ghsl_multicity_baseline"
    )

    TARGET = Path(
        "/home/ubuntu/ghsl_multicity/delhi_ncr"
    )

    CONFIG_PATH = Path(
        sys.argv[1]
    )

    TEMPLATE_NAMES = [
        "population_epoch_template.py",
        "smod_epoch_template.py",
        "built_surface_epoch_template.py",
    ]


    def sha256_file(path):
        digest = hashlib.sha256()

        with path.open("rb") as handle:
            for block in iter(
                lambda: handle.read(
                    1024 * 1024
                ),
                b"",
            ):
                digest.update(block)

        return digest.hexdigest()


    def compile_source(source, path):
        compile(
            source,
            str(path),
            "exec",
        )


    # ========================================================
    # LOAD CONFIGURATIONS
    # ========================================================

    if not CONFIG_PATH.is_file():
        raise FileNotFoundError(
            CONFIG_PATH
        )

    config = json.loads(
        CONFIG_PATH.read_text(
            encoding="utf-8"
        )
    )

    source_config_path = (
        BASELINE
        / "source_city_config.json"
    )

    if not source_config_path.is_file():
        raise FileNotFoundError(
            source_config_path
        )

    source_config = json.loads(
        source_config_path.read_text(
            encoding="utf-8"
        )
    )


    # ========================================================
    # VALIDATE DELHI NCR CONFIGURATION
    # ========================================================

    expected_extent = {
        "min_x": 7103000.0,
        "min_y": 3446000.0,
        "max_x": 7173000.0,
        "max_y": 3516000.0,
    }

    if (
        config.get("city_slug")
        != "delhi_ncr"
    ):
        raise RuntimeError(
            "Delhi NCR configuration has the wrong slug."
        )

    if (
        config.get("city_name")
        != "Delhi NCR"
    ):
        raise RuntimeError(
            "Delhi NCR configuration has the wrong name."
        )

    if (
        config.get("remote_work_dir")
        != str(TARGET)
    ):
        raise RuntimeError(
            "Delhi NCR configuration has the wrong "
            "remote workspace."
        )

    if (
        config.get("hdfs_output_base")
        != (
            "hdfs://localhost:9000"
            "/data/ghsl/curated/cities/"
            "city=delhi_ncr"
        )
    ):
        raise RuntimeError(
            "Delhi NCR configuration has the wrong "
            "HDFS output root."
        )

    if (
        config.get("expected_grid_cells")
        != 4900
    ):
        raise RuntimeError(
            "Delhi NCR expected_grid_cells must be 4900."
        )

    if (
        config.get("extent_mollweide")
        != expected_extent
    ):
        raise RuntimeError(
            "Delhi NCR configuration has the wrong extent."
        )


    # ========================================================
    # VALIDATE BASELINE INPUTS
    # ========================================================

    baseline_driver = (
        BASELINE
        / "batch"
        / "run_city_epochs.py"
    )

    baseline_manifest = (
        BASELINE
        / "baseline_manifest.json"
    )

    baseline_checksums = (
        BASELINE
        / "SHA256SUMS"
    )

    required_baseline_files = [
        baseline_driver,
        baseline_manifest,
        baseline_checksums,
        source_config_path,
    ]

    for required_path in (
        required_baseline_files
    ):
        if not required_path.is_file():
            raise FileNotFoundError(
                required_path
            )

    baseline_templates = {}

    for template_name in TEMPLATE_NAMES:
        template_path = (
            BASELINE
            / "templates"
            / template_name
        )

        if not template_path.is_file():
            raise FileNotFoundError(
                template_path
            )

        baseline_templates[
            template_name
        ] = template_path


    # ========================================================
    # CHECK FOR AN EXISTING WORKSPACE
    # ========================================================

    workspace_manifest = (
        TARGET
        / "delhi_ncr_workspace_manifest.json"
    )

    if TARGET.exists():
        if workspace_manifest.is_file():
            print("=" * 80)
            print("DELHI NCR WORKSPACE CREATION")
            print("=" * 80)

            print(
                "Existing completed Delhi NCR "
                "workspace found."
            )

            print(
                "Workspace:",
                TARGET,
            )

            print(
                "Existing workspace was not "
                "overwritten."
            )

            print(
                "DELHI_NCR_WORKSPACE_ALREADY_EXISTS"
            )

            print(
                "DELHI_NCR_WORKSPACE_BUILD_SUCCESS"
            )

            raise SystemExit(0)

        raise RuntimeError(
            "An incomplete Delhi NCR workspace already "
            "exists and was not overwritten: "
            + str(TARGET)
        )


    # ========================================================
    # BUILD PARAMETER REPLACEMENTS
    # ========================================================

    source_root = str(
        source_config[
            "remote_work_dir"
        ]
    )

    target_root = str(
        config[
            "remote_work_dir"
        ]
    )

    source_slug = str(
        source_config[
            "city_slug"
        ]
    )

    target_slug = str(
        config[
            "city_slug"
        ]
    )

    source_name = str(
        source_config[
            "city_name"
        ]
    )

    target_name = str(
        config[
            "city_name"
        ]
    )

    source_extent = (
        source_config[
            "extent_mollweide"
        ]
    )

    target_extent = (
        config[
            "extent_mollweide"
        ]
    )

    replacements = [
        (
            source_root,
            target_root,
        ),
        (
            "city=" + source_slug,
            "city=" + target_slug,
        ),
        (
            source_slug,
            target_slug,
        ),
        (
            source_name,
            target_name,
        ),
    ]

    for key in [
        "min_x",
        "min_y",
        "max_x",
        "max_y",
    ]:
        replacements.append(
            (
                str(
                    float(
                        source_extent[key]
                    )
                ),
                str(
                    float(
                        target_extent[key]
                    )
                ),
            )
        )

        replacements.append(
            (
                str(
                    int(
                        source_extent[key]
                    )
                ),
                str(
                    int(
                        target_extent[key]
                    )
                ),
            )
        )


    def parameterize(source):
        output = source

        for old_value, new_value in replacements:
            output = output.replace(
                old_value,
                new_value,
            )

        return output


    # ========================================================
    # CREATE WORKSPACE
    # ========================================================

    try:
        (
            TARGET
            / "templates"
        ).mkdir(
            parents=True,
            exist_ok=False,
        )

        (
            TARGET
            / "programs"
        ).mkdir()

        (
            TARGET
            / "batch"
        ).mkdir()


        # ====================================================
        # PARAMETERIZE TEMPLATES
        # ====================================================

        template_hashes = {}

        for template_name in TEMPLATE_NAMES:
            source_path = (
                baseline_templates[
                    template_name
                ]
            )

            destination_path = (
                TARGET
                / "templates"
                / template_name
            )

            template_source = (
                source_path.read_text(
                    encoding="utf-8"
                )
            )

            template_source = parameterize(
                template_source
            )

            compile_source(
                template_source,
                destination_path,
            )

            destination_path.write_text(
                template_source,
                encoding="utf-8",
            )

            py_compile.compile(
                str(destination_path),
                doraise=True,
            )

            template_hashes[
                template_name
            ] = sha256_file(
                destination_path
            )


        # ====================================================
        # PARAMETERIZE DRIVER
        # ====================================================

        driver_destination = (
            TARGET
            / "batch"
            / "run_city_epochs.py"
        )

        driver_source = (
            baseline_driver.read_text(
                encoding="utf-8"
            )
        )

        driver_source = parameterize(
            driver_source
        )


        # ====================================================
        # REFRESH TEMPLATE CHECKSUMS THROUGH AST
        # ====================================================

        driver_tree = ast.parse(
            driver_source
        )

        checksum_node = None

        for node in driver_tree.body:
            if not isinstance(
                node,
                ast.Assign,
            ):
                continue

            if any(
                isinstance(
                    target,
                    ast.Name,
                )
                and target.id
                == "EXPECTED_TEMPLATE_SHA256"
                for target
                in node.targets
            ):
                checksum_node = node
                break

        if checksum_node is None:
            raise RuntimeError(
                "EXPECTED_TEMPLATE_SHA256 "
                "was not found."
            )

        new_checksum_dictionary = {
            "population": (
                template_hashes[
                    "population_epoch_template.py"
                ]
            ),
            "smod": (
                template_hashes[
                    "smod_epoch_template.py"
                ]
            ),
            "built_surface": (
                template_hashes[
                    "built_surface_epoch_template.py"
                ]
            ),
        }

        driver_lines = (
            driver_source.splitlines(
                keepends=True
            )
        )

        checksum_assignment = (
            "EXPECTED_TEMPLATE_SHA256 = "
            + repr(
                new_checksum_dictionary
            )
            + chr(10)
        )

        driver_lines[
            checksum_node.lineno - 1:
            checksum_node.end_lineno
        ] = [
            checksum_assignment
        ]

        driver_source = "".join(
            driver_lines
        )

        compile_source(
            driver_source,
            driver_destination,
        )

        driver_destination.write_text(
            driver_source,
            encoding="utf-8",
        )

        py_compile.compile(
            str(driver_destination),
            doraise=True,
        )


        # ====================================================
        # SAVE CONFIGURATION
        # ====================================================

        delhi_config_path = (
            TARGET
            / "city_config.json"
        )

        delhi_config_path.write_text(
            json.dumps(
                config,
                indent=2,
                sort_keys=True,
            )
            + chr(10),
            encoding="utf-8",
        )


        # ====================================================
        # CODE-LEVEL PARAMETER AUDIT
        #
        # The full display name is validated in city_config,
        # not required inside driver/template source.
        # ====================================================

        combined_source = (
            driver_source
        )

        for template_name in TEMPLATE_NAMES:
            combined_source += (
                TARGET
                / "templates"
                / template_name
            ).read_text(
                encoding="utf-8"
            )

        forbidden_values = [
            source_root,
            "city=" + source_slug,
            source_name,
        ]

        remaining_forbidden = [
            value
            for value in forbidden_values
            if value in combined_source
        ]

        if remaining_forbidden:
            raise RuntimeError(
                "Hyderabad leakage remains: "
                + repr(
                    remaining_forbidden
                )
            )

        required_code_values = [
            target_root,
            "city=" + target_slug,
            "7103000",
            "3446000",
            "7173000",
            "3516000",
        ]

        missing_code_values = [
            value
            for value
            in required_code_values
            if value
            not in combined_source
        ]

        if missing_code_values:
            raise RuntimeError(
                "Required Delhi NCR code parameters "
                "are missing: "
                + repr(
                    missing_code_values
                )
            )


        # ====================================================
        # CONFIGURATION-LEVEL AUDIT
        # ====================================================

        saved_config = json.loads(
            delhi_config_path.read_text(
                encoding="utf-8"
            )
        )

        if (
            saved_config["city_name"]
            != "Delhi NCR"
        ):
            raise RuntimeError(
                "Delhi NCR city name is incorrect "
                "in city_config.json."
            )

        if (
            saved_config["city_slug"]
            != "delhi_ncr"
        ):
            raise RuntimeError(
                "Delhi NCR city slug is incorrect "
                "in city_config.json."
            )

        if (
            saved_config["extent_mollweide"]
            != expected_extent
        ):
            raise RuntimeError(
                "Delhi NCR extent is incorrect "
                "in city_config.json."
            )


        # ====================================================
        # VERIFY CHECKSUM DICTIONARY FROM SAVED DRIVER
        # ====================================================

        saved_driver_tree = ast.parse(
            driver_destination.read_text(
                encoding="utf-8"
            )
        )

        saved_checksum_dictionary = None

        for node in saved_driver_tree.body:
            if not isinstance(
                node,
                ast.Assign,
            ):
                continue

            if any(
                isinstance(
                    target,
                    ast.Name,
                )
                and target.id
                == "EXPECTED_TEMPLATE_SHA256"
                for target
                in node.targets
            ):
                saved_checksum_dictionary = (
                    ast.literal_eval(
                        node.value
                    )
                )

                break

        if (
            saved_checksum_dictionary
            != new_checksum_dictionary
        ):
            raise RuntimeError(
                "Saved driver checksum dictionary "
                "does not match Delhi NCR templates."
            )


        # ====================================================
        # REMOVE GENERATED BYTECODE
        # ====================================================

        for pycache in TARGET.rglob(
            "__pycache__"
        ):
            shutil.rmtree(
                pycache
            )


        # ====================================================
        # SAVE WORKSPACE MANIFEST
        # ====================================================

        manifest = {
            "city_slug": target_slug,
            "city_name": target_name,
            "source_baseline": str(
                BASELINE
            ),
            "workspace": str(
                TARGET
            ),
            "driver_sha256": sha256_file(
                driver_destination
            ),
            "template_sha256": (
                new_checksum_dictionary
            ),
            "extent_mollweide": (
                expected_extent
            ),
            "expected_grid_cells": 4900,
            "created_at_utc": (
                time.strftime(
                    "%Y-%m-%dT%H:%M:%SZ",
                    time.gmtime(),
                )
            ),
        }

        workspace_manifest.write_text(
            json.dumps(
                manifest,
                indent=2,
                sort_keys=True,
            )
            + chr(10),
            encoding="utf-8",
        )


    except Exception:
        # Do not leave behind a partial Delhi NCR workspace.
        if TARGET.exists():
            shutil.rmtree(
                TARGET
            )

        raise


    # ========================================================
    # FINAL REPORT
    # ========================================================

    print("=" * 80)
    print("DELHI NCR WORKSPACE CREATION")
    print("=" * 80)

    print(
        "Workspace:",
        TARGET,
    )

    print(
        "Driver:",
        TARGET
        / "batch"
        / "run_city_epochs.py",
    )

    print(
        "Configuration:",
        TARGET
        / "city_config.json",
    )

    print(
        "Manifest:",
        workspace_manifest,
    )

    print()
    print("Template parameterization: PASS")
    print("Template compilation: PASS")
    print("Driver parameterization: PASS")
    print("Driver compilation: PASS")
    print("Template checksum refresh: PASS")
    print("Saved checksum verification: PASS")
    print("Hyderabad leakage audit: PASS")
    print("Delhi NCR code-parameter audit: PASS")
    print("Delhi NCR configuration audit: PASS")
    print("Delhi NCR extent validation: PASS")
    print("Partial-build cleanup protection: PASS")
    print("Spark launched: NO")
    print("HDFS modified: NO")
    print("DELHI_NCR_WORKSPACE_BUILD_SUCCESS")
    '''
).strip() + "\n"


# ============================================================
# LOCAL BUILDER VALIDATION
# ============================================================

compile(
    remote_builder,
    "build_delhi_ncr_workspace.py",
    "exec",
)

builder_sha256 = hashlib.sha256(
    remote_builder.encode("utf-8")
).hexdigest()

print(
    "LOCAL DELHI NCR BUILDER COMPILE: PASS"
)

print(
    "Expected builder SHA-256:",
    builder_sha256,
)


# ============================================================
# TRANSFER AND EXECUTE
# ============================================================

builder_b64 = base64.b64encode(
    remote_builder.encode("utf-8")
).decode("ascii")


remote_command = f"""
set -e

BUILDER="/tmp/build_delhi_ncr_workspace.py"
CONFIG="/tmp/delhi_ncr_city_config.json"

EXPECTED_BUILDER_SHA256="{builder_sha256}"

echo "{builder_b64}" \
| base64 -d \
> "${{BUILDER}}"

echo "{config_b64}" \
| base64 -d \
> "${{CONFIG}}"

ACTUAL_BUILDER_SHA256="$(
    sha256sum "${{BUILDER}}" \
    | awk '{{print $1}}'
)"

echo "Remote builder SHA-256: ${{ACTUAL_BUILDER_SHA256}}"

if [ "${{ACTUAL_BUILDER_SHA256}}" != "${{EXPECTED_BUILDER_SHA256}}" ]; then
    echo "ERROR: Builder checksum mismatch."
    exit 20
fi

echo "REMOTE BUILDER CHECKSUM: PASS"

python3 -m py_compile "${{BUILDER}}"

echo "REMOTE BUILDER COMPILE: PASS"

python3 "${{BUILDER}}" "${{CONFIG}}"
"""


build_result = run_remote(
    remote_command,
    timeout=300,
    show_output=True,
)


print()
print("=" * 80)
print("CELL 4 RESULT")
print("=" * 80)

print(
    "Return code:",
    build_result.returncode,
)

print(
    "Workspace:",
    ACTIVE_CITY["remote_work_dir"],
)

print("Spark launched: NO")
print("HDFS modified: NO")
print("CELL 4: SUCCESS")
```

```python
# ============================================================
# CELL 5: DELHI NCR STATIC AND LEAKAGE AUDIT
#
# This version deliberately avoids:
#   - outer f-strings around embedded Python
#   - fragile grep checks for Python spacing
#   - creating __pycache__ inside the workspace
#
# It verifies:
#   - required Delhi NCR files
#   - Python compilation
#   - Delhi NCR configuration
#   - driver functions
#   - 4,900-cell expectation
#   - template checksums
#   - workspace-manifest checksums
#   - Delhi NCR paths and extent
#   - absence of Hyderabad leakage
#   - generalized endpoint logic
#   - generalized 2020 guards
#   - absence of stale launch state
#
# Read-only toward source files and HDFS.
# ============================================================

root = ACTIVE_CITY["remote_work_dir"]

if ACTIVE_CITY_SLUG != "delhi_ncr":
    raise RuntimeError(
        "ACTIVE_CITY_SLUG must be delhi_ncr."
    )


remote_audit = r'''
set -e

ROOT="__DELHI_NCR_ROOT__"
DRIVER="$ROOT/batch/run_city_epochs.py"
CONFIG="$ROOT/city_config.json"
MANIFEST="$ROOT/delhi_ncr_workspace_manifest.json"

POPULATION_TEMPLATE="$ROOT/templates/population_epoch_template.py"
SMOD_TEMPLATE="$ROOT/templates/smod_epoch_template.py"
BUILT_TEMPLATE="$ROOT/templates/built_surface_epoch_template.py"

COMPILE_ROOT="/tmp/delhi_ncr_static_compile_check"


echo "============================================================"
echo "DELHI NCR STATIC AND LEAKAGE AUDIT"
echo "============================================================"


# ============================================================
# REQUIRED FILES
# ============================================================

for required_file in \
    "$DRIVER" \
    "$CONFIG" \
    "$MANIFEST" \
    "$POPULATION_TEMPLATE" \
    "$SMOD_TEMPLATE" \
    "$BUILT_TEMPLATE"

do
    if [ ! -f "$required_file" ]; then
        echo "ERROR: Required file is missing:"
        echo "$required_file"
        exit 20
    fi
done

echo "Required Delhi NCR files: PASS"


# ============================================================
# COMPILE OUTSIDE THE WORKSPACE
# ============================================================

rm -rf "$COMPILE_ROOT"

mkdir -p "$COMPILE_ROOT"

PYTHONPYCACHEPREFIX="$COMPILE_ROOT" \
python3 -m py_compile "$DRIVER"

PYTHONPYCACHEPREFIX="$COMPILE_ROOT" \
python3 -m py_compile "$POPULATION_TEMPLATE"

PYTHONPYCACHEPREFIX="$COMPILE_ROOT" \
python3 -m py_compile "$SMOD_TEMPLATE"

PYTHONPYCACHEPREFIX="$COMPILE_ROOT" \
python3 -m py_compile "$BUILT_TEMPLATE"

echo "Delhi NCR Python compilation: PASS"


# ============================================================
# STRUCTURAL PYTHON AUDIT
#
# All paths are command-line arguments. No shell or notebook
# interpolation occurs inside this Python source.
# ============================================================

python3 - \
    "$ROOT" \
    "$DRIVER" \
    "$CONFIG" \
    "$MANIFEST" \
    "$POPULATION_TEMPLATE" \
    "$SMOD_TEMPLATE" \
    "$BUILT_TEMPLATE" <<'PY'
from pathlib import Path
import ast
import hashlib
import json
import sys


root = Path(sys.argv[1])
driver_path = Path(sys.argv[2])
config_path = Path(sys.argv[3])
manifest_path = Path(sys.argv[4])

population_template = Path(sys.argv[5])
smod_template = Path(sys.argv[6])
built_template = Path(sys.argv[7])


def sha256_file(path):
    digest = hashlib.sha256()

    with path.open("rb") as handle:
        for block in iter(
            lambda: handle.read(1024 * 1024),
            b"",
        ):
            digest.update(block)

    return digest.hexdigest()


# ============================================================
# CONFIGURATION AUDIT
# ============================================================

config = json.loads(
    config_path.read_text(
        encoding="utf-8"
    )
)

expected_config = {
    "city_slug": "delhi_ncr",
    "city_name": "Delhi NCR",
    "remote_work_dir":
        "/home/ubuntu/ghsl_multicity/delhi_ncr",
    "hdfs_output_base":
        "hdfs://localhost:9000"
        "/data/ghsl/curated/cities/"
        "city=delhi_ncr",
    "expected_grid_cells": 4900,
    "expected_complete_cells": 4900,
    "expected_partial_cells": 0,
    "target_crs": "ESRI:54009",
}

for key, expected_value in expected_config.items():
    actual_value = config.get(key)

    if actual_value != expected_value:
        raise RuntimeError(
            "Configuration mismatch for "
            + key
            + ": expected "
            + repr(expected_value)
            + ", found "
            + repr(actual_value)
        )


expected_extent = {
    "min_x": 7103000.0,
    "min_y": 3446000.0,
    "max_x": 7173000.0,
    "max_y": 3516000.0,
}

actual_extent = config.get(
    "extent_mollweide"
)

if actual_extent != expected_extent:
    raise RuntimeError(
        "Delhi NCR extent mismatch. Expected "
        + repr(expected_extent)
        + ", found "
        + repr(actual_extent)
    )

if (
    expected_extent["max_x"]
    - expected_extent["min_x"]
    != 70000.0
):
    raise RuntimeError(
        "Delhi NCR X extent is not 70 km."
    )

if (
    expected_extent["max_y"]
    - expected_extent["min_y"]
    != 70000.0
):
    raise RuntimeError(
        "Delhi NCR Y extent is not 70 km."
    )

print("Delhi NCR configuration: PASS")


# ============================================================
# DRIVER STRUCTURE
# ============================================================

driver_source = driver_path.read_text(
    encoding="utf-8"
)

driver_tree = ast.parse(
    driver_source
)

function_names = set()

for node in driver_tree.body:
    if isinstance(
        node,
        ast.FunctionDef,
    ):
        function_names.add(
            node.name
        )


required_functions = {
    "generate_population_program",
    "generate_smod_program",
    "generate_built_program",
    "ensure_stage",
    "verify_template",
    "replace_exact_count",
}

missing_functions = sorted(
    required_functions
    - function_names
)

if missing_functions:
    raise RuntimeError(
        "Required driver functions are missing: "
        + repr(missing_functions)
    )

print("Required production functions: PASS")


# ============================================================
# TOP-LEVEL ASSIGNMENTS
# ============================================================

assignments = {}

for node in driver_tree.body:
    if not isinstance(
        node,
        ast.Assign,
    ):
        continue

    try:
        assignment_value = ast.literal_eval(
            node.value
        )
    except Exception:
        continue

    for target in node.targets:
        if isinstance(
            target,
            ast.Name,
        ):
            assignments[
                target.id
            ] = assignment_value


# Find a structural expected row/cell constant equal to 4900.

grid_expectations = {}

for name, value in assignments.items():
    uppercase_name = name.upper()

    if "EXPECTED" not in uppercase_name:
        continue

    if (
        "ROW" not in uppercase_name
        and "CELL" not in uppercase_name
    ):
        continue

    grid_expectations[name] = value


if 4900 not in grid_expectations.values():
    raise RuntimeError(
        "No expected row/cell constant equal "
        "to 4900 was found. Candidates: "
        + repr(grid_expectations)
    )

print(
    "Expected 4,900-cell constant: PASS",
    grid_expectations,
)


# ============================================================
# TEMPLATE CHECKSUM AUDIT
# ============================================================

driver_checksums = assignments.get(
    "EXPECTED_TEMPLATE_SHA256"
)

if not isinstance(
    driver_checksums,
    dict,
):
    raise RuntimeError(
        "EXPECTED_TEMPLATE_SHA256 is missing "
        "or is not a dictionary."
    )


actual_checksums = {
    "population": sha256_file(
        population_template
    ),
    "smod": sha256_file(
        smod_template
    ),
    "built_surface": sha256_file(
        built_template
    ),
}


if driver_checksums != actual_checksums:
    raise RuntimeError(
        "Driver template checksums do not match. "
        "Driver="
        + repr(driver_checksums)
        + ", actual="
        + repr(actual_checksums)
    )

print("Driver template checksums: PASS")


# ============================================================
# WORKSPACE MANIFEST AUDIT
# ============================================================

manifest = json.loads(
    manifest_path.read_text(
        encoding="utf-8"
    )
)

if (
    manifest.get("city_slug")
    != "delhi_ncr"
):
    raise RuntimeError(
        "Workspace manifest has the wrong slug."
    )

if (
    manifest.get("city_name")
    != "Delhi NCR"
):
    raise RuntimeError(
        "Workspace manifest has the wrong name."
    )

if (
    manifest.get("driver_sha256")
    != sha256_file(driver_path)
):
    raise RuntimeError(
        "Workspace driver checksum mismatch."
    )

if (
    manifest.get("template_sha256")
    != actual_checksums
):
    raise RuntimeError(
        "Workspace template checksum mismatch."
    )

print("Workspace manifest checksums: PASS")


# ============================================================
# CITY PATH AND LEAKAGE AUDIT
# ============================================================

combined_source = driver_source

combined_source += (
    population_template.read_text(
        encoding="utf-8"
    )
)

combined_source += (
    smod_template.read_text(
        encoding="utf-8"
    )
)

combined_source += (
    built_template.read_text(
        encoding="utf-8"
    )
)


required_delhi_ncr_values = [
    "/home/ubuntu/ghsl_multicity/delhi_ncr",
    "city=delhi_ncr",
    "7103000",
    "3446000",
    "7173000",
    "3516000",
]

missing_delhi_ncr_values = []

for value in required_delhi_ncr_values:
    if value not in combined_source:
        missing_delhi_ncr_values.append(
            value
        )

if missing_delhi_ncr_values:
    raise RuntimeError(
        "Required Delhi NCR values are missing: "
        + repr(missing_delhi_ncr_values)
    )


forbidden_hyderabad_values = [
    "/home/ubuntu/ghsl_multicity/hyderabad",
    "city=hyderabad",
    "Hyderabad Metropolitan Region",
]

remaining_hyderabad_values = []

for value in forbidden_hyderabad_values:
    if value in combined_source:
        remaining_hyderabad_values.append(
            value
        )

if remaining_hyderabad_values:
    raise RuntimeError(
        "Hyderabad leakage remains: "
        + repr(
            remaining_hyderabad_values
        )
    )

print(
    "Delhi NCR paths and extent in code: PASS"
)

print("Hyderabad leakage audit: PASS")


# ============================================================
# GENERALIZED TEMPLATE LOGIC
# ============================================================

population_source = (
    population_template.read_text(
        encoding="utf-8"
    )
)

smod_source = (
    smod_template.read_text(
        encoding="utf-8"
    )
)

built_source = (
    built_template.read_text(
        encoding="utf-8"
    )
)


if (
    ".parquet(population_1975_path)"
    in population_source
):
    raise RuntimeError(
        "Population template still contains "
        "the obsolete 1975 self-reference."
    )

if (
    ".parquet(SMOD_1975_PATH)"
    in smod_source
):
    raise RuntimeError(
        "SMOD template still contains "
        "the obsolete 1975 self-reference."
    )

if (
    "product=population/"
    not in built_source
):
    raise RuntimeError(
        "Built template does not contain "
        "the population-grid reference."
    )

print("Generalized template logic: PASS")


# ============================================================
# GENERALIZED 2020 GUARDS
# ============================================================

def is_epoch_not_2020(node):
    return (
        isinstance(
            node,
            ast.Compare,
        )
        and isinstance(
            node.left,
            ast.Name,
        )
        and node.left.id == "epoch"
        and len(node.ops) == 1
        and isinstance(
            node.ops[0],
            ast.NotEq,
        )
        and len(
            node.comparators
        ) == 1
        and isinstance(
            node.comparators[0],
            ast.Constant,
        )
        and node.comparators[0].value
        == 2020
    )


def is_active_2020_check(node):
    return (
        isinstance(
            node,
            ast.Compare,
        )
        and isinstance(
            node.left,
            ast.Constant,
        )
        and node.left.value
        == "EPOCH = 2020"
        and len(node.ops) == 1
        and isinstance(
            node.ops[0],
            ast.In,
        )
        and len(
            node.comparators
        ) == 1
        and isinstance(
            node.comparators[0],
            ast.Name,
        )
        and node.comparators[0].id
        == "program"
    )


guards_checked = 0

for node in ast.walk(
    driver_tree
):
    if not isinstance(
        node,
        ast.If,
    ):
        continue

    messages = []

    for child in ast.walk(node):
        if (
            isinstance(
                child,
                ast.Constant,
            )
            and isinstance(
                child.value,
                str,
            )
        ):
            messages.append(
                child.value
            )

    if not any(
        "retains active 2020"
        in message
        for message in messages
    ):
        continue

    if not (
        isinstance(
            node.test,
            ast.BoolOp,
        )
        and isinstance(
            node.test.op,
            ast.And,
        )
    ):
        raise RuntimeError(
            "A retained-active-2020 guard "
            "is not generalized."
        )

    has_epoch_condition = any(
        is_epoch_not_2020(
            condition
        )
        for condition
        in node.test.values
    )

    has_program_condition = any(
        is_active_2020_check(
            condition
        )
        for condition
        in node.test.values
    )

    if not (
        has_epoch_condition
        and has_program_condition
    ):
        raise RuntimeError(
            "A 2020 guard is missing one "
            "of its required conditions."
        )

    guards_checked += 1


if guards_checked == 0:
    raise RuntimeError(
        "No generalized 2020 guards were found."
    )

print(
    "Generalized 2020 guards: PASS",
    guards_checked,
)


print()
print("DELHI_NCR_STRUCTURAL_AUDIT_SUCCESS")
PY


# ============================================================
# LAUNCH-STATE AUDIT
# ============================================================

if [ -e "$ROOT/batch/city.pid" ]; then
    echo "ERROR: Delhi NCR city.pid already exists."
    exit 30
fi

if [ -e "$ROOT/batch/city.log" ]; then
    echo "ERROR: Delhi NCR city.log already exists."
    exit 31
fi

echo "No stale Delhi NCR launch state: PASS"


# ============================================================
# CLEAN TEMPORARY COMPILATION OUTPUT
# ============================================================

rm -rf "$COMPILE_ROOT"


echo
echo "DELHI_NCR_STATIC_AUDIT_SUCCESS"
'''


# Replace a literal placeholder instead of using an f-string.

cmd = remote_audit.replace(
    "__DELHI_NCR_ROOT__",
    root,
)


audit_result = run_remote(
    cmd,
    timeout=300,
    show_output=True,
)


print()
print("=" * 80)
print("CELL 5 RESULT")
print("=" * 80)

print(
    "Return code:",
    audit_result.returncode,
)

print("Spark launched: NO")
print("HDFS modified: NO")
print("Delhi NCR source files modified: NO")
print("CELL 5: SUCCESS")
```

```python
# ============================================================
# CELL 6: ALL-EPOCH GENERATOR SMOKE TEST
#
# Generates and compiles:
#   10 population programs
#   10 SMOD programs
#   10 built-surface programs
#
# Total: 30 generated programs
#
# This cell does not:
#   - launch Spark
#   - execute generated programs
#   - read raster data
#   - write to HDFS
#   - modify the Delhi NCR workspace
#
# Generated programs are written only to /tmp and deleted.
# ============================================================

root = ACTIVE_CITY["remote_work_dir"]

if ACTIVE_CITY_SLUG != "delhi_ncr":
    raise RuntimeError(
        "ACTIVE_CITY_SLUG must be delhi_ncr."
    )


remote_smoke_test = r'''
set -e

ROOT="__DELHI_NCR_ROOT__"
DRIVER="$ROOT/batch/run_city_epochs.py"

echo "============================================================"
echo "DELHI NCR ALL-EPOCH GENERATOR SMOKE TEST"
echo "============================================================"

if [ ! -f "$DRIVER" ]; then
    echo "ERROR: Delhi NCR driver is missing:"
    echo "$DRIVER"
    exit 20
fi


python3 - "$ROOT" "$DRIVER" <<'PY'
from pathlib import Path
import ast
import py_compile
import shutil
import sys
import tempfile


root = Path(sys.argv[1])
driver_path = Path(sys.argv[2])

driver_source = driver_path.read_text(
    encoding="utf-8"
)

driver_tree = ast.parse(
    driver_source
)


# ============================================================
# BUILD A DEFINITIONS-ONLY DRIVER MODULE
#
# Keep imports, constants, assignments, and functions.
# Exclude the top-level production loop so no Spark job starts.
# ============================================================

allowed_node_types = (
    ast.Import,
    ast.ImportFrom,
    ast.Assign,
    ast.AnnAssign,
    ast.FunctionDef,
    ast.AsyncFunctionDef,
    ast.ClassDef,
)

definition_nodes = [
    node
    for node in driver_tree.body
    if isinstance(
        node,
        allowed_node_types,
    )
]

definition_module = ast.Module(
    body=definition_nodes,
    type_ignores=[],
)

ast.fix_missing_locations(
    definition_module
)


namespace = {
    "__file__": str(driver_path),
    "__name__": "delhi_ncr_generator_smoke_test",
}

exec(
    compile(
        definition_module,
        str(driver_path),
        "exec",
    ),
    namespace,
)


# ============================================================
# REQUIRED FACTORIES
# ============================================================

required_factories = {
    "population":
        "generate_population_program",
    "smod":
        "generate_smod_program",
    "built_surface":
        "generate_built_program",
}

factories = {}

for stage, function_name in (
    required_factories.items()
):
    factory = namespace.get(
        function_name
    )

    if not callable(factory):
        raise RuntimeError(
            "Required factory is unavailable: "
            + function_name
        )

    factories[stage] = factory


# ============================================================
# TEST CONFIGURATION
# ============================================================

epochs = [
    1975,
    1980,
    1985,
    1990,
    1995,
    2000,
    2005,
    2010,
    2015,
    2020,
]

expected_program_count = (
    len(epochs)
    * len(factories)
)

temporary_root = Path(
    tempfile.mkdtemp(
        prefix="delhi_ncr_generator_smoke_"
    )
)

generated_records = []


# ============================================================
# GENERATE AND COMPILE ALL 30 PROGRAMS
# ============================================================

try:
    for epoch in epochs:
        epoch_directory = (
            temporary_root
            / ("epoch_" + str(epoch))
        )

        epoch_directory.mkdir()

        for stage, factory in (
            factories.items()
        ):
            generated_object = factory(
                epoch,
                epoch_directory,
            )

            # The production factories normally return a Path.
            # Support source text as a defensive fallback.
            if isinstance(
                generated_object,
                Path,
            ):
                program_path = generated_object

            elif (
                isinstance(
                    generated_object,
                    str,
                )
                and "\n"
                not in generated_object
            ):
                program_path = Path(
                    generated_object
                )

            elif isinstance(
                generated_object,
                str,
            ):
                program_path = (
                    epoch_directory
                    / (
                        stage
                        + "_"
                        + str(epoch)
                        + ".py"
                    )
                )

                program_path.write_text(
                    generated_object,
                    encoding="utf-8",
                )

            else:
                raise RuntimeError(
                    "Unexpected factory return type "
                    "for "
                    + stage
                    + " "
                    + str(epoch)
                    + ": "
                    + type(
                        generated_object
                    ).__name__
                )


            if not program_path.is_file():
                raise RuntimeError(
                    "Factory did not create a program "
                    "for "
                    + stage
                    + " "
                    + str(epoch)
                    + ": "
                    + str(program_path)
                )


            # Compile every generated program.
            py_compile.compile(
                str(program_path),
                doraise=True,
            )


            program_source = (
                program_path.read_text(
                    encoding="utf-8"
                )
            )


            # --------------------------------------------
            # Delhi NCR path validation
            # --------------------------------------------

            if (
                "city=delhi_ncr"
                not in program_source
            ):
                raise RuntimeError(
                    "Delhi NCR HDFS path is absent from "
                    + str(program_path)
                )

            if (
                "/home/ubuntu/ghsl_multicity/"
                "delhi_ncr"
                not in program_source
                and stage
                in {
                    "population",
                    "smod",
                    "built_surface",
                }
            ):
                # This is informational in most generated
                # programs, but reject only if neither Delhi NCR
                # workspace nor Delhi NCR HDFS identity appears.
                if (
                    "delhi_ncr"
                    not in program_source
                ):
                    raise RuntimeError(
                        "Delhi NCR identity is absent from "
                        + str(program_path)
                    )


            # --------------------------------------------
            # Hyderabad leakage validation
            # --------------------------------------------

            forbidden_values = [
                "city=hyderabad",
                "/home/ubuntu/ghsl_multicity/"
                "hyderabad",
                "Hyderabad Metropolitan Region",
            ]

            remaining_forbidden = [
                value
                for value
                in forbidden_values
                if value in program_source
            ]

            if remaining_forbidden:
                raise RuntimeError(
                    "Hyderabad leakage in "
                    + str(program_path)
                    + ": "
                    + repr(
                        remaining_forbidden
                    )
                )


            # --------------------------------------------
            # Active epoch validation
            # --------------------------------------------

            expected_epoch_assignment = (
                "EPOCH = "
                + str(epoch)
            )

            compact_program_source = (
                program_source.replace(
                    " ",
                    ""
                )
            )

            compact_expected_assignment = (
                expected_epoch_assignment.replace(
                    " ",
                    ""
                )
            )

            if (
                compact_expected_assignment
                not in compact_program_source
            ):
                raise RuntimeError(
                    "Active epoch assignment is "
                    "incorrect in "
                    + str(program_path)
                    + ". Expected "
                    + expected_epoch_assignment
                )


            # --------------------------------------------
            # Expected HDFS epoch path validation
            # --------------------------------------------

            expected_epoch_path = (
                "epoch="
                + str(epoch)
                + "/"
            )

            if (
                expected_epoch_path
                not in program_source
            ):
                raise RuntimeError(
                    "Expected epoch path is absent "
                    "from "
                    + str(program_path)
                    + ": "
                    + expected_epoch_path
                )


            generated_records.append(
                {
                    "epoch": epoch,
                    "stage": stage,
                    "path": str(
                        program_path
                    ),
                }
            )

            print(
                "PASS:",
                epoch,
                stage,
                program_path.name,
            )


    # ========================================================
    # FINAL COUNTS
    # ========================================================

    if (
        len(generated_records)
        != expected_program_count
    ):
        raise RuntimeError(
            "Expected "
            + str(
                expected_program_count
            )
            + " generated programs, found "
            + str(
                len(generated_records)
            )
        )


    stage_counts = {}

    for stage in factories:
        stage_counts[stage] = sum(
            1
            for record
            in generated_records
            if record["stage"]
            == stage
        )


    expected_stage_count = len(
        epochs
    )

    for stage, count in (
        stage_counts.items()
    ):
        if count != expected_stage_count:
            raise RuntimeError(
                "Expected "
                + str(
                    expected_stage_count
                )
                + " "
                + stage
                + " programs, found "
                + str(count)
            )


    print()

    print(
        "Population programs:",
        stage_counts["population"],
    )

    print(
        "SMOD programs:",
        stage_counts["smod"],
    )

    print(
        "Built-surface programs:",
        stage_counts[
            "built_surface"
        ],
    )

    print(
        "Total generated programs:",
        len(generated_records),
    )

    print()

    print(
        "ALL_30_GENERATED_PROGRAMS_COMPILE_PASS"
    )


finally:
    # No generated smoke-test artifacts persist.
    if temporary_root.exists():
        shutil.rmtree(
            temporary_root
        )


print(
    "Temporary smoke-test directory removed: PASS"
)

print(
    "ALL_EPOCH_GENERATOR_SMOKE_TEST_SUCCESS"
)
PY


echo
echo "Spark launched: NO"
echo "HDFS modified: NO"
echo "Delhi NCR workspace modified: NO"
echo "CELL_6_REMOTE_SUCCESS"
'''


# Replace a literal placeholder.
# Do not wrap the embedded Python in an outer f-string.

cmd = remote_smoke_test.replace(
    "__DELHI_NCR_ROOT__",
    root,
)


smoke_result = run_remote(
    cmd,
    timeout=300,
    show_output=True,
)


print()
print("=" * 80)
print("CELL 6 RESULT")
print("=" * 80)

print(
    "Return code:",
    smoke_result.returncode,
)

print(
    "Expected generated programs:",
    30,
)

print("Spark launched: NO")
print("HDFS modified: NO")
print("Delhi NCR workspace modified: NO")
print("CELL 6: SUCCESS")
```

```python
# ============================================================
# CELL 7: DETACHED DELHI NCR PRODUCTION LAUNCH
#
# Launches the full 1975-2020 Delhi NCR production pipeline.
#
# Safety:
#   - Refuses a duplicate active launch
#   - Loads Hadoop/Sedona environment
#   - Verifies HDFS daemons and safe mode
#   - Verifies Delhi NCR driver and template checksums
#   - Writes PID before returning
#   - Detects immediate startup failure
#   - Leaves the job detached on EC2
# ============================================================

root = ACTIVE_CITY["remote_work_dir"]

if ACTIVE_CITY_SLUG != "delhi_ncr":
    raise RuntimeError(
        "ACTIVE_CITY_SLUG must be delhi_ncr."
    )


remote_launch = r'''
set -e

source /home/ubuntu/ghsl-sedona-env.sh

export HADOOP_HOME=/usr/local/hadoop
export PATH="${HADOOP_HOME}/bin:${HADOOP_HOME}/sbin:${PATH}"

ROOT="__DELHI_NCR_ROOT__"
DRIVER="$ROOT/batch/run_city_epochs.py"
PID_FILE="$ROOT/batch/city.pid"
LOG_FILE="$ROOT/batch/city.log"


echo "============================================================"
echo "DELHI NCR PRODUCTION LAUNCH"
echo "============================================================"


# ------------------------------------------------------------
# Verify HDFS services.
# ------------------------------------------------------------

for daemon in \
    NameNode \
    DataNode \
    SecondaryNameNode

do
    jps \
        | awk '{print $2}' \
        | grep -qx "$daemon"
done

hdfs dfsadmin -safemode get \
    | grep -qi "Safe mode is OFF"

echo "HDFS services: PASS"


# ------------------------------------------------------------
# Verify Delhi NCR production files.
# ------------------------------------------------------------

if [ ! -f "$DRIVER" ]; then
    echo "ERROR: Delhi NCR driver is missing."
    exit 20
fi

python3 -m py_compile "$DRIVER"

for template in "$ROOT"/templates/*.py; do
    python3 -m py_compile "$template"
done

echo "Delhi NCR Python compilation: PASS"


# ------------------------------------------------------------
# Refuse duplicate launch.
# ------------------------------------------------------------

if [ -f "$PID_FILE" ]; then
    OLD_PID="$(cat "$PID_FILE")"

    if kill -0 "$OLD_PID" 2>/dev/null; then
        echo "ERROR: Delhi NCR job is already running."
        echo "Active PID: $OLD_PID"
        exit 30
    fi

    echo "Removing stale PID file:"
    echo "$PID_FILE"

    rm -f "$PID_FILE"
fi


# ------------------------------------------------------------
# Preserve an old log if one unexpectedly exists.
# ------------------------------------------------------------

if [ -f "$LOG_FILE" ]; then
    LOG_BACKUP="$LOG_FILE.before_launch.$(
        date -u +%Y%m%dT%H%M%SZ
    )"

    mv "$LOG_FILE" "$LOG_BACKUP"

    echo "Previous log moved to:"
    echo "$LOG_BACKUP"
fi


# ------------------------------------------------------------
# Launch detached production driver.
# ------------------------------------------------------------

nohup python3 "$DRIVER" \
    >"$LOG_FILE" \
    2>&1 \
    </dev/null &

NEW_PID="$!"

echo "$NEW_PID" > "$PID_FILE"

echo "New PID: $NEW_PID"
echo "Log: $LOG_FILE"


# ------------------------------------------------------------
# Allow startup checks to run.
# ------------------------------------------------------------

sleep 5


# ------------------------------------------------------------
# Confirm process survived startup.
# ------------------------------------------------------------

if ! kill -0 "$NEW_PID" 2>/dev/null; then
    echo
    echo "ERROR: Delhi NCR driver exited during startup."

    echo
    echo "Latest 100 log lines:"

    if [ -f "$LOG_FILE" ]; then
        tail -n 100 "$LOG_FILE"
    fi

    exit 31
fi


echo

ps -p "$NEW_PID" \
    -o pid,etime,%cpu,%mem,cmd

echo
echo "DETACHED_LAUNCH_SUCCESS"
'''


cmd = remote_launch.replace(
    "__DELHI_NCR_ROOT__",
    root,
)


launch_result = run_remote(
    cmd,
    timeout=90,
    show_output=True,
)


print()
print("=" * 80)
print("CELL 7 RESULT")
print("=" * 80)

print(
    "Return code:",
    launch_result.returncode,
)

print("Safe to close Colab.")
print("Monitor with Cell 8.")
print("CELL 7: SUCCESS")
```

```python
# ============================================================
# CELL 8: STABLE DELHI NCR MONITOR WITH CLEAR FINAL STATUS
#
# Read-only:
#   - Does not launch a job
#   - Does not stop a job
#   - Does not modify HDFS
# ============================================================

root = ACTIVE_CITY["remote_work_dir"]
city_name = ACTIVE_CITY["city_name"]

if ACTIVE_CITY_SLUG != "delhi_ncr":
    raise RuntimeError(
        "ACTIVE_CITY_SLUG must be delhi_ncr."
    )


remote_monitor = r'''
set +e

source /home/ubuntu/ghsl-sedona-env.sh 2>/dev/null

export HADOOP_HOME=/usr/local/hadoop
export PATH="$HADOOP_HOME/bin:$HADOOP_HOME/sbin:$PATH"

ROOT="__DELHI_NCR_ROOT__"
CITY_NAME="__CITY_NAME__"

PID_FILE="$ROOT/batch/city.pid"
LOG_FILE="$ROOT/batch/city.log"
MANIFEST="$ROOT/batch/batch_manifest.tsv"


get_job_status() {
    RECORDED_PID="N/A"
    JOB_STATUS="UNKNOWN"

    if [ -f "$PID_FILE" ]; then
        RECORDED_PID="$(cat "$PID_FILE")"

        if kill -0 "$RECORDED_PID" 2>/dev/null; then
            JOB_STATUS="RUNNING"
        else
            JOB_STATUS="STOPPED"
        fi
    fi
}


# ============================================================
# INITIAL STATUS
# ============================================================

get_job_status

echo "============================================================"
echo "DELHI NCR JOB STATUS"
echo "============================================================"

echo "City: $CITY_NAME"
echo "Recorded PID: $RECORDED_PID"
echo "STATUS=$JOB_STATUS"

if [ "$JOB_STATUS" = "RUNNING" ]; then
    ps -p "$RECORDED_PID" \
        -o pid,etime,%cpu,%mem,cmd
fi


# ============================================================
# LATEST LOG
# ============================================================

echo
echo "============================================================"
echo "LATEST 150 LOG LINES"
echo "============================================================"

if [ -f "$LOG_FILE" ]; then
    tail -n 150 "$LOG_FILE"
else
    echo "Log file is missing."
fi


# ============================================================
# CURRENT MANIFEST
# ============================================================

echo
echo "============================================================"
echo "CURRENT DELHI NCR MANIFEST"
echo "============================================================"

if [ -f "$MANIFEST" ]; then
    cat "$MANIFEST"
else
    echo "Manifest has not been created."
fi


# ============================================================
# SIMPLE PROGRESS COUNTS
# ============================================================

echo
echo "============================================================"
echo "CURRENT PROGRESS SUMMARY"
echo "============================================================"

COMPLETED_STAGES=0
COMPLETED_EPOCHS=0

if [ -f "$MANIFEST" ]; then
    COMPLETED_STAGES="$(
        grep -E \
            $'\t(built_and_validated|existing_valid_skipped)\t' \
            "$MANIFEST" \
            | wc -l \
            | tr -d ' '
    )"

    for epoch in \
        1975 1980 1985 1990 1995 \
        2000 2005 2010 2015 2020
    do
        EPOCH_RECORDS="$(
            awk -F '\t' \
                -v target_epoch="$epoch" \
                '$1 == target_epoch {print}' \
                "$MANIFEST" \
                | grep -E \
                    $'\t(built_and_validated|existing_valid_skipped)\t' \
                | wc -l \
                | tr -d ' '
        )"

        if [ "$EPOCH_RECORDS" = "4" ]; then
            COMPLETED_EPOCHS="$(
                expr "$COMPLETED_EPOCHS" + 1
            )"
        fi
    done
fi

echo "Completed stage records: $COMPLETED_STAGES / 40"
echo "Fully completed epochs: $COMPLETED_EPOCHS / 10"


# ============================================================
# ACTIVE JAVA PROCESSES
# ============================================================

echo
echo "============================================================"
echo "ACTIVE SPARK / JAVA PROCESSES"
echo "============================================================"

jps


# ============================================================
# FINAL STATUS
# ============================================================

get_job_status

COMPLETION_RESULT="UNKNOWN"
FINAL_MESSAGE="DELHI NCR JOB STATUS UNKNOWN"

if [ "$JOB_STATUS" = "RUNNING" ]; then
    COMPLETION_RESULT="IN_PROGRESS"
    FINAL_MESSAGE="DELHI NCR JOB IS STILL RUNNING"
fi

if [ "$JOB_STATUS" = "STOPPED" ]; then
    if [ -f "$LOG_FILE" ]; then
        if grep -q "BATCH PIPELINE: SUCCESS" "$LOG_FILE"; then
            COMPLETION_RESULT="SUCCESS"
            FINAL_MESSAGE="DELHI NCR JOB COMPLETED SUCCESSFULLY"
        elif grep -q "Traceback (most recent call last)" "$LOG_FILE"; then
            COMPLETION_RESULT="FAILED"
            FINAL_MESSAGE="DELHI NCR JOB STOPPED WITH FAILURE"
        else
            COMPLETION_RESULT="UNKNOWN"
            FINAL_MESSAGE="DELHI NCR JOB STOPPED, RESULT UNKNOWN"
        fi
    fi
fi


# ============================================================
# FINAL BLOCK
# ============================================================

echo
echo
echo "############################################################"
echo "############################################################"
echo
echo "              FINAL DELHI NCR PIPELINE STATUS"
echo
echo "City: $CITY_NAME"
echo "PID: $RECORDED_PID"
echo "FINAL_STATUS=$JOB_STATUS"
echo "COMPLETION_RESULT=$COMPLETION_RESULT"
echo "COMPLETED_STAGES=$COMPLETED_STAGES/40"
echo "COMPLETED_EPOCHS=$COMPLETED_EPOCHS/10"
echo
echo ">>> $FINAL_MESSAGE <<<"
echo

if [ "$JOB_STATUS" = "RUNNING" ]; then
    ps -p "$RECORDED_PID" \
        -o pid,etime,%cpu,%mem,cmd

    echo
    echo "Action: Let the EC2 job continue."
    echo "Do not rerun Cell 7."
fi

if [ "$COMPLETION_RESULT" = "SUCCESS" ]; then
    echo "Action: Run Cell 9 final audit."
fi

if [ "$COMPLETION_RESULT" = "FAILED" ]; then
    echo "Latest failure markers:"

    grep -n \
        -E \
        "Traceback|RuntimeError|AnalysisException|Return code: 1" \
        "$LOG_FILE" \
        | tail -n 20

    echo
    echo "Action: Do not rerun Cell 7."
fi

echo
echo "############################################################"
echo "############################################################"

exit 0
'''


cmd = (
    remote_monitor
    .replace(
        "__DELHI_NCR_ROOT__",
        root,
    )
    .replace(
        "__CITY_NAME__",
        city_name,
    )
)


monitor_result = run_remote(
    cmd,
    timeout=180,
    show_output=True,
)


print()
print("=" * 80)
print("CELL 8 RESULT")
print("=" * 80)

print(
    "Monitor return code:",
    monitor_result.returncode,
)

print("No new job was launched.")
print("No running process was stopped.")
print("No HDFS data was modified.")
```
