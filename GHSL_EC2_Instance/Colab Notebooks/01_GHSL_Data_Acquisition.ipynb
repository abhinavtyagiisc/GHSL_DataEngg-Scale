So the raster uses the World Mollweide projected coordinate system.

Why Mollweide?

Mollweide is an equal-area projection. That matters for this project because we will compare areas occupied by settlement classes across India and through time.

In an equal-area projection:

Equal-sized pixels represent equal ground area consistently across the map.

That makes area-based calculations more defensible than using latitude-longitude degrees directly.

The official GHSL source also lists Mollweide as a coordinate system for GHS-SMOD and identifies the product as a 1 km settlement-classification grid.

Before transferring the file into HDFS, we must confirm that the exported file preserved the source raster’s structure.

rasterio.open() opens the GeoTIFF as a raster dataset reader. Rasterio exposes the coordinate reference system through src.crs and the pixel-to-coordinate mapping through src.transform.

What each property tells us
src.driver: confirms the file format, normally GTiff.
src.count: number of raster bands, expected 1.
src.width, src.height: number of pixel columns and rows.
src.dtypes: storage datatype for the raster values.
src.crs: exported coordinate reference system.
src.transform: exported pixel size and grid origin.
src.nodata: value representing pixels without valid data.
src.bounds: projected geographic extent of the exported raster.

We preserve: ghsl_smod_india_2020.tif as the untouched raw export. We create: ghsl_smod_india_2020_clean.tif as the curated Spark input.

Raw layer → Original Earth Engine export

Curated layer → Validated file with explicit NoData metadata

```python
import rasterio

clean_tif_path = (
    "/content/drive/MyDrive/GHSL_India/"
    "ghsl_smod_india_2020_clean.tif"
)

with rasterio.open(tif_path) as src:
    profile = src.profile.copy()

    profile.update(
        nodata=0,
        compress="lzw"
    )

    with rasterio.open(
        clean_tif_path,
        "w",
        **profile
    ) as dst:
        dst.write(src.read(1), 1)

print("Created:", clean_tif_path)
```

```python
import numpy as np
import rasterio

with rasterio.open(clean_tif_path) as src:
    clean_data = src.read(1)
    clean_mask = src.read_masks(1)

    print("No-data value:", src.nodata)
    print("Unique stored values:", np.unique(clean_data))
    print("Unique mask values:", np.unique(clean_mask))

    print(
        "Invalid pixel count:",
        np.count_nonzero(clean_mask == 0)
    )

    print(
        "Valid pixel count:",
        np.count_nonzero(clean_mask == 255)
    )
```

Next step only: turn the checks into automated assertions Why write assertions?

Until now, we manually inspected outputs. An assertion converts an expected property into an executable data-quality rule.

assert src.count == 1

Stop execution if the exported file does not contain exactly one raster band.
assert src.nodata == 0
means-  Stop execution unless zero is formally defined as NoData.

```python
import numpy as np
import rasterio

expected_class_codes = {
    10, 11, 12, 13,
    21, 22, 23, 30
}

with rasterio.open(clean_tif_path) as src:
    data = src.read(1)
    mask = src.read_masks(1)

    stored_values = set(
        np.unique(data).tolist()
    )

    valid_values = set(
        np.unique(data[mask == 255]).tolist()
    )

    assert src.count == 1
    assert src.nodata == 0
    assert stored_values == expected_class_codes | {0}
    assert valid_values == expected_class_codes
    assert np.count_nonzero(mask == 0) == 6_844_740
    assert np.count_nonzero(mask == 255) == 3_178_596

print("All GeoTIFF validation checks passed.")
```

```python
# Inspect every band's native projection, affine transform,
# nominal resolution, and Earth Engine data type.

for dataset_name, image in ghsl_images.items():
    print("\n" + "=" * 85)
    print("Dataset:", dataset_name)
    print("=" * 85)

    band_names = image.bandNames().getInfo()

    for band_name in band_names:
        band = image.select(band_name)
        projection = band.projection()

        projection_info = projection.getInfo()
        nominal_scale = projection.nominalScale().getInfo()
        band_type = band.bandTypes().getInfo()

        print("\nBand:", band_name)
        print("CRS:", projection_info.get("crs"))
        print("Transform:", projection_info.get("transform"))
        print("Nominal scale:", nominal_scale, "metres")
        print("Data type:", band_type)
```

```python
# Verify the CRS, numeric affine transforms, and exact nesting
# of the 1 km SMOD grid within the shared 100 m GHSL grid.

grid_details = {}

for dataset_name, image in ghsl_images.items():
    print("\n" + "=" * 85)
    print("Dataset:", dataset_name)
    print("=" * 85)

    for band_name in image.bandNames().getInfo():
        band = image.select(band_name)
        projection = band.projection()

        # projection.getInfo() returns the numeric transform list
        projection_info = projection.getInfo()

        crs = projection.crs().getInfo()
        transform = projection_info["transform"]
        nominal_scale = projection.nominalScale().getInfo()

        grid_details[(dataset_name, band_name)] = {
            "crs": crs,
            "transform": transform,
            "nominal_scale": nominal_scale,
        }

        print("\nBand:", band_name)
        print("CRS reported by Earth Engine:", crs)
        print("Numeric transform:", transform)
        print("Nominal scale:", nominal_scale, "metres")
        print("Transform Python type:", type(transform).__name__)


# Extract the grid definitions
smod_grid = grid_details[
    ("SMOD 2020", "smod_code")
]

population_grid = grid_details[
    ("Population 2020", "population_count")
]

built_grid = grid_details[
    ("Built-up Surface 2020", "built_surface")
]

built_nres_grid = grid_details[
    ("Built-up Surface 2020", "built_surface_nres")
]

height_grid = grid_details[
    ("Building Height 2018", "built_height")
]


# Compare the CRS values of the three native 100 m products
same_100m_crs = (
    population_grid["crs"]
    == built_grid["crs"]
    == built_nres_grid["crs"]
    == height_grid["crs"]
)

# Compare their complete six-value affine transforms
same_100m_transform = (
    population_grid["transform"]
    == built_grid["transform"]
    == built_nres_grid["transform"]
    == height_grid["transform"]
)

# Compare the X and Y origins of SMOD and the native 100 m grid
same_origin_as_smod = (
    population_grid["transform"][2]
    == smod_grid["transform"][2]
    and
    population_grid["transform"][5]
    == smod_grid["transform"][5]
)

# Compare pixel sizes
pixel_ratio_x = (
    smod_grid["transform"][0]
    / population_grid["transform"][0]
)

pixel_ratio_y = (
    abs(smod_grid["transform"][4])
    / abs(population_grid["transform"][4])
)

# Check whether the 1 km origin aligns exactly with the 100 m grid
x_origin_offset = (
    smod_grid["transform"][2]
    - population_grid["transform"][2]
)

y_origin_offset = (
    smod_grid["transform"][5]
    - population_grid["transform"][5]
)


print("\n" + "=" * 85)
print("GRID ALIGNMENT SUMMARY")
print("=" * 85)

print("All 100 m products use the same CRS:", same_100m_crs)
print("All 100 m products use the same transform:", same_100m_transform)
print("SMOD and 100 m grids use the same origin:", same_origin_as_smod)

print("X-origin offset:", x_origin_offset, "metres")
print("Y-origin offset:", y_origin_offset, "metres")

print("SMOD-to-100 m pixel ratio in X:", pixel_ratio_x)
print("SMOD-to-100 m pixel ratio in Y:", pixel_ratio_y)


exact_nesting_confirmed = (
    same_100m_transform
    and same_origin_as_smod
    and pixel_ratio_x == 10
    and pixel_ratio_y == 10
)

print("\n" + "-" * 85)

if exact_nesting_confirmed:
    print("Result: EXACT NESTING CONFIRMED")
    print(
        "Population, built-up surface, and building height "
        "share one canonical 100 m grid."
    )
    print(
        "Each 1 km SMOD cell corresponds exactly to "
        "10 x 10 cells on the canonical 100 m grid."
    )
    print(
        "SMOD can therefore be mapped to the 100 m grid "
        "using nearest-neighbour categorical assignment."
    )
else:
    print("Result: GRID DIFFERENCES REQUIRE FURTHER INVESTIGATION")
```

```python
# -------------------------------------------------------------------
# CELL 6: PREPARE AND COMBINE THE GHSL ANALYTICAL BANDS
# -------------------------------------------------------------------

# Use the native Population 2020 grid as the canonical 100 m grid.
# Population, built-up surface, and building height were already
# confirmed to have exactly the same 100 m transform.

canonical_projection_100m = (
    ghsl_images["Population 2020"]
    .select("population_count")
    .projection()
)

canonical_projection_info = canonical_projection_100m.getInfo()
canonical_transform_100m = canonical_projection_info["transform"]

# The CRS string is not being returned through getInfo() in this
# Earth Engine environment, but the native projection object remains
# valid and can still be used for reductions and exports.
canonical_scale_100m = (
    canonical_projection_100m
    .nominalScale()
    .getInfo()
)


# -------------------------------------------------------------------
# 1. SMOD 2020
# -------------------------------------------------------------------
# Native source resolution: 1 km.
#
# Important:
# Nearest-neighbour is Earth Engine's default resampling method.
# Therefore, do NOT call .resample("nearest").
#
# We also preserve SMOD's actual native 1 km projection here.
# The common 100 m output grid will be specified during export or
# reduction rather than falsely changing SMOD's source projection.

smod_2020 = (
    ghsl_images["SMOD 2020"]
    .select("smod_code")
    .rename("smod_code")
)


# -------------------------------------------------------------------
# 2. POPULATION 2020
# -------------------------------------------------------------------
# Native resolution: 100 m.

population_2020_100m = (
    ghsl_images["Population 2020"]
    .select("population_count")
    .rename("population_count")
)


# -------------------------------------------------------------------
# 3. TOTAL BUILT-UP SURFACE 2020
# -------------------------------------------------------------------
# Native resolution: 100 m.

built_surface_2020_100m = (
    ghsl_images["Built-up Surface 2020"]
    .select("built_surface")
    .rename("built_surface")
)


# -------------------------------------------------------------------
# 4. NON-RESIDENTIAL BUILT-UP SURFACE 2020
# -------------------------------------------------------------------
# Native resolution: 100 m.

built_surface_nres_2020_100m = (
    ghsl_images["Built-up Surface 2020"]
    .select("built_surface_nres")
    .rename("built_surface_nres")
)


# -------------------------------------------------------------------
# 5. BUILDING HEIGHT 2018
# -------------------------------------------------------------------
# Native resolution: 100 m.
# This remains explicitly labelled as a 2018 cross-sectional variable.

built_height_2018_100m = (
    ghsl_images["Building Height 2018"]
    .select("built_height")
    .rename("built_height_2018")
)


# -------------------------------------------------------------------
# COMBINE THE FIVE BANDS
# -------------------------------------------------------------------
#
# The individual bands retain their native source projections:
#
# - smod_code: 1 km
# - all other bands: 100 m
#
# When this image is reduced or exported on the canonical 100 m grid,
# Earth Engine will assign the SMOD category using its default
# nearest-neighbour behavior.

combined_pilot_image = (
    ee.Image.cat([
        smod_2020,
        population_2020_100m,
        built_surface_2020_100m,
        built_surface_nres_2020_100m,
        built_height_2018_100m,
    ])
    .clip(bengaluru_pilot)
)


# -------------------------------------------------------------------
# VERIFY THE COMBINED IMAGE
# -------------------------------------------------------------------

print("=" * 85)
print("COMBINED BENGALURU PILOT IMAGE")
print("=" * 85)

combined_band_names = combined_pilot_image.bandNames().getInfo()

print("\nBands:")
print(combined_band_names)

print("\nBand data types:")
print(combined_pilot_image.bandTypes().getInfo())

print("\nCanonical 100 m transform:")
print(canonical_transform_100m)

print("\nCanonical nominal scale:")
print(canonical_scale_100m, "metres")

print("\nNative source resolution by band:")
print("- smod_code: 1000 m")
print("- population_count: 100 m")
print("- built_surface: 100 m")
print("- built_surface_nres: 100 m")
print("- built_height_2018: 100 m")

print("\nTemporal interpretation:")
print("- smod_code: 2020")
print("- population_count: 2020")
print("- built_surface: 2020")
print("- built_surface_nres: 2020")
print("- built_height_2018: 2018")

expected_bands = [
    "smod_code",
    "population_count",
    "built_surface",
    "built_surface_nres",
    "built_height_2018",
]

assert combined_band_names == expected_bands, (
    "Unexpected combined band structure: "
    f"{combined_band_names}"
)

print("\nValidation:")
print("All five expected analytical bands are present.")

print("\nImportant:")
print(
    "SMOD retains its true 1 km source resolution."
)
print(
    "Nearest-neighbour assignment to the 100 m analytical grid "
    "will be applied during reduction/export."
)
print(
    "Building height remains a separate 2018 "
    "cross-sectional enrichment variable."
)
```

```python
# -------------------------------------------------------------------
# CELL 7: CALCULATE BENGALURU PILOT STATISTICS
# -------------------------------------------------------------------

print("=" * 85)
print("CALCULATING BENGALURU PILOT STATISTICS")
print("=" * 85)
print("This operation may take a short while.\n")


# -------------------------------------------------------------------
# 1. SMOD CLASS DISTRIBUTION, 2020
# -------------------------------------------------------------------
# Use the native 1 km scale because SMOD is genuinely a 1 km product.

smod_histogram = (
    smod_2020
    .reduceRegion(
        reducer=ee.Reducer.frequencyHistogram(),
        geometry=bengaluru_pilot,
        scale=1000,
        maxPixels=10_000_000,
        tileScale=4
    )
    .get("smod_code")
    .getInfo()
)


# Official SMOD class names
smod_class_names = {
    10: "Water",
    11: "Very low density rural",
    12: "Low density rural",
    13: "Rural cluster",
    21: "Suburban or peri-urban",
    22: "Semi-dense urban cluster",
    23: "Dense urban cluster",
    30: "Urban centre",
}


# -------------------------------------------------------------------
# 2. POPULATION TOTAL, 2020
# -------------------------------------------------------------------
# population_count represents population associated with each source cell.
# Use the native 100 m grid.

population_statistics = (
    population_2020_100m
    .reduceRegion(
        reducer=(
            ee.Reducer.sum()
            .combine(
                reducer2=ee.Reducer.minMax(),
                sharedInputs=True
            )
            .combine(
                reducer2=ee.Reducer.count(),
                sharedInputs=True
            )
        ),
        geometry=bengaluru_pilot,
        scale=100,
        maxPixels=20_000_000,
        tileScale=4
    )
    .getInfo()
)


# -------------------------------------------------------------------
# 3. BUILT-UP SURFACE TOTALS, 2020
# -------------------------------------------------------------------
# The exact unit interpretation will continue to be preserved from the
# official metadata. The raw totals are calculated here without changing
# the source values.

built_statistics = (
    ee.Image.cat([
        built_surface_2020_100m,
        built_surface_nres_2020_100m
    ])
    .reduceRegion(
        reducer=(
            ee.Reducer.sum()
            .combine(
                reducer2=ee.Reducer.minMax(),
                sharedInputs=True
            )
            .combine(
                reducer2=ee.Reducer.count(),
                sharedInputs=True
            )
        ),
        geometry=bengaluru_pilot,
        scale=100,
        maxPixels=20_000_000,
        tileScale=4
    )
    .getInfo()
)


# -------------------------------------------------------------------
# 4. BUILDING-HEIGHT STATISTICS, 2018
# -------------------------------------------------------------------
# Building height is handled separately because it belongs only to 2018.
# Summing height values would not be meaningful, so calculate descriptive
# statistics instead.

height_statistics = (
    built_height_2018_100m
    .reduceRegion(
        reducer=(
            ee.Reducer.mean()
            .combine(
                reducer2=ee.Reducer.median(),
                sharedInputs=True
            )
            .combine(
                reducer2=ee.Reducer.minMax(),
                sharedInputs=True
            )
            .combine(
                reducer2=ee.Reducer.count(),
                sharedInputs=True
            )
        ),
        geometry=bengaluru_pilot,
        scale=100,
        maxPixels=20_000_000,
        tileScale=4
    )
    .getInfo()
)


# -------------------------------------------------------------------
# PRINT RESULTS
# -------------------------------------------------------------------

print("\n" + "=" * 85)
print("1. SMOD CLASS DISTRIBUTION, 2020")
print("=" * 85)

if smod_histogram:
    total_smod_cells = sum(
        float(count)
        for count in smod_histogram.values()
    )

    for raw_code, raw_count in sorted(
        smod_histogram.items(),
        key=lambda item: float(item[0])
    ):
        code = int(float(raw_code))
        count = float(raw_count)

        percentage = (
            100.0 * count / total_smod_cells
            if total_smod_cells > 0
            else 0.0
        )

        class_name = smod_class_names.get(
            code,
            "Unknown or invalid class"
        )

        print(
            f"{code:>2} | "
            f"{class_name:<30} | "
            f"{count:>10,.0f} cells | "
            f"{percentage:>6.2f}%"
        )

    print("\nTotal valid SMOD cells:", f"{total_smod_cells:,.0f}")

else:
    print("No valid SMOD values were returned.")


print("\n" + "=" * 85)
print("2. POPULATION STATISTICS, 2020")
print("=" * 85)

for key, value in population_statistics.items():
    print(f"{key}: {value}")


print("\n" + "=" * 85)
print("3. BUILT-UP-SURFACE STATISTICS, 2020")
print("=" * 85)

for key, value in built_statistics.items():
    print(f"{key}: {value}")


print("\n" + "=" * 85)
print("4. BUILDING-HEIGHT STATISTICS, 2018")
print("=" * 85)

for key, value in height_statistics.items():
    print(f"{key}: {value}")


# -------------------------------------------------------------------
# BASIC VALIDATION
# -------------------------------------------------------------------

assert population_statistics, (
    "No population statistics were returned."
)

assert built_statistics, (
    "No built-up-surface statistics were returned."
)

assert height_statistics, (
    "No building-height statistics were returned."
)

print("\n" + "=" * 85)
print("VALIDATION RESULT")
print("=" * 85)
print("All four GHSL products returned data for the Bengaluru pilot.")
```

```python
# -------------------------------------------------------------------
# CELL 8: MEANINGFUL BUILT-UP AND BUILDING-HEIGHT STATISTICS
# -------------------------------------------------------------------

# The previous height mean included zero-height cells across the whole
# pilot region. Create masks so height statistics describe built cells.

# Cells containing some built-up surface in 2020
built_cell_mask_2020 = built_surface_2020_100m.gt(0)

# Cells containing a positive building-height estimate in 2018
positive_height_mask_2018 = built_height_2018_100m.gt(0)

# Height values only where height is positive
positive_height_2018 = (
    built_height_2018_100m
    .updateMask(positive_height_mask_2018)
)

# Height values where:
# 1. Height is positive in 2018
# 2. Built-up surface is positive in 2020
#
# Important:
# This is a cross-year contextual comparison, not a pure 2018 match.
height_on_built_cells = (
    built_height_2018_100m
    .updateMask(
        positive_height_mask_2018
        .And(built_cell_mask_2020)
    )
)


# -------------------------------------------------------------------
# COUNT TOTAL, BUILT, AND POSITIVE-HEIGHT CELLS
# -------------------------------------------------------------------

cell_count_image = ee.Image.constant(1).rename("cell")

total_100m_cell_count = (
    cell_count_image
    .reduceRegion(
        reducer=ee.Reducer.count(),
        geometry=bengaluru_pilot,
        scale=100,
        maxPixels=20_000_000,
        tileScale=4
    )
    .get("cell")
    .getInfo()
)

built_cell_count = (
    cell_count_image
    .updateMask(built_cell_mask_2020)
    .reduceRegion(
        reducer=ee.Reducer.count(),
        geometry=bengaluru_pilot,
        scale=100,
        maxPixels=20_000_000,
        tileScale=4
    )
    .get("cell")
    .getInfo()
)

positive_height_cell_count = (
    cell_count_image
    .updateMask(positive_height_mask_2018)
    .reduceRegion(
        reducer=ee.Reducer.count(),
        geometry=bengaluru_pilot,
        scale=100,
        maxPixels=20_000_000,
        tileScale=4
    )
    .get("cell")
    .getInfo()
)

height_on_built_cell_count = (
    cell_count_image
    .updateMask(
        positive_height_mask_2018
        .And(built_cell_mask_2020)
    )
    .reduceRegion(
        reducer=ee.Reducer.count(),
        geometry=bengaluru_pilot,
        scale=100,
        maxPixels=20_000_000,
        tileScale=4
    )
    .get("cell")
    .getInfo()
)


# -------------------------------------------------------------------
# POSITIVE BUILDING-HEIGHT DISTRIBUTION
# -------------------------------------------------------------------

positive_height_statistics = (
    positive_height_2018
    .reduceRegion(
        reducer=(
            ee.Reducer.mean()
            .combine(
                reducer2=ee.Reducer.median(),
                sharedInputs=True
            )
            .combine(
                reducer2=ee.Reducer.minMax(),
                sharedInputs=True
            )
            .combine(
                reducer2=ee.Reducer.percentile(
                    [25, 75, 90, 95]
                ),
                sharedInputs=True
            )
            .combine(
                reducer2=ee.Reducer.count(),
                sharedInputs=True
            )
        ),
        geometry=bengaluru_pilot,
        scale=100,
        maxPixels=20_000_000,
        tileScale=4
    )
    .getInfo()
)


# -------------------------------------------------------------------
# HEIGHT DISTRIBUTION ON CELLS ALSO BUILT IN 2020
# -------------------------------------------------------------------

height_on_built_statistics = (
    height_on_built_cells
    .reduceRegion(
        reducer=(
            ee.Reducer.mean()
            .combine(
                reducer2=ee.Reducer.median(),
                sharedInputs=True
            )
            .combine(
                reducer2=ee.Reducer.percentile(
                    [25, 75, 90, 95]
                ),
                sharedInputs=True
            )
            .combine(
                reducer2=ee.Reducer.count(),
                sharedInputs=True
            )
        ),
        geometry=bengaluru_pilot,
        scale=100,
        maxPixels=20_000_000,
        tileScale=4
    )
    .getInfo()
)


# -------------------------------------------------------------------
# DERIVED SUMMARY VALUES
# -------------------------------------------------------------------

population_total = population_statistics[
    "population_count_sum"
]

built_surface_total_m2 = built_statistics[
    "built_surface_sum"
]

built_surface_nres_total_m2 = built_statistics[
    "built_surface_nres_sum"
]

built_surface_total_km2 = (
    built_surface_total_m2 / 1_000_000
)

built_surface_nres_total_km2 = (
    built_surface_nres_total_m2 / 1_000_000
)

nonresidential_share_percent = (
    100.0
    * built_surface_nres_total_m2
    / built_surface_total_m2
    if built_surface_total_m2 > 0
    else 0.0
)

built_cell_share_percent = (
    100.0
    * built_cell_count
    / total_100m_cell_count
    if total_100m_cell_count > 0
    else 0.0
)

positive_height_share_percent = (
    100.0
    * positive_height_cell_count
    / total_100m_cell_count
    if total_100m_cell_count > 0
    else 0.0
)

height_coverage_of_built_cells_percent = (
    100.0
    * height_on_built_cell_count
    / built_cell_count
    if built_cell_count > 0
    else 0.0
)


# -------------------------------------------------------------------
# PRINT RESULTS
# -------------------------------------------------------------------

print("=" * 85)
print("REFINED BENGALURU PILOT SUMMARY")
print("=" * 85)

print("\n2020 population:")
print(f"{population_total:,.0f} people")

print("\n2020 total built-up surface:")
print(f"{built_surface_total_m2:,.2f} square metres")
print(f"{built_surface_total_km2:,.2f} square kilometres")

print("\n2020 non-residential built-up surface:")
print(f"{built_surface_nres_total_m2:,.2f} square metres")
print(
    f"{built_surface_nres_total_km2:,.2f} "
    "square kilometres"
)

print("\nNon-residential share of built-up surface:")
print(f"{nonresidential_share_percent:.2f}%")

print("\n100 m cell coverage:")
print(f"Total cells: {total_100m_cell_count:,.0f}")
print(f"Cells with built surface > 0: {built_cell_count:,.0f}")
print(f"Built-cell share: {built_cell_share_percent:.2f}%")

print(
    "Cells with positive 2018 height: "
    f"{positive_height_cell_count:,.0f}"
)
print(
    "Positive-height share of all cells: "
    f"{positive_height_share_percent:.2f}%"
)

print(
    "Cells with positive 2018 height and positive "
    f"2020 built surface: {height_on_built_cell_count:,.0f}"
)

print(
    "2018 height coverage among cells built in 2020: "
    f"{height_coverage_of_built_cells_percent:.2f}%"
)


print("\n" + "=" * 85)
print("POSITIVE BUILDING HEIGHTS, 2018")
print("=" * 85)

for key, value in positive_height_statistics.items():
    print(f"{key}: {value}")


print("\n" + "=" * 85)
print("2018 HEIGHT WHERE 2020 BUILT SURFACE IS POSITIVE")
print("=" * 85)

for key, value in height_on_built_statistics.items():
    print(f"{key}: {value}")


# -------------------------------------------------------------------
# VALIDATION
# -------------------------------------------------------------------

assert total_100m_cell_count > 0
assert built_cell_count > 0
assert positive_height_cell_count > 0

assert built_surface_nres_total_m2 <= built_surface_total_m2, (
    "Non-residential built-up surface exceeds total built-up surface."
)

print("\n" + "=" * 85)
print("VALIDATION RESULT")
print("=" * 85)

print("Built-up and positive-height statistics were calculated successfully.")
print(
    "The earlier all-cell building-height mean should not be used "
    "for urban-form interpretation."
)
print(
    "Use the positive-height distribution and report its coverage "
    "instead."
)
```

```python
# -------------------------------------------------------------------
# CELL 9: CORRECT GRID-ALIGNED COVERAGE COUNTS
# -------------------------------------------------------------------
#
# Important:
# Do not pass "ESRI:54009" to Earth Engine because the server does not
# parse that identifier in this operation.
#
# Reuse the native Population 2020 projection object, which represents
# the canonical GHSL 100 m grid already confirmed in Cells 3 and 4.

canonical_projection_100m = (
    population_2020_100m
    .select("population_count")
    .projection()
)


def count_valid_pixels(image, band_name):
    """
    Count unmasked pixels using the native canonical GHSL 100 m
    projection.

    The input image must already be aligned with, or evaluated on,
    the shared GHSL grid.
    """

    result = (
        image
        .select(band_name)
        .reduceRegion(
            reducer=ee.Reducer.count(),
            geometry=bengaluru_pilot,
            crs=canonical_projection_100m,
            scale=100,
            maxPixels=20_000_000,
            tileScale=4,
        )
        .get(band_name)
        .getInfo()
    )

    return int(result or 0)


# -------------------------------------------------------------------
# VALID SOURCE-PIXEL COUNTS
# -------------------------------------------------------------------

population_valid_count = count_valid_pixels(
    population_2020_100m,
    "population_count",
)

built_valid_count = count_valid_pixels(
    built_surface_2020_100m,
    "built_surface",
)

height_valid_count = count_valid_pixels(
    built_height_2018_100m,
    "built_height_2018",
)


# -------------------------------------------------------------------
# POSITIVE-VALUE MASKS
# -------------------------------------------------------------------

built_positive_image = (
    built_surface_2020_100m
    .gt(0)
    .selfMask()
    .rename("built_positive")
)

height_positive_image = (
    built_height_2018_100m
    .gt(0)
    .selfMask()
    .rename("height_positive")
)


built_positive_count = count_valid_pixels(
    built_positive_image,
    "built_positive",
)

height_positive_count = count_valid_pixels(
    height_positive_image,
    "height_positive",
)


# -------------------------------------------------------------------
# CROSS-YEAR INTERSECTION
# -------------------------------------------------------------------
#
# This compares:
# - positive built-up surface in 2020
# - positive building height in 2018
#
# Therefore, it is contextual cross-year coverage, not a direct
# same-year comparison.

height_and_built_positive_image = (
    built_surface_2020_100m
    .gt(0)
    .And(
        built_height_2018_100m.gt(0)
    )
    .selfMask()
    .rename("height_and_built_positive")
)

height_and_built_positive_count = count_valid_pixels(
    height_and_built_positive_image,
    "height_and_built_positive",
)


# Built in 2020 but without positive building height in 2018
built_without_positive_height_image = (
    built_surface_2020_100m
    .gt(0)
    .And(
        built_height_2018_100m.lte(0)
    )
    .selfMask()
    .rename("built_without_positive_height")
)

built_without_positive_height_count = count_valid_pixels(
    built_without_positive_height_image,
    "built_without_positive_height",
)


# Positive height in 2018 but without positive built surface in 2020
height_without_positive_built_image = (
    built_height_2018_100m
    .gt(0)
    .And(
        built_surface_2020_100m.lte(0)
    )
    .selfMask()
    .rename("height_without_positive_built")
)

height_without_positive_built_count = count_valid_pixels(
    height_without_positive_built_image,
    "height_without_positive_built",
)


# -------------------------------------------------------------------
# COVERAGE PERCENTAGES
# -------------------------------------------------------------------

built_positive_share_of_valid = (
    100.0 * built_positive_count / built_valid_count
    if built_valid_count > 0
    else 0.0
)

height_positive_share_of_valid = (
    100.0 * height_positive_count / height_valid_count
    if height_valid_count > 0
    else 0.0
)

height_coverage_of_built_cells = (
    100.0
    * height_and_built_positive_count
    / built_positive_count
    if built_positive_count > 0
    else 0.0
)

built_overlap_of_height_cells = (
    100.0
    * height_and_built_positive_count
    / height_positive_count
    if height_positive_count > 0
    else 0.0
)


# -------------------------------------------------------------------
# INTERNAL CONSISTENCY CHECKS
# -------------------------------------------------------------------

built_partition_total = (
    height_and_built_positive_count
    + built_without_positive_height_count
)

height_partition_total = (
    height_and_built_positive_count
    + height_without_positive_built_count
)


# -------------------------------------------------------------------
# PRINT RESULTS
# -------------------------------------------------------------------

print("=" * 85)
print("GRID-ALIGNED 100 M COVERAGE COUNTS")
print("=" * 85)

print("\nCanonical grid:")
print("Native Population 2020 GHSL projection")
print("Nominal scale: 100 metres")
print(
    "Transform:",
    canonical_projection_100m.getInfo()["transform"]
)

print("\nValid source pixels:")
print(
    f"Population valid pixels:      "
    f"{population_valid_count:,.0f}"
)
print(
    f"Built-surface valid pixels:   "
    f"{built_valid_count:,.0f}"
)
print(
    f"Building-height valid pixels: "
    f"{height_valid_count:,.0f}"
)

print("\nPositive-value pixels:")
print(
    f"Built surface > 0 in 2020:   "
    f"{built_positive_count:,.0f}"
)
print(
    f"Building height > 0 in 2018: "
    f"{height_positive_count:,.0f}"
)

print("\nCross-year intersection:")
print(
    "Positive built + positive height:    "
    f"{height_and_built_positive_count:,.0f}"
)
print(
    "Positive built, no positive height: "
    f"{built_without_positive_height_count:,.0f}"
)
print(
    "Positive height, no positive built: "
    f"{height_without_positive_built_count:,.0f}"
)

print("\nCoverage percentages:")
print(
    "Positive built-cell share of valid built pixels: "
    f"{built_positive_share_of_valid:.2f}%"
)
print(
    "Positive height-cell share of valid height pixels: "
    f"{height_positive_share_of_valid:.2f}%"
)
print(
    "Positive 2018 height coverage among positive "
    "2020 built cells: "
    f"{height_coverage_of_built_cells:.2f}%"
)
print(
    "Positive 2020 built overlap among positive "
    "2018 height cells: "
    f"{built_overlap_of_height_cells:.2f}%"
)

print("\nPartition checks:")
print(
    "Positive built count:  ",
    f"{built_positive_count:,.0f}",
)
print(
    "Built partition total: ",
    f"{built_partition_total:,.0f}",
)
print(
    "Positive height count: ",
    f"{height_positive_count:,.0f}",
)
print(
    "Height partition total:",
    f"{height_partition_total:,.0f}",
)


# -------------------------------------------------------------------
# VALIDATION
# -------------------------------------------------------------------

assert population_valid_count > 0
assert built_valid_count > 0
assert height_valid_count > 0

assert built_positive_count <= built_valid_count
assert height_positive_count <= height_valid_count

assert built_partition_total == built_positive_count, (
    "Built-cell partition does not equal the positive "
    "built-cell count."
)

assert height_partition_total == height_positive_count, (
    "Height-cell partition does not equal the positive "
    "height-cell count."
)

print("\n" + "=" * 85)
print("VALIDATION RESULT")
print("=" * 85)

print(
    "All counts were evaluated using the native "
    "canonical GHSL 100 m projection."
)
print(
    "The built and height intersection partitions "
    "are internally consistent."
)
```

```python
# -------------------------------------------------------------------
# CELL 12: VERIFY 2015 AND 2020 GRID COMPATIBILITY
# -------------------------------------------------------------------

# Select and rename the required 2015 bands.
smod_2015 = (
    ghsl_images_2015["SMOD 2015"]
    .select("smod_code")
    .rename("smod_code")
)

population_2015_100m = (
    ghsl_images_2015["Population 2015"]
    .select("population_count")
    .rename("population_count")
)

built_surface_2015_100m = (
    ghsl_images_2015["Built-up Surface 2015"]
    .select("built_surface")
    .rename("built_surface")
)

built_surface_nres_2015_100m = (
    ghsl_images_2015["Built-up Surface 2015"]
    .select("built_surface_nres")
    .rename("built_surface_nres")
)


def get_grid_information(image, band_name):
    """
    Return the native projection information for one image band.
    """

    band = image.select(band_name)
    projection = band.projection()
    projection_info = projection.getInfo()

    return {
        "transform": projection_info["transform"],
        "nominal_scale": projection.nominalScale().getInfo(),
        "data_type": band.bandTypes().getInfo(),
    }


# Read the 2015 grid definitions.
grid_2015 = {
    "smod": get_grid_information(
        smod_2015,
        "smod_code",
    ),
    "population": get_grid_information(
        population_2015_100m,
        "population_count",
    ),
    "built_surface": get_grid_information(
        built_surface_2015_100m,
        "built_surface",
    ),
    "built_surface_nres": get_grid_information(
        built_surface_nres_2015_100m,
        "built_surface_nres",
    ),
}


# Read the 2020 grid definitions.
grid_2020 = {
    "smod": get_grid_information(
        smod_2020,
        "smod_code",
    ),
    "population": get_grid_information(
        population_2020_100m,
        "population_count",
    ),
    "built_surface": get_grid_information(
        built_surface_2020_100m,
        "built_surface",
    ),
    "built_surface_nres": get_grid_information(
        built_surface_nres_2020_100m,
        "built_surface_nres",
    ),
}


# -------------------------------------------------------------------
# COMPARE 2015 AND 2020 FOR EACH PRODUCT
# -------------------------------------------------------------------

print("=" * 85)
print("2015-TO-2020 GRID COMPATIBILITY")
print("=" * 85)

all_grid_checks_passed = True

for product_name in [
    "smod",
    "population",
    "built_surface",
    "built_surface_nres",
]:
    info_2015 = grid_2015[product_name]
    info_2020 = grid_2020[product_name]

    same_transform = (
        info_2015["transform"]
        == info_2020["transform"]
    )

    same_scale = (
        info_2015["nominal_scale"]
        == info_2020["nominal_scale"]
    )

    product_passed = (
        same_transform
        and same_scale
    )

    all_grid_checks_passed = (
        all_grid_checks_passed
        and product_passed
    )

    print("\n" + "-" * 85)
    print("Product:", product_name)

    print(
        "2015 transform:",
        info_2015["transform"],
    )

    print(
        "2020 transform:",
        info_2020["transform"],
    )

    print(
        "2015 nominal scale:",
        info_2015["nominal_scale"],
        "metres",
    )

    print(
        "2020 nominal scale:",
        info_2020["nominal_scale"],
        "metres",
    )

    print(
        "Same transform:",
        same_transform,
    )

    print(
        "Same nominal scale:",
        same_scale,
    )

    print(
        "Result:",
        "PASS" if product_passed else "FAIL",
    )


# -------------------------------------------------------------------
# VERIFY CROSS-PRODUCT ALIGNMENT WITHIN 2015
# -------------------------------------------------------------------

smod_2015_transform = (
    grid_2015["smod"]["transform"]
)

population_2015_transform = (
    grid_2015["population"]["transform"]
)

built_2015_transform = (
    grid_2015["built_surface"]["transform"]
)

built_nres_2015_transform = (
    grid_2015["built_surface_nres"]["transform"]
)


same_2015_100m_grid = (
    population_2015_transform
    == built_2015_transform
    == built_nres_2015_transform
)

same_2015_origin = (
    smod_2015_transform[2]
    == population_2015_transform[2]
    and
    smod_2015_transform[5]
    == population_2015_transform[5]
)

smod_to_100m_ratio_x = (
    smod_2015_transform[0]
    / population_2015_transform[0]
)

smod_to_100m_ratio_y = (
    abs(smod_2015_transform[4])
    / abs(population_2015_transform[4])
)


print("\n" + "=" * 85)
print("2015 CROSS-PRODUCT ALIGNMENT")
print("=" * 85)

print(
    "Population and both built-surface bands "
    "share one 100 m grid:",
    same_2015_100m_grid,
)

print(
    "SMOD and the 100 m grid use the same origin:",
    same_2015_origin,
)

print(
    "SMOD-to-100 m pixel ratio in X:",
    smod_to_100m_ratio_x,
)

print(
    "SMOD-to-100 m pixel ratio in Y:",
    smod_to_100m_ratio_y,
)


exact_2015_nesting = (
    same_2015_100m_grid
    and same_2015_origin
    and smod_to_100m_ratio_x == 10
    and smod_to_100m_ratio_y == 10
)


# -------------------------------------------------------------------
# FINAL VALIDATION
# -------------------------------------------------------------------

assert all_grid_checks_passed, (
    "At least one product changed grid between 2015 and 2020."
)

assert exact_2015_nesting, (
    "The 2015 SMOD and 100 m products are not exactly nested."
)


print("\n" + "=" * 85)
print("FINAL VALIDATION RESULT")
print("=" * 85)

print(
    "All 2015 products use the same grids as their "
    "2020 counterparts."
)

print(
    "The 2015 population and built-surface bands "
    "share the canonical 100 m grid."
)

print(
    "The 2015 SMOD grid is exactly nested as "
    "10 x 10 cells over the canonical 100 m grid."
)

print(
    "The 2015-to-2020 comparison can therefore be "
    "performed without shifting the source grids."
)
```

```python
# -------------------------------------------------------------------
# CELL 15C: CONSOLIDATE THE SMOD 2020 EXPORT SAFELY
# -------------------------------------------------------------------
#
# The completed SMOD 2020 export was written to:
# GHSL_Bengaluru_Pilot (1)
#
# This cell copies it into the main pilot folder.
# It does not delete or modify the original file.

from pathlib import Path
import shutil
import hashlib


source_smod_2020 = Path(
    "/content/drive/MyDrive/"
    "GHSL_Bengaluru_Pilot (1)/"
    "ghsl_bengaluru_smod_2020_1km.tif"
)

destination_folder = Path(
    "/content/drive/MyDrive/"
    "GHSL_Bengaluru_Pilot"
)

destination_smod_2020 = (
    destination_folder
    / "ghsl_bengaluru_smod_2020_1km.tif"
)


def calculate_sha256(file_path, chunk_size=1024 * 1024):
    """
    Calculate the SHA-256 checksum of a file.
    """

    sha256 = hashlib.sha256()

    with file_path.open("rb") as file_handle:
        while True:
            chunk = file_handle.read(chunk_size)

            if not chunk:
                break

            sha256.update(chunk)

    return sha256.hexdigest()


print("=" * 85)
print("CONSOLIDATING SMOD 2020 EXPORT")
print("=" * 85)

print("\nSource:")
print(source_smod_2020)

print("\nDestination:")
print(destination_smod_2020)


# -------------------------------------------------------------------
# VALIDATE THE SOURCE
# -------------------------------------------------------------------

if not source_smod_2020.exists():
    raise FileNotFoundError(
        "The discovered SMOD 2020 source file no longer exists:\n"
        f"{source_smod_2020}"
    )

source_size = source_smod_2020.stat().st_size

if source_size <= 0:
    raise ValueError(
        "The SMOD 2020 source file is empty."
    )

source_checksum = calculate_sha256(
    source_smod_2020
)

print("\nSource size:")
print(
    f"{source_size:,} bytes "
    f"({source_size / (1024 ** 2):.4f} MiB)"
)

print("\nSource SHA-256:")
print(source_checksum)


# -------------------------------------------------------------------
# CREATE THE DESTINATION FOLDER IF REQUIRED
# -------------------------------------------------------------------

destination_folder.mkdir(
    parents=True,
    exist_ok=True,
)


# -------------------------------------------------------------------
# COPY THE FILE SAFELY
# -------------------------------------------------------------------

if destination_smod_2020.exists():
    existing_destination_size = (
        destination_smod_2020.stat().st_size
    )

    existing_destination_checksum = (
        calculate_sha256(
            destination_smod_2020
        )
    )

    print(
        "\nA destination file already exists."
    )

    print(
        "Existing destination size:",
        f"{existing_destination_size:,} bytes",
    )

    print(
        "Existing destination SHA-256:",
        existing_destination_checksum,
    )

    if (
        existing_destination_size == source_size
        and existing_destination_checksum == source_checksum
    ):
        print(
            "\nThe existing destination is identical "
            "to the source. No copy was required."
        )

    else:
        raise FileExistsError(
            "A different file already exists at the destination. "
            "The file was not overwritten."
        )

else:
    shutil.copy2(
        source_smod_2020,
        destination_smod_2020,
    )

    print(
        "\nThe SMOD 2020 file was copied into "
        "the main pilot folder."
    )


# -------------------------------------------------------------------
# VERIFY THE COPIED FILE
# -------------------------------------------------------------------

if not destination_smod_2020.exists():
    raise FileNotFoundError(
        "The destination file does not exist after copying."
    )

destination_size = (
    destination_smod_2020.stat().st_size
)

destination_checksum = calculate_sha256(
    destination_smod_2020
)

print("\nDestination size:")
print(
    f"{destination_size:,} bytes "
    f"({destination_size / (1024 ** 2):.4f} MiB)"
)

print("\nDestination SHA-256:")
print(destination_checksum)


assert destination_size == source_size, (
    "Source and destination sizes do not match."
)

assert destination_checksum == source_checksum, (
    "Source and destination checksums do not match."
)


# -------------------------------------------------------------------
# VERIFY THE COMPLETE PILOT INVENTORY
# -------------------------------------------------------------------

expected_filenames = [
    "ghsl_bengaluru_smod_2015_1km.tif",
    "ghsl_bengaluru_smod_2020_1km.tif",
    "ghsl_bengaluru_population_2015_100m.tif",
    "ghsl_bengaluru_population_2020_100m.tif",
    "ghsl_bengaluru_built_surface_2015_100m.tif",
    "ghsl_bengaluru_built_surface_2020_100m.tif",
    "ghsl_bengaluru_building_height_2018_100m.tif",
]

missing_files = []

print("\n" + "=" * 85)
print("FINAL PILOT FILE INVENTORY")
print("=" * 85)

for filename in expected_filenames:
    file_path = destination_folder / filename

    if file_path.exists():
        size_mib = (
            file_path.stat().st_size
            / (1024 ** 2)
        )

        print(
            f"FOUND   {filename:<58} "
            f"{size_mib:>8.4f} MiB"
        )

    else:
        print(
            f"MISSING {filename}"
        )

        missing_files.append(
            filename
        )


assert not missing_files, (
    "The pilot inventory remains incomplete: "
    + ", ".join(missing_files)
)


print("\n" + "=" * 85)
print("VALIDATION RESULT")
print("=" * 85)

print(
    "All seven Bengaluru pilot GeoTIFF files are now "
    "present in the main export folder."
)

print(
    "The copied SMOD 2020 file exactly matches its "
    "source according to file size and SHA-256 checksum."
)

print(
    "The original file in GHSL_Bengaluru_Pilot (1) "
    "has been retained for safety."
)
```

```python
# -------------------------------------------------------------------
# CELL 16: VALIDATE ALL BENGALURU PILOT GEOTIFF FILES
# -------------------------------------------------------------------

from pathlib import Path
import hashlib
import json

import numpy as np
import rasterio


PILOT_FOLDER = Path(
    "/content/drive/MyDrive/GHSL_Bengaluru_Pilot"
)


# -------------------------------------------------------------------
# FILE DEFINITIONS
# -------------------------------------------------------------------

pilot_files = {
    "smod_2015": (
        PILOT_FOLDER
        / "ghsl_bengaluru_smod_2015_1km.tif"
    ),
    "smod_2020": (
        PILOT_FOLDER
        / "ghsl_bengaluru_smod_2020_1km.tif"
    ),
    "population_2015": (
        PILOT_FOLDER
        / "ghsl_bengaluru_population_2015_100m.tif"
    ),
    "population_2020": (
        PILOT_FOLDER
        / "ghsl_bengaluru_population_2020_100m.tif"
    ),
    "built_surface_2015": (
        PILOT_FOLDER
        / "ghsl_bengaluru_built_surface_2015_100m.tif"
    ),
    "built_surface_2020": (
        PILOT_FOLDER
        / "ghsl_bengaluru_built_surface_2020_100m.tif"
    ),
    "building_height_2018": (
        PILOT_FOLDER
        / "ghsl_bengaluru_building_height_2018_100m.tif"
    ),
}


expected_band_counts = {
    "smod_2015": 1,
    "smod_2020": 1,
    "population_2015": 1,
    "population_2020": 1,
    "built_surface_2015": 2,
    "built_surface_2020": 2,
    "building_height_2018": 1,
}


expected_resolutions = {
    "smod_2015": 1000.0,
    "smod_2020": 1000.0,
    "population_2015": 100.0,
    "population_2020": 100.0,
    "built_surface_2015": 100.0,
    "built_surface_2020": 100.0,
    "building_height_2018": 100.0,
}


valid_smod_codes = {
    10,
    11,
    12,
    13,
    21,
    22,
    23,
    30,
}


# -------------------------------------------------------------------
# CHECKSUM FUNCTION
# -------------------------------------------------------------------

def calculate_sha256(file_path, chunk_size=1024 * 1024):
    sha256 = hashlib.sha256()

    with file_path.open("rb") as file_handle:
        while True:
            chunk = file_handle.read(chunk_size)

            if not chunk:
                break

            sha256.update(chunk)

    return sha256.hexdigest()


# -------------------------------------------------------------------
# BAND STATISTICS FUNCTION
# -------------------------------------------------------------------

def calculate_band_statistics(dataset, band_index):
    """
    Read one raster band as a masked array and calculate
    validation statistics.
    """

    band = dataset.read(
        band_index,
        masked=True,
    )

    compressed_values = band.compressed()

    total_pixel_count = int(band.size)
    masked_pixel_count = int(
        np.ma.count_masked(band)
    )
    valid_pixel_count = int(
        compressed_values.size
    )

    if valid_pixel_count > 0:
        minimum = float(
            np.min(compressed_values)
        )
        maximum = float(
            np.max(compressed_values)
        )
        mean = float(
            np.mean(
                compressed_values,
                dtype=np.float64,
            )
        )
        zero_count = int(
            np.count_nonzero(
                compressed_values == 0
            )
        )
        negative_count = int(
            np.count_nonzero(
                compressed_values < 0
            )
        )
        nan_count = int(
            np.count_nonzero(
                np.isnan(compressed_values)
            )
        ) if np.issubdtype(
            compressed_values.dtype,
            np.floating,
        ) else 0

        infinite_count = int(
            np.count_nonzero(
                np.isinf(compressed_values)
            )
        ) if np.issubdtype(
            compressed_values.dtype,
            np.floating,
        ) else 0

        unique_values = np.unique(
            compressed_values
        )

        if unique_values.size <= 25:
            unique_values_for_report = [
                float(value)
                for value in unique_values
            ]
        else:
            unique_values_for_report = None

    else:
        minimum = None
        maximum = None
        mean = None
        zero_count = 0
        negative_count = 0
        nan_count = 0
        infinite_count = 0
        unique_values_for_report = []

    return {
        "band_index": band_index,
        "dtype": str(band.dtype),
        "total_pixel_count": total_pixel_count,
        "valid_pixel_count": valid_pixel_count,
        "masked_pixel_count": masked_pixel_count,
        "minimum": minimum,
        "maximum": maximum,
        "mean": mean,
        "zero_count": zero_count,
        "negative_count": negative_count,
        "nan_count": nan_count,
        "infinite_count": infinite_count,
        "unique_values_if_small": (
            unique_values_for_report
        ),
    }


# -------------------------------------------------------------------
# VALIDATE INDIVIDUAL FILES
# -------------------------------------------------------------------

validation_report = {}
validation_errors = []


print("=" * 100)
print("BENGALURU PILOT GEOTIFF VALIDATION")
print("=" * 100)


for dataset_name, file_path in pilot_files.items():
    print("\n" + "=" * 100)
    print("Dataset:", dataset_name)
    print("File:", file_path.name)
    print("=" * 100)

    if not file_path.exists():
        validation_errors.append(
            f"{dataset_name}: file does not exist"
        )

        print("ERROR: File does not exist.")
        continue

    file_size_bytes = file_path.stat().st_size
    checksum = calculate_sha256(file_path)

    print(
        "File size:",
        f"{file_size_bytes:,} bytes",
        f"({file_size_bytes / (1024 ** 2):.4f} MiB)",
    )

    print("SHA-256:", checksum)

    try:
        with rasterio.open(file_path) as dataset:
            transform = dataset.transform

            x_resolution = abs(
                float(transform.a)
            )

            y_resolution = abs(
                float(transform.e)
            )

            metadata = {
                "file_name": file_path.name,
                "file_size_bytes": file_size_bytes,
                "sha256": checksum,
                "driver": dataset.driver,
                "band_count": dataset.count,
                "width": dataset.width,
                "height": dataset.height,
                "dtypes": list(dataset.dtypes),
                "crs": (
                    dataset.crs.to_string()
                    if dataset.crs is not None
                    else None
                ),
                "transform": [
                    float(transform.a),
                    float(transform.b),
                    float(transform.c),
                    float(transform.d),
                    float(transform.e),
                    float(transform.f),
                ],
                "x_resolution": x_resolution,
                "y_resolution": y_resolution,
                "bounds": [
                    float(dataset.bounds.left),
                    float(dataset.bounds.bottom),
                    float(dataset.bounds.right),
                    float(dataset.bounds.top),
                ],
                "nodata": dataset.nodata,
                "nodatavals": [
                    None if value is None else float(value)
                    for value in dataset.nodatavals
                ],
                "descriptions": list(
                    dataset.descriptions
                ),
                "color_interpretations": [
                    item.name
                    for item in dataset.colorinterp
                ],
                "band_statistics": [],
            }

            print("Driver:", dataset.driver)
            print("CRS:", metadata["crs"])
            print(
                "Dimensions:",
                dataset.width,
                "x",
                dataset.height,
            )
            print("Band count:", dataset.count)
            print("Data types:", dataset.dtypes)
            print(
                "Resolution:",
                x_resolution,
                "x",
                y_resolution,
                "metres",
            )
            print("Transform:", metadata["transform"])
            print("Bounds:", metadata["bounds"])
            print("NoData:", dataset.nodata)
            print(
                "Band descriptions:",
                dataset.descriptions,
            )

            expected_band_count = (
                expected_band_counts[dataset_name]
            )

            expected_resolution = (
                expected_resolutions[dataset_name]
            )

            if dataset.count != expected_band_count:
                validation_errors.append(
                    f"{dataset_name}: expected "
                    f"{expected_band_count} band(s), "
                    f"found {dataset.count}"
                )

            if not np.isclose(
                x_resolution,
                expected_resolution,
            ):
                validation_errors.append(
                    f"{dataset_name}: unexpected "
                    f"X resolution {x_resolution}"
                )

            if not np.isclose(
                y_resolution,
                expected_resolution,
            ):
                validation_errors.append(
                    f"{dataset_name}: unexpected "
                    f"Y resolution {y_resolution}"
                )

            if dataset.crs is None:
                validation_errors.append(
                    f"{dataset_name}: CRS is missing"
                )

            for band_index in range(
                1,
                dataset.count + 1,
            ):
                statistics = (
                    calculate_band_statistics(
                        dataset,
                        band_index,
                    )
                )

                metadata["band_statistics"].append(
                    statistics
                )

                print(
                    f"\nBand {band_index} statistics:"
                )

                for key, value in statistics.items():
                    print(
                        f"  {key}: {value}"
                    )

                if statistics["valid_pixel_count"] == 0:
                    validation_errors.append(
                        f"{dataset_name}, band "
                        f"{band_index}: no valid pixels"
                    )

                if statistics["negative_count"] > 0:
                    validation_errors.append(
                        f"{dataset_name}, band "
                        f"{band_index}: negative values found"
                    )

                if statistics["nan_count"] > 0:
                    validation_errors.append(
                        f"{dataset_name}, band "
                        f"{band_index}: NaN values found"
                    )

                if statistics["infinite_count"] > 0:
                    validation_errors.append(
                        f"{dataset_name}, band "
                        f"{band_index}: infinite values found"
                    )

            validation_report[dataset_name] = metadata

    except Exception as error:
        validation_errors.append(
            f"{dataset_name}: Rasterio error: {error}"
        )

        print(
            "ERROR while opening or inspecting raster:",
            error,
        )


# -------------------------------------------------------------------
# VALIDATE SMOD CLASS CODES
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("SMOD CLASS-CODE VALIDATION")
print("=" * 100)


for dataset_name in [
    "smod_2015",
    "smod_2020",
]:
    file_path = pilot_files[dataset_name]

    with rasterio.open(file_path) as dataset:
        smod_array = dataset.read(
            1,
            masked=True,
        )

        observed_codes = {
            int(value)
            for value in np.unique(
                smod_array.compressed()
            )
        }

        invalid_codes = (
            observed_codes
            - valid_smod_codes
            - {0}
        )

        print(
            f"\n{dataset_name}:"
        )
        print(
            "Observed codes:",
            sorted(observed_codes),
        )
        print(
            "Invalid codes:",
            sorted(invalid_codes),
        )

        if invalid_codes:
            validation_errors.append(
                f"{dataset_name}: invalid SMOD "
                f"codes {sorted(invalid_codes)}"
            )


# -------------------------------------------------------------------
# PAIRWISE ALIGNMENT CHECKS
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("PAIRWISE GRID ALIGNMENT")
print("=" * 100)


def compare_grids(
    first_name,
    second_name,
    require_same_dimensions=True,
):
    first = validation_report[first_name]
    second = validation_report[second_name]

    same_crs = (
        first["crs"] == second["crs"]
    )

    same_transform = (
        first["transform"]
        == second["transform"]
    )

    same_dimensions = (
        first["width"] == second["width"]
        and first["height"] == second["height"]
    )

    same_bounds = (
        np.allclose(
            first["bounds"],
            second["bounds"],
        )
    )

    passed = (
        same_crs
        and same_transform
        and same_bounds
        and (
            same_dimensions
            if require_same_dimensions
            else True
        )
    )

    print(
        f"\n{first_name} vs {second_name}"
    )
    print("Same CRS:", same_crs)
    print("Same transform:", same_transform)
    print("Same dimensions:", same_dimensions)
    print("Same bounds:", same_bounds)
    print(
        "Result:",
        "PASS" if passed else "FAIL",
    )

    if not passed:
        validation_errors.append(
            f"Grid mismatch: {first_name} "
            f"vs {second_name}"
        )

    return passed


compare_grids(
    "smod_2015",
    "smod_2020",
)

compare_grids(
    "population_2015",
    "population_2020",
)

compare_grids(
    "built_surface_2015",
    "built_surface_2020",
)

compare_grids(
    "population_2015",
    "built_surface_2015",
)

compare_grids(
    "population_2020",
    "built_surface_2020",
)

compare_grids(
    "population_2020",
    "building_height_2018",
)


# -------------------------------------------------------------------
# VERIFY 1 KM TO 100 M NESTING
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("SMOD-TO-100 M NESTING")
print("=" * 100)


smod_metadata = validation_report[
    "smod_2020"
]

population_metadata = validation_report[
    "population_2020"
]

smod_transform = smod_metadata["transform"]
population_transform = (
    population_metadata["transform"]
)

same_origin = (
    np.isclose(
        smod_transform[2],
        population_transform[2],
    )
    and
    np.isclose(
        smod_transform[5],
        population_transform[5],
    )
)

pixel_ratio_x = (
    smod_metadata["x_resolution"]
    / population_metadata["x_resolution"]
)

pixel_ratio_y = (
    smod_metadata["y_resolution"]
    / population_metadata["y_resolution"]
)

width_ratio = (
    population_metadata["width"]
    / smod_metadata["width"]
)

height_ratio = (
    population_metadata["height"]
    / smod_metadata["height"]
)

print("Same grid origin:", same_origin)
print(
    "Pixel-size ratio X:",
    pixel_ratio_x,
)
print(
    "Pixel-size ratio Y:",
    pixel_ratio_y,
)
print(
    "Raster-width ratio:",
    width_ratio,
)
print(
    "Raster-height ratio:",
    height_ratio,
)


exact_pixel_nesting = (
    same_origin
    and np.isclose(pixel_ratio_x, 10.0)
    and np.isclose(pixel_ratio_y, 10.0)
)

if not exact_pixel_nesting:
    validation_errors.append(
        "SMOD and 100 m exports do not have "
        "the expected 10 x 10 nesting relationship"
    )


# -------------------------------------------------------------------
# SAVE MACHINE-READABLE REPORT
# -------------------------------------------------------------------

validation_report_path = (
    PILOT_FOLDER
    / "bengaluru_pilot_raster_validation.json"
)

with validation_report_path.open(
    "w",
    encoding="utf-8",
) as report_file:
    json.dump(
        validation_report,
        report_file,
        indent=2,
    )


print("\n" + "=" * 100)
print("VALIDATION REPORT SAVED")
print("=" * 100)

print(validation_report_path)


# -------------------------------------------------------------------
# FINAL RESULT
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("FINAL VALIDATION RESULT")
print("=" * 100)


if validation_errors:
    print(
        "VALIDATION COMPLETED WITH ISSUES:"
    )

    for index, error in enumerate(
        validation_errors,
        start=1,
    ):
        print(
            f"{index}. {error}"
        )

    raise AssertionError(
        "One or more GeoTIFF validation checks failed. "
        "Do not upload the files to HDFS yet."
    )

else:
    print(
        "SUCCESS: All seven Bengaluru pilot GeoTIFFs "
        "passed the current validation checks."
    )

    print(
        "The files are readable and contain the "
        "expected bands, resolutions, values, and grids."
    )

    print(
        "The pilot files are ready for the local "
        "curation and HDFS-ingestion stages."
    )
```

```python
# -------------------------------------------------------------------
# CELL 17: CREATE AND VALIDATE CURATED 100 M PILOT RASTERS
# -------------------------------------------------------------------
#
# Cell 16 validated the raw Earth Engine exports and found:
#
# 1. Population and building-height files contain NaN values, but their
#    GeoTIFF metadata does not declare a NoData value.
#
# 2. Built-up rasters contain valid zero values, but cells outside the
#    common GHSL coverage need to be distinguished from valid zero.
#
# 3. SMOD and the 100 m rasters were independently cropped to their
#    native grids. The global grids are nested, but the cropped files
#    do not start at the same top-left coordinate.
#
# This cell creates analysis-ready curated copies.
#
# RAW FILES ARE NOT MODIFIED.
#
# Curated conventions:
#
# Population and building height:
#     dtype  = float32
#     NoData = -9999.0
#
# Built-up surface:
#     dtype  = uint16
#     NoData = 65535
#     Valid zero remains zero.
#
# SMOD on the 100 m analytical grid:
#     dtype  = uint8
#     NoData = 0
#     Valid SMOD values remain unchanged.
#
# SMOD is expanded using nearest-neighbour categorical reprojection.
# Its analytical grid becomes 100 m, but its semantic/source
# resolution remains 1 km.

from pathlib import Path
import json
import hashlib

import numpy as np
import rasterio
from rasterio.warp import reproject, Resampling


# -------------------------------------------------------------------
# PATHS AND CONSTANTS
# -------------------------------------------------------------------

RAW_FOLDER = Path(
    "/content/drive/MyDrive/GHSL_Bengaluru_Pilot"
)

CURATED_FOLDER = RAW_FOLDER / "curated"

CURATED_FOLDER.mkdir(
    parents=True,
    exist_ok=True,
)

FLOAT_NODATA = np.float32(-9999.0)
BUILT_NODATA = np.uint16(65535)
SMOD_NODATA = np.uint8(0)

VALID_SMOD_CODES = {
    10,
    11,
    12,
    13,
    21,
    22,
    23,
    30,
}


raw_files = {
    "smod_2015": (
        RAW_FOLDER
        / "ghsl_bengaluru_smod_2015_1km.tif"
    ),
    "smod_2020": (
        RAW_FOLDER
        / "ghsl_bengaluru_smod_2020_1km.tif"
    ),
    "population_2015": (
        RAW_FOLDER
        / "ghsl_bengaluru_population_2015_100m.tif"
    ),
    "population_2020": (
        RAW_FOLDER
        / "ghsl_bengaluru_population_2020_100m.tif"
    ),
    "built_surface_2015": (
        RAW_FOLDER
        / "ghsl_bengaluru_built_surface_2015_100m.tif"
    ),
    "built_surface_2020": (
        RAW_FOLDER
        / "ghsl_bengaluru_built_surface_2020_100m.tif"
    ),
    "building_height_2018": (
        RAW_FOLDER
        / "ghsl_bengaluru_building_height_2018_100m.tif"
    ),
}


curated_files = {
    "population_2015": (
        CURATED_FOLDER
        / "ghsl_bengaluru_population_2015_100m_curated.tif"
    ),
    "population_2020": (
        CURATED_FOLDER
        / "ghsl_bengaluru_population_2020_100m_curated.tif"
    ),
    "built_surface_2015": (
        CURATED_FOLDER
        / "ghsl_bengaluru_built_surface_2015_100m_curated.tif"
    ),
    "built_surface_2020": (
        CURATED_FOLDER
        / "ghsl_bengaluru_built_surface_2020_100m_curated.tif"
    ),
    "building_height_2018": (
        CURATED_FOLDER
        / "ghsl_bengaluru_building_height_2018_100m_curated.tif"
    ),
    "smod_2015_100m": (
        CURATED_FOLDER
        / "ghsl_bengaluru_smod_2015_on_100m_grid.tif"
    ),
    "smod_2020_100m": (
        CURATED_FOLDER
        / "ghsl_bengaluru_smod_2020_on_100m_grid.tif"
    ),
}


for dataset_name, file_path in raw_files.items():
    if not file_path.exists():
        raise FileNotFoundError(
            f"Missing raw file for {dataset_name}: {file_path}"
        )


# -------------------------------------------------------------------
# HELPER: CHECKSUM
# -------------------------------------------------------------------

def calculate_sha256(file_path, chunk_size=1024 * 1024):
    sha256 = hashlib.sha256()

    with file_path.open("rb") as file_handle:
        while True:
            chunk = file_handle.read(chunk_size)

            if not chunk:
                break

            sha256.update(chunk)

    return sha256.hexdigest()


# -------------------------------------------------------------------
# HELPER: LOAD THE COMMON 100 M COVERAGE MASK
# -------------------------------------------------------------------

def load_population_coverage_mask(population_path):
    """
    Return True where the raw population raster contains a finite value.

    Zero population is valid and remains inside the coverage mask.
    NaN or infinity is considered outside valid analytical coverage.
    """

    with rasterio.open(population_path) as source:
        population = source.read(1)

        coverage_mask = np.isfinite(population)

        grid = {
            "crs": source.crs,
            "transform": source.transform,
            "width": source.width,
            "height": source.height,
            "bounds": source.bounds,
            "profile": source.profile.copy(),
        }

    return coverage_mask, grid


coverage_mask_2015, grid_2015 = (
    load_population_coverage_mask(
        raw_files["population_2015"]
    )
)

coverage_mask_2020, grid_2020 = (
    load_population_coverage_mask(
        raw_files["population_2020"]
    )
)


assert grid_2015["crs"] == grid_2020["crs"]
assert grid_2015["transform"] == grid_2020["transform"]
assert grid_2015["width"] == grid_2020["width"]
assert grid_2015["height"] == grid_2020["height"]
assert grid_2015["bounds"] == grid_2020["bounds"]


# The raw validation showed the same number and location pattern of
# NaN cells. Confirm whether the coverage masks are identical.

coverage_masks_identical = np.array_equal(
    coverage_mask_2015,
    coverage_mask_2020,
)

print("=" * 100)
print("COMMON COVERAGE MASK")
print("=" * 100)

print(
    "2015 valid population cells:",
    f"{np.count_nonzero(coverage_mask_2015):,}",
)

print(
    "2020 valid population cells:",
    f"{np.count_nonzero(coverage_mask_2020):,}",
)

print(
    "2015 and 2020 coverage masks are identical:",
    coverage_masks_identical,
)


if not coverage_masks_identical:
    print(
        "Warning: The 2015 and 2020 masks differ. "
        "Each epoch will retain its own coverage mask."
    )


# Building height is cross-sectional for 2018.
# Use the union of valid historical coverage only as its external
# analytical footprint. Its own finite-value mask is also retained.

height_context_coverage_mask = (
    coverage_mask_2015
    | coverage_mask_2020
)


# -------------------------------------------------------------------
# CURATE POPULATION
# -------------------------------------------------------------------

def curate_population(
    source_path,
    destination_path,
    expected_coverage_mask,
    epoch,
):
    with rasterio.open(source_path) as source:
        source_array = source.read(1).astype(
            np.float32,
            copy=False,
        )

        if source_array.shape != expected_coverage_mask.shape:
            raise ValueError(
                f"Coverage-mask shape mismatch for {source_path.name}."
            )

        finite_mask = np.isfinite(source_array)

        if not np.array_equal(
            finite_mask,
            expected_coverage_mask,
        ):
            raise ValueError(
                f"Population validity mask changed unexpectedly for "
                f"{source_path.name}."
            )

        curated_array = np.full(
            source_array.shape,
            FLOAT_NODATA,
            dtype=np.float32,
        )

        curated_array[finite_mask] = (
            source_array[finite_mask]
        )

        profile = source.profile.copy()

        profile.update(
            driver="GTiff",
            dtype="float32",
            count=1,
            nodata=float(FLOAT_NODATA),
            compress="LZW",
            tiled=True,
            BIGTIFF="IF_SAFER",
        )

        with rasterio.open(
            destination_path,
            "w",
            **profile,
        ) as destination:
            destination.write(
                curated_array,
                1,
            )

            destination.set_band_description(
                1,
                "population_count",
            )

            destination.update_tags(
                product="GHS_POP",
                epoch=str(epoch),
                analytical_resolution_m="100",
                source_resolution_m="100",
                nodata_policy="non_finite_to_minus_9999",
                valid_zero_policy="preserved",
            )

    return {
        "source": str(source_path),
        "destination": str(destination_path),
        "epoch": epoch,
        "valid_cells": int(
            np.count_nonzero(finite_mask)
        ),
        "nodata_cells": int(
            np.count_nonzero(~finite_mask)
        ),
    }


# -------------------------------------------------------------------
# CURATE BUILDING HEIGHT
# -------------------------------------------------------------------

def curate_building_height(
    source_path,
    destination_path,
    external_coverage_mask,
):
    with rasterio.open(source_path) as source:
        source_array = source.read(1).astype(
            np.float32,
            copy=False,
        )

        if source_array.shape != external_coverage_mask.shape:
            raise ValueError(
                "Building-height and analytical-mask shapes differ."
            )

        own_finite_mask = np.isfinite(
            source_array
        )

        final_valid_mask = (
            own_finite_mask
            & external_coverage_mask
        )

        curated_array = np.full(
            source_array.shape,
            FLOAT_NODATA,
            dtype=np.float32,
        )

        curated_array[final_valid_mask] = (
            source_array[final_valid_mask]
        )

        profile = source.profile.copy()

        profile.update(
            driver="GTiff",
            dtype="float32",
            count=1,
            nodata=float(FLOAT_NODATA),
            compress="LZW",
            tiled=True,
            BIGTIFF="IF_SAFER",
        )

        with rasterio.open(
            destination_path,
            "w",
            **profile,
        ) as destination:
            destination.write(
                curated_array,
                1,
            )

            destination.set_band_description(
                1,
                "built_height_2018",
            )

            destination.update_tags(
                product="GHS_BUILT_H",
                reference_year="2018",
                temporal_policy="cross_sectional_only",
                analytical_resolution_m="100",
                source_resolution_m="100",
                nodata_policy="non_finite_to_minus_9999",
                valid_zero_policy="preserved",
            )

    return {
        "source": str(source_path),
        "destination": str(destination_path),
        "reference_year": 2018,
        "valid_cells": int(
            np.count_nonzero(final_valid_mask)
        ),
        "nodata_cells": int(
            np.count_nonzero(~final_valid_mask)
        ),
    }


# -------------------------------------------------------------------
# CURATE BUILT-UP SURFACE
# -------------------------------------------------------------------

def curate_built_surface(
    source_path,
    destination_path,
    coverage_mask,
    epoch,
):
    with rasterio.open(source_path) as source:
        if source.count != 2:
            raise ValueError(
                f"{source_path.name} must contain two bands."
            )

        total_built = source.read(1)
        nonresidential_built = source.read(2)

        if total_built.shape != coverage_mask.shape:
            raise ValueError(
                f"Coverage-mask shape mismatch for {source_path.name}."
            )

        curated_total = np.full(
            total_built.shape,
            BUILT_NODATA,
            dtype=np.uint16,
        )

        curated_nonresidential = np.full(
            nonresidential_built.shape,
            BUILT_NODATA,
            dtype=np.uint16,
        )

        curated_total[coverage_mask] = (
            total_built[coverage_mask]
        )

        curated_nonresidential[coverage_mask] = (
            nonresidential_built[coverage_mask]
        )

        # Validate that non-residential surface does not exceed
        # total built-up surface on valid analytical cells.

        incompatible_mask = (
            coverage_mask
            & (
                nonresidential_built
                > total_built
            )
        )

        incompatible_count = int(
            np.count_nonzero(
                incompatible_mask
            )
        )

        if incompatible_count > 0:
            raise ValueError(
                f"{source_path.name} contains "
                f"{incompatible_count:,} valid cells where "
                "non-residential built surface exceeds total "
                "built surface."
            )

        profile = source.profile.copy()

        profile.update(
            driver="GTiff",
            dtype="uint16",
            count=2,
            nodata=int(BUILT_NODATA),
            compress="LZW",
            tiled=True,
            BIGTIFF="IF_SAFER",
        )

        with rasterio.open(
            destination_path,
            "w",
            **profile,
        ) as destination:
            destination.write(
                curated_total,
                1,
            )

            destination.write(
                curated_nonresidential,
                2,
            )

            destination.set_band_description(
                1,
                "built_surface",
            )

            destination.set_band_description(
                2,
                "built_surface_nres",
            )

            destination.update_tags(
                product="GHS_BUILT_S",
                epoch=str(epoch),
                analytical_resolution_m="100",
                source_resolution_m="100",
                nodata_policy="outside_population_coverage_to_65535",
                valid_zero_policy="preserved",
            )

    return {
        "source": str(source_path),
        "destination": str(destination_path),
        "epoch": epoch,
        "valid_cells": int(
            np.count_nonzero(
                coverage_mask
            )
        ),
        "nodata_cells": int(
            np.count_nonzero(
                ~coverage_mask
            )
        ),
        "nonresidential_exceeds_total_cells": (
            incompatible_count
        ),
    }


# -------------------------------------------------------------------
# EXPAND SMOD TO THE EXACT 100 M ANALYTICAL GRID
# -------------------------------------------------------------------

def expand_smod_to_100m(
    smod_source_path,
    population_reference_path,
    destination_path,
    coverage_mask,
    epoch,
):
    with rasterio.open(
        smod_source_path
    ) as smod_source, rasterio.open(
        population_reference_path
    ) as population_reference:

        if smod_source.crs != population_reference.crs:
            raise ValueError(
                "SMOD and population CRS values do not match."
            )

        if (
            population_reference.height,
            population_reference.width,
        ) != coverage_mask.shape:
            raise ValueError(
                "SMOD destination mask and population grid differ."
            )

        source_smod = smod_source.read(
            1
        ).astype(
            np.int16,
            copy=False,
        )

        destination_smod = np.full(
            (
                population_reference.height,
                population_reference.width,
            ),
            int(SMOD_NODATA),
            dtype=np.uint8,
        )

        reproject(
            source=source_smod,
            destination=destination_smod,
            src_transform=smod_source.transform,
            src_crs=smod_source.crs,
            src_nodata=0,
            dst_transform=population_reference.transform,
            dst_crs=population_reference.crs,
            dst_nodata=int(SMOD_NODATA),
            resampling=Resampling.nearest,
            init_dest_nodata=True,
        )

        # Apply the epoch-specific common analytical coverage mask.
        destination_smod[
            ~coverage_mask
        ] = SMOD_NODATA

        observed_codes = {
            int(value)
            for value in np.unique(
                destination_smod
            )
        }

        invalid_codes = (
            observed_codes
            - VALID_SMOD_CODES
            - {int(SMOD_NODATA)}
        )

        if invalid_codes:
            raise ValueError(
                f"Invalid SMOD values after reprojection: "
                f"{sorted(invalid_codes)}"
            )

        profile = population_reference.profile.copy()

        profile.update(
            driver="GTiff",
            dtype="uint8",
            count=1,
            nodata=int(SMOD_NODATA),
            compress="LZW",
            tiled=True,
            BIGTIFF="IF_SAFER",
        )

        with rasterio.open(
            destination_path,
            "w",
            **profile,
        ) as destination:
            destination.write(
                destination_smod,
                1,
            )

            destination.set_band_description(
                1,
                "smod_code",
            )

            destination.update_tags(
                product="GHS_SMOD",
                epoch=str(epoch),
                source_resolution_m="1000",
                analytical_resolution_m="100",
                semantic_resolution_m="1000",
                assignment_method="nearest_neighbour",
                nodata_code="0",
            )

    return {
        "source": str(smod_source_path),
        "population_grid_reference": str(
            population_reference_path
        ),
        "destination": str(destination_path),
        "epoch": epoch,
        "observed_codes": sorted(
            observed_codes
        ),
        "nodata_cells": int(
            np.count_nonzero(
                destination_smod
                == SMOD_NODATA
            )
        ),
    }


# -------------------------------------------------------------------
# RUN CURATION
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("CREATING CURATED RASTERS")
print("=" * 100)


curation_operations = {}


curation_operations[
    "population_2015"
] = curate_population(
    raw_files["population_2015"],
    curated_files["population_2015"],
    coverage_mask_2015,
    epoch=2015,
)

print(
    "Created:",
    curated_files["population_2015"].name,
)


curation_operations[
    "population_2020"
] = curate_population(
    raw_files["population_2020"],
    curated_files["population_2020"],
    coverage_mask_2020,
    epoch=2020,
)

print(
    "Created:",
    curated_files["population_2020"].name,
)


curation_operations[
    "building_height_2018"
] = curate_building_height(
    raw_files["building_height_2018"],
    curated_files["building_height_2018"],
    height_context_coverage_mask,
)

print(
    "Created:",
    curated_files["building_height_2018"].name,
)


curation_operations[
    "built_surface_2015"
] = curate_built_surface(
    raw_files["built_surface_2015"],
    curated_files["built_surface_2015"],
    coverage_mask_2015,
    epoch=2015,
)

print(
    "Created:",
    curated_files["built_surface_2015"].name,
)


curation_operations[
    "built_surface_2020"
] = curate_built_surface(
    raw_files["built_surface_2020"],
    curated_files["built_surface_2020"],
    coverage_mask_2020,
    epoch=2020,
)

print(
    "Created:",
    curated_files["built_surface_2020"].name,
)


curation_operations[
    "smod_2015_100m"
] = expand_smod_to_100m(
    raw_files["smod_2015"],
    raw_files["population_2015"],
    curated_files["smod_2015_100m"],
    coverage_mask_2015,
    epoch=2015,
)

print(
    "Created:",
    curated_files["smod_2015_100m"].name,
)


curation_operations[
    "smod_2020_100m"
] = expand_smod_to_100m(
    raw_files["smod_2020"],
    raw_files["population_2020"],
    curated_files["smod_2020_100m"],
    coverage_mask_2020,
    epoch=2020,
)

print(
    "Created:",
    curated_files["smod_2020_100m"].name,
)


# -------------------------------------------------------------------
# VALIDATE THE CURATED GRID
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("CURATED GRID VALIDATION")
print("=" * 100)


curated_metadata = {}


for dataset_name, file_path in curated_files.items():
    with rasterio.open(file_path) as dataset:
        metadata = {
            "file": str(file_path),
            "file_size_bytes": file_path.stat().st_size,
            "sha256": calculate_sha256(file_path),
            "width": dataset.width,
            "height": dataset.height,
            "count": dataset.count,
            "dtypes": list(dataset.dtypes),
            "crs": (
                dataset.crs.to_string()
                if dataset.crs is not None
                else None
            ),
            "transform": [
                float(dataset.transform.a),
                float(dataset.transform.b),
                float(dataset.transform.c),
                float(dataset.transform.d),
                float(dataset.transform.e),
                float(dataset.transform.f),
            ],
            "bounds": [
                float(dataset.bounds.left),
                float(dataset.bounds.bottom),
                float(dataset.bounds.right),
                float(dataset.bounds.top),
            ],
            "nodata": (
                None
                if dataset.nodata is None
                else float(dataset.nodata)
            ),
            "descriptions": list(
                dataset.descriptions
            ),
        }

        curated_metadata[
            dataset_name
        ] = metadata

        print(f"\n{dataset_name}")
        print(
            "  Dimensions:",
            dataset.width,
            "x",
            dataset.height,
        )
        print(
            "  Band count:",
            dataset.count,
        )
        print(
            "  Data types:",
            dataset.dtypes,
        )
        print(
            "  CRS:",
            metadata["crs"],
        )
        print(
            "  Transform:",
            metadata["transform"],
        )
        print(
            "  Bounds:",
            metadata["bounds"],
        )
        print(
            "  NoData:",
            dataset.nodata,
        )
        print(
            "  Descriptions:",
            dataset.descriptions,
        )


reference = curated_metadata[
    "population_2020"
]

grid_mismatches = []


for dataset_name, metadata in curated_metadata.items():
    same_crs = (
        metadata["crs"]
        == reference["crs"]
    )

    same_transform = np.allclose(
        metadata["transform"],
        reference["transform"],
    )

    same_dimensions = (
        metadata["width"]
        == reference["width"]
        and metadata["height"]
        == reference["height"]
    )

    same_bounds = np.allclose(
        metadata["bounds"],
        reference["bounds"],
    )

    passed = (
        same_crs
        and same_transform
        and same_dimensions
        and same_bounds
    )

    print(
        f"\nGrid check {dataset_name}:",
        "PASS" if passed else "FAIL",
    )

    if not passed:
        grid_mismatches.append(
            dataset_name
        )


assert not grid_mismatches, (
    "Curated grid mismatch for: "
    + ", ".join(grid_mismatches)
)


# -------------------------------------------------------------------
# VALIDATE FLOAT RASTERS
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("FLOAT NODATA VALIDATION")
print("=" * 100)


for dataset_name in [
    "population_2015",
    "population_2020",
    "building_height_2018",
]:
    with rasterio.open(
        curated_files[dataset_name]
    ) as dataset:
        array = dataset.read(1)

        nan_count = int(
            np.count_nonzero(
                np.isnan(array)
            )
        )

        infinite_count = int(
            np.count_nonzero(
                np.isinf(array)
            )
        )

        nodata_count = int(
            np.count_nonzero(
                array == FLOAT_NODATA
            )
        )

        valid_values = array[
            array != FLOAT_NODATA
        ]

        print(f"\n{dataset_name}")
        print("  NaN cells:", nan_count)
        print(
            "  Infinite cells:",
            infinite_count,
        )
        print(
            "  Explicit NoData cells:",
            nodata_count,
        )
        print(
            "  Valid zero cells:",
            int(
                np.count_nonzero(
                    valid_values == 0
                )
            ),
        )
        print(
            "  Valid minimum:",
            float(
                np.min(valid_values)
            ),
        )
        print(
            "  Valid maximum:",
            float(
                np.max(valid_values)
            ),
        )

        assert dataset.nodata == float(
            FLOAT_NODATA
        )
        assert nan_count == 0
        assert infinite_count == 0
        assert valid_values.size > 0
        assert np.min(valid_values) >= 0


# -------------------------------------------------------------------
# VALIDATE BUILT-UP RASTERS
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("BUILT-UP NODATA VALIDATION")
print("=" * 100)


for dataset_name in [
    "built_surface_2015",
    "built_surface_2020",
]:
    with rasterio.open(
        curated_files[dataset_name]
    ) as dataset:
        total_built = dataset.read(1)
        nonresidential_built = dataset.read(2)

        nodata_mask = (
            total_built == BUILT_NODATA
        )

        second_band_nodata_mask = (
            nonresidential_built
            == BUILT_NODATA
        )

        print(f"\n{dataset_name}")
        print(
            "  NoData cells, band 1:",
            f"{np.count_nonzero(nodata_mask):,}",
        )
        print(
            "  NoData cells, band 2:",
            f"{np.count_nonzero(second_band_nodata_mask):,}",
        )
        print(
            "  Valid zero cells, band 1:",
            f"{np.count_nonzero((total_built == 0) & ~nodata_mask):,}",
        )
        print(
            "  Valid zero cells, band 2:",
            f"{np.count_nonzero((nonresidential_built == 0) & ~second_band_nodata_mask):,}",
        )

        assert dataset.nodata == int(
            BUILT_NODATA
        )
        assert np.array_equal(
            nodata_mask,
            second_band_nodata_mask,
        )

        valid_mask = ~nodata_mask

        assert np.all(
            nonresidential_built[valid_mask]
            <= total_built[valid_mask]
        )


# -------------------------------------------------------------------
# VALIDATE CURATED SMOD
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("CURATED SMOD VALIDATION")
print("=" * 100)


for dataset_name in [
    "smod_2015_100m",
    "smod_2020_100m",
]:
    with rasterio.open(
        curated_files[dataset_name]
    ) as dataset:
        smod = dataset.read(1)

        observed_codes = {
            int(value)
            for value in np.unique(smod)
        }

        invalid_codes = (
            observed_codes
            - VALID_SMOD_CODES
            - {int(SMOD_NODATA)}
        )

        print(f"\n{dataset_name}")
        print(
            "  Observed codes:",
            sorted(observed_codes),
        )
        print(
            "  Invalid codes:",
            sorted(invalid_codes),
        )
        print(
            "  NoData cells:",
            f"{np.count_nonzero(smod == SMOD_NODATA):,}",
        )

        print(
            "  Source resolution metadata:",
            dataset.tags().get(
                "source_resolution_m"
            ),
        )

        print(
            "  Analytical resolution metadata:",
            dataset.tags().get(
                "analytical_resolution_m"
            ),
        )

        print(
            "  Assignment method:",
            dataset.tags().get(
                "assignment_method"
            ),
        )

        assert dataset.nodata == int(
            SMOD_NODATA
        )
        assert not invalid_codes
        assert (
            dataset.tags().get(
                "source_resolution_m"
            )
            == "1000"
        )
        assert (
            dataset.tags().get(
                "analytical_resolution_m"
            )
            == "100"
        )
        assert (
            dataset.tags().get(
                "assignment_method"
            )
            == "nearest_neighbour"
        )


# -------------------------------------------------------------------
# SAVE CURATION MANIFEST
# -------------------------------------------------------------------

manifest_path = (
    CURATED_FOLDER
    / "bengaluru_pilot_curation_manifest.json"
)


manifest = {
    "status": "success",
    "analytical_grid": {
        "resolution_m": 100,
        "crs": reference["crs"],
        "transform": reference["transform"],
        "width": reference["width"],
        "height": reference["height"],
        "bounds": reference["bounds"],
    },
    "nodata_policy": {
        "population": float(
            FLOAT_NODATA
        ),
        "building_height": float(
            FLOAT_NODATA
        ),
        "built_surface": int(
            BUILT_NODATA
        ),
        "smod": int(
            SMOD_NODATA
        ),
    },
    "smod_policy": {
        "source_resolution_m": 1000,
        "analytical_resolution_m": 100,
        "semantic_resolution_m": 1000,
        "assignment_method": (
            "nearest_neighbour"
        ),
    },
    "building_height_policy": {
        "reference_year": 2018,
        "temporal_policy": (
            "cross_sectional_only"
        ),
    },
    "coverage_masks_identical_2015_2020": (
        coverage_masks_identical
    ),
    "operations": curation_operations,
    "outputs": curated_metadata,
}


with manifest_path.open(
    "w",
    encoding="utf-8",
) as manifest_file:
    json.dump(
        manifest,
        manifest_file,
        indent=2,
    )


print("\n" + "=" * 100)
print("FINAL CURATION RESULT")
print("=" * 100)

print(
    "SUCCESS: All seven curated rasters share the exact "
    "same 100 m grid."
)

print(
    "Population and height NaN values were converted "
    "to explicit -9999 NoData."
)

print(
    "Built-up pixels outside the common coverage mask "
    "were converted to 65535 NoData."
)

print(
    "Valid population, height and built-up zero values "
    "were preserved."
)

print(
    "SMOD 2015 and 2020 were mapped to the exact 100 m "
    "analytical grid using nearest-neighbour assignment."
)

print(
    "SMOD remains semantically a 1 km classification."
)

print(
    "Raw Earth Engine exports were not modified."
)

print("\nCurated folder:")
print(CURATED_FOLDER)

print("\nCuration manifest:")
print(manifest_path)
```

```python
# -------------------------------------------------------------------
# CELL 18: CONVERT CURATED BENGALURU RASTERS TO PARQUET
# -------------------------------------------------------------------
#
# Outputs:
#
# 1. cell_reference_100m
#    One stable row per valid analytical cell.
#
# 2. grid_year/epoch=2015
#    One row per valid 100 m cell for 2015.
#
# 3. grid_year/epoch=2020
#    One row per valid 100 m cell for 2020.
#
# 4. building_height_2018
#    Separate cross-sectional building-height table.
#
# Building height is intentionally NOT inserted into every historical
# grid-year record.

from pathlib import Path
import json
import hashlib

import numpy as np
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq
import rasterio


# -------------------------------------------------------------------
# PATHS
# -------------------------------------------------------------------

CURATED_FOLDER = Path(
    "/content/drive/MyDrive/"
    "GHSL_Bengaluru_Pilot/curated"
)

PARQUET_ROOT = Path(
    "/content/drive/MyDrive/"
    "GHSL_Bengaluru_Pilot/parquet"
)

CELL_REFERENCE_FOLDER = (
    PARQUET_ROOT / "cell_reference_100m"
)

GRID_YEAR_FOLDER = (
    PARQUET_ROOT / "grid_year"
)

HEIGHT_FOLDER = (
    PARQUET_ROOT / "building_height_2018"
)

CELL_REFERENCE_FOLDER.mkdir(
    parents=True,
    exist_ok=True,
)

GRID_YEAR_FOLDER.mkdir(
    parents=True,
    exist_ok=True,
)

HEIGHT_FOLDER.mkdir(
    parents=True,
    exist_ok=True,
)


curated_paths = {
    "population_2015": (
        CURATED_FOLDER
        / "ghsl_bengaluru_population_2015_100m_curated.tif"
    ),
    "population_2020": (
        CURATED_FOLDER
        / "ghsl_bengaluru_population_2020_100m_curated.tif"
    ),
    "built_surface_2015": (
        CURATED_FOLDER
        / "ghsl_bengaluru_built_surface_2015_100m_curated.tif"
    ),
    "built_surface_2020": (
        CURATED_FOLDER
        / "ghsl_bengaluru_built_surface_2020_100m_curated.tif"
    ),
    "building_height_2018": (
        CURATED_FOLDER
        / "ghsl_bengaluru_building_height_2018_100m_curated.tif"
    ),
    "smod_2015": (
        CURATED_FOLDER
        / "ghsl_bengaluru_smod_2015_on_100m_grid.tif"
    ),
    "smod_2020": (
        CURATED_FOLDER
        / "ghsl_bengaluru_smod_2020_on_100m_grid.tif"
    ),
}


for dataset_name, file_path in curated_paths.items():
    if not file_path.exists():
        raise FileNotFoundError(
            f"Missing curated raster for {dataset_name}: "
            f"{file_path}"
        )


# -------------------------------------------------------------------
# CONSTANTS
# -------------------------------------------------------------------

GRID_VERSION = "ghsl_100m_bengaluru_v1"

FLOAT_NODATA = np.float32(-9999.0)
BUILT_NODATA = np.uint16(65535)
SMOD_NODATA = np.uint8(0)

SMOD_CLASS_NAMES = {
    10: "Water",
    11: "Very low density rural",
    12: "Low density rural",
    13: "Rural cluster",
    21: "Suburban or peri-urban",
    22: "Semi-dense urban cluster",
    23: "Dense urban cluster",
    30: "Urban centre",
}


SMOD_GROUPS = {
    10: "water",
    11: "rural",
    12: "rural",
    13: "rural_cluster",
    21: "suburban_peri_urban",
    22: "semi_dense_urban_cluster",
    23: "dense_urban_cluster",
    30: "urban_centre",
}


# -------------------------------------------------------------------
# HELPER FUNCTIONS
# -------------------------------------------------------------------

def calculate_sha256(file_path, chunk_size=1024 * 1024):
    sha256 = hashlib.sha256()

    with file_path.open("rb") as file_handle:
        while True:
            chunk = file_handle.read(chunk_size)

            if not chunk:
                break

            sha256.update(chunk)

    return sha256.hexdigest()


def read_single_band(file_path):
    with rasterio.open(file_path) as dataset:
        return (
            dataset.read(1),
            dataset.profile.copy(),
            dataset.transform,
            dataset.crs,
        )


def read_built_bands(file_path):
    with rasterio.open(file_path) as dataset:
        return (
            dataset.read(1),
            dataset.read(2),
            dataset.profile.copy(),
            dataset.transform,
            dataset.crs,
        )


def write_parquet_table(
    dataframe,
    output_path,
):
    output_path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    table = pa.Table.from_pandas(
        dataframe,
        preserve_index=False,
    )

    pq.write_table(
        table,
        output_path,
        compression="snappy",
        use_dictionary=True,
        write_statistics=True,
    )


# -------------------------------------------------------------------
# LOAD THE REFERENCE GRID
# -------------------------------------------------------------------

with rasterio.open(
    curated_paths["population_2020"]
) as reference_dataset:
    grid_width = reference_dataset.width
    grid_height = reference_dataset.height
    grid_transform = reference_dataset.transform
    grid_crs = reference_dataset.crs
    grid_bounds = reference_dataset.bounds


print("=" * 100)
print("PARQUET CONVERSION")
print("=" * 100)

print("\nAnalytical grid:")
print("Width:", grid_width)
print("Height:", grid_height)
print("CRS:", grid_crs)
print("Transform:", grid_transform)
print("Bounds:", grid_bounds)


# -------------------------------------------------------------------
# CREATE STABLE GRID INDICES AND CELL IDS
# -------------------------------------------------------------------

row_indices, column_indices = np.indices(
    (
        grid_height,
        grid_width,
    ),
    dtype=np.int32,
)

flat_rows = row_indices.ravel()
flat_columns = column_indices.ravel()

# Stable numeric cell ID within this versioned grid.
#
# cell_id is 1-based so that zero is never a valid cell ID.

flat_cell_ids = (
    flat_rows.astype(np.int64)
    * np.int64(grid_width)
    + flat_columns.astype(np.int64)
    + np.int64(1)
)


# Cell-centre projected coordinates.
#
# The curated grid has no rotation, but the formula is kept explicit.

flat_x = (
    grid_transform.c
    + (
        flat_columns.astype(np.float64)
        + 0.5
    )
    * grid_transform.a
)

flat_y = (
    grid_transform.f
    + (
        flat_rows.astype(np.float64)
        + 0.5
    )
    * grid_transform.e
)


# -------------------------------------------------------------------
# BUILD EACH HISTORICAL GRID-YEAR PARTITION
# -------------------------------------------------------------------

grid_year_summaries = {}


for epoch in [
    2015,
    2020,
]:
    print("\n" + "=" * 100)
    print("BUILDING GRID-YEAR PARTITION:", epoch)
    print("=" * 100)

    population, _, _, _ = read_single_band(
        curated_paths[
            f"population_{epoch}"
        ]
    )

    built_surface, built_nres, _, _, _ = (
        read_built_bands(
            curated_paths[
                f"built_surface_{epoch}"
            ]
        )
    )

    smod, _, _, _ = read_single_band(
        curated_paths[
            f"smod_{epoch}"
        ]
    )

    # A valid analytical row requires valid values from all three
    # historical products.
    valid_mask = (
        (population != FLOAT_NODATA)
        & (built_surface != BUILT_NODATA)
        & (built_nres != BUILT_NODATA)
        & (smod != SMOD_NODATA)
    )

    flat_valid_mask = valid_mask.ravel()

    valid_population = population[
        valid_mask
    ].astype(
        np.float32,
        copy=False,
    )

    valid_built_surface = built_surface[
        valid_mask
    ].astype(
        np.uint16,
        copy=False,
    )

    valid_built_nres = built_nres[
        valid_mask
    ].astype(
        np.uint16,
        copy=False,
    )

    valid_smod = smod[
        valid_mask
    ].astype(
        np.uint8,
        copy=False,
    )

    built_fraction = (
        valid_built_surface.astype(
            np.float32
        )
        / np.float32(10000.0)
    )

    dataframe = pd.DataFrame(
        {
            "grid_version": GRID_VERSION,
            "cell_id": flat_cell_ids[
                flat_valid_mask
            ],
            "epoch": np.full(
                np.count_nonzero(
                    flat_valid_mask
                ),
                epoch,
                dtype=np.int16,
            ),
            "row_index": flat_rows[
                flat_valid_mask
            ],
            "column_index": flat_columns[
                flat_valid_mask
            ],
            "centroid_x": flat_x[
                flat_valid_mask
            ],
            "centroid_y": flat_y[
                flat_valid_mask
            ],
            "analytical_resolution_m": (
                np.full(
                    np.count_nonzero(
                        flat_valid_mask
                    ),
                    100,
                    dtype=np.int16,
                )
            ),
            "smod_source_resolution_m": (
                np.full(
                    np.count_nonzero(
                        flat_valid_mask
                    ),
                    1000,
                    dtype=np.int16,
                )
            ),
            "smod_code": valid_smod,
            "population_count": (
                valid_population
            ),
            "built_surface": (
                valid_built_surface
            ),
            "built_surface_nres": (
                valid_built_nres
            ),
            "built_fraction": (
                built_fraction
            ),
        }
    )

    dataframe[
        "smod_class_name"
    ] = dataframe[
        "smod_code"
    ].map(
        SMOD_CLASS_NAMES
    ).astype("category")

    dataframe[
        "smod_group"
    ] = dataframe[
        "smod_code"
    ].map(
        SMOD_GROUPS
    ).astype("category")

    # Stable ordering improves reproducibility and later temporal joins.
    dataframe.sort_values(
        by="cell_id",
        inplace=True,
        ignore_index=True,
    )

    output_folder = (
        GRID_YEAR_FOLDER
        / f"epoch={epoch}"
    )

    output_file = (
        output_folder
        / "part-00000.parquet"
    )

    write_parquet_table(
        dataframe,
        output_file,
    )

    duplicate_count = int(
        dataframe.duplicated(
            subset=[
                "cell_id",
                "epoch",
            ]
        ).sum()
    )

    summary = {
        "epoch": epoch,
        "row_count": int(
            len(dataframe)
        ),
        "unique_cell_count": int(
            dataframe[
                "cell_id"
            ].nunique()
        ),
        "duplicate_cell_epoch_count": (
            duplicate_count
        ),
        "population_sum": float(
            dataframe[
                "population_count"
            ].sum()
        ),
        "built_surface_sum_m2": int(
            dataframe[
                "built_surface"
            ].sum()
        ),
        "built_surface_nres_sum_m2": int(
            dataframe[
                "built_surface_nres"
            ].sum()
        ),
        "minimum_cell_id": int(
            dataframe[
                "cell_id"
            ].min()
        ),
        "maximum_cell_id": int(
            dataframe[
                "cell_id"
            ].max()
        ),
        "output_file": str(
            output_file
        ),
        "output_size_bytes": (
            output_file.stat().st_size
        ),
        "sha256": calculate_sha256(
            output_file
        ),
    }

    grid_year_summaries[
        str(epoch)
    ] = summary

    print("Rows:", f"{len(dataframe):,}")
    print(
        "Unique cells:",
        f"{summary['unique_cell_count']:,}",
    )
    print(
        "Duplicate cell-year rows:",
        duplicate_count,
    )
    print(
        "Population total:",
        f"{summary['population_sum']:,.2f}",
    )
    print(
        "Built surface total:",
        f"{summary['built_surface_sum_m2']:,}",
        "square metres",
    )
    print(
        "Output:",
        output_file,
    )
    print(
        "Output size:",
        f"{summary['output_size_bytes'] / (1024 ** 2):.2f}",
        "MiB",
    )

    assert duplicate_count == 0
    assert (
        summary["row_count"]
        == summary["unique_cell_count"]
    )

    del dataframe
    del population
    del built_surface
    del built_nres
    del smod


# -------------------------------------------------------------------
# CREATE CELL-REFERENCE TABLE
# -------------------------------------------------------------------
#
# Use the union of cells valid in either historical epoch.
# Because the masks are currently identical, this should match both
# grid-year partitions.

print("\n" + "=" * 100)
print("BUILDING CELL-REFERENCE TABLE")
print("=" * 100)


population_2015, _, _, _ = read_single_band(
    curated_paths["population_2015"]
)

population_2020, _, _, _ = read_single_band(
    curated_paths["population_2020"]
)

smod_2015, _, _, _ = read_single_band(
    curated_paths["smod_2015"]
)

smod_2020, _, _, _ = read_single_band(
    curated_paths["smod_2020"]
)


cell_reference_mask = (
    (
        population_2015
        != FLOAT_NODATA
    )
    | (
        population_2020
        != FLOAT_NODATA
    )
) & (
    (
        smod_2015
        != SMOD_NODATA
    )
    | (
        smod_2020
        != SMOD_NODATA
    )
)


flat_cell_reference_mask = (
    cell_reference_mask.ravel()
)


cell_reference_dataframe = pd.DataFrame(
    {
        "grid_version": (
            GRID_VERSION
        ),
        "cell_id": flat_cell_ids[
            flat_cell_reference_mask
        ],
        "row_index": flat_rows[
            flat_cell_reference_mask
        ],
        "column_index": flat_columns[
            flat_cell_reference_mask
        ],
        "centroid_x": flat_x[
            flat_cell_reference_mask
        ],
        "centroid_y": flat_y[
            flat_cell_reference_mask
        ],
        "cell_area_m2": np.full(
            np.count_nonzero(
                flat_cell_reference_mask
            ),
            10000,
            dtype=np.int32,
        ),
        "analytical_resolution_m": (
            np.full(
                np.count_nonzero(
                    flat_cell_reference_mask
                ),
                100,
                dtype=np.int16,
            )
        ),
    }
)


cell_reference_dataframe.sort_values(
    by="cell_id",
    inplace=True,
    ignore_index=True,
)


cell_reference_output = (
    CELL_REFERENCE_FOLDER
    / "part-00000.parquet"
)


write_parquet_table(
    cell_reference_dataframe,
    cell_reference_output,
)


cell_reference_duplicate_count = int(
    cell_reference_dataframe.duplicated(
        subset=[
            "cell_id",
        ]
    ).sum()
)


assert cell_reference_duplicate_count == 0


print(
    "Rows:",
    f"{len(cell_reference_dataframe):,}",
)

print(
    "Duplicate cell IDs:",
    cell_reference_duplicate_count,
)

print(
    "Output:",
    cell_reference_output,
)

print(
    "Output size:",
    f"{cell_reference_output.stat().st_size / (1024 ** 2):.2f}",
    "MiB",
)


# -------------------------------------------------------------------
# CREATE 2018 BUILDING-HEIGHT TABLE
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("BUILDING 2018 HEIGHT TABLE")
print("=" * 100)


height_2018, _, _, _ = read_single_band(
    curated_paths[
        "building_height_2018"
    ]
)


height_valid_mask = (
    (height_2018 != FLOAT_NODATA)
    & cell_reference_mask
)


flat_height_valid_mask = (
    height_valid_mask.ravel()
)


height_dataframe = pd.DataFrame(
    {
        "grid_version": (
            GRID_VERSION
        ),
        "cell_id": flat_cell_ids[
            flat_height_valid_mask
        ],
        "reference_year": np.full(
            np.count_nonzero(
                flat_height_valid_mask
            ),
            2018,
            dtype=np.int16,
        ),
        "built_height_m": height_2018[
            height_valid_mask
        ].astype(
            np.float32,
            copy=False,
        ),
        "source_resolution_m": (
            np.full(
                np.count_nonzero(
                    flat_height_valid_mask
                ),
                100,
                dtype=np.int16,
            )
        ),
        "temporal_policy": (
            "cross_sectional_only"
        ),
    }
)


height_dataframe.sort_values(
    by="cell_id",
    inplace=True,
    ignore_index=True,
)


height_output = (
    HEIGHT_FOLDER
    / "part-00000.parquet"
)


write_parquet_table(
    height_dataframe,
    height_output,
)


height_duplicate_count = int(
    height_dataframe.duplicated(
        subset=[
            "cell_id",
            "reference_year",
        ]
    ).sum()
)


assert height_duplicate_count == 0


print(
    "Rows:",
    f"{len(height_dataframe):,}",
)

print(
    "Positive-height rows:",
    f"{np.count_nonzero(height_dataframe['built_height_m'] > 0):,}",
)

print(
    "Zero-height rows:",
    f"{np.count_nonzero(height_dataframe['built_height_m'] == 0):,}",
)

print(
    "Output:",
    height_output,
)

print(
    "Output size:",
    f"{height_output.stat().st_size / (1024 ** 2):.2f}",
    "MiB",
)


# -------------------------------------------------------------------
# VALIDATE TEMPORAL CELL OVERLAP
# -------------------------------------------------------------------

table_2015 = pq.read_table(
    GRID_YEAR_FOLDER
    / "epoch=2015"
    / "part-00000.parquet",
    columns=[
        "cell_id",
    ],
)

table_2020 = pq.read_table(
    GRID_YEAR_FOLDER
    / "epoch=2020"
    / "part-00000.parquet",
    columns=[
        "cell_id",
    ],
)


cells_2015 = set(
    table_2015[
        "cell_id"
    ].to_pylist()
)

cells_2020 = set(
    table_2020[
        "cell_id"
    ].to_pylist()
)


common_cell_count = len(
    cells_2015
    & cells_2020
)

only_2015_count = len(
    cells_2015
    - cells_2020
)

only_2020_count = len(
    cells_2020
    - cells_2015
)


print("\n" + "=" * 100)
print("TEMPORAL CELL OVERLAP")
print("=" * 100)

print(
    "Cells in both epochs:",
    f"{common_cell_count:,}",
)

print(
    "Cells only in 2015:",
    f"{only_2015_count:,}",
)

print(
    "Cells only in 2020:",
    f"{only_2020_count:,}",
)


# -------------------------------------------------------------------
# SAVE PARQUET MANIFEST
# -------------------------------------------------------------------

parquet_manifest = {
    "status": "success",
    "grid_version": (
        GRID_VERSION
    ),
    "grid": {
        "width": grid_width,
        "height": grid_height,
        "crs": str(grid_crs),
        "transform": [
            float(grid_transform.a),
            float(grid_transform.b),
            float(grid_transform.c),
            float(grid_transform.d),
            float(grid_transform.e),
            float(grid_transform.f),
        ],
        "bounds": [
            float(grid_bounds.left),
            float(grid_bounds.bottom),
            float(grid_bounds.right),
            float(grid_bounds.top),
        ],
        "analytical_resolution_m": 100,
        "cell_area_m2": 10000,
    },
    "grid_year": (
        grid_year_summaries
    ),
    "cell_reference": {
        "row_count": int(
            len(
                cell_reference_dataframe
            )
        ),
        "duplicate_cell_id_count": (
            cell_reference_duplicate_count
        ),
        "output_file": str(
            cell_reference_output
        ),
        "output_size_bytes": (
            cell_reference_output
            .stat()
            .st_size
        ),
        "sha256": calculate_sha256(
            cell_reference_output
        ),
    },
    "building_height_2018": {
        "row_count": int(
            len(height_dataframe)
        ),
        "duplicate_cell_year_count": (
            height_duplicate_count
        ),
        "output_file": str(
            height_output
        ),
        "output_size_bytes": (
            height_output
            .stat()
            .st_size
        ),
        "sha256": calculate_sha256(
            height_output
        ),
        "temporal_policy": (
            "cross_sectional_only"
        ),
    },
    "temporal_overlap": {
        "common_cells": (
            common_cell_count
        ),
        "only_2015": (
            only_2015_count
        ),
        "only_2020": (
            only_2020_count
        ),
    },
}


manifest_path = (
    PARQUET_ROOT
    / "bengaluru_pilot_parquet_manifest.json"
)


with manifest_path.open(
    "w",
    encoding="utf-8",
) as manifest_file:
    json.dump(
        parquet_manifest,
        manifest_file,
        indent=2,
    )


# -------------------------------------------------------------------
# FINAL RESULT
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("FINAL PARQUET RESULT")
print("=" * 100)

print(
    "SUCCESS: The curated 100 m rasters were converted "
    "to Spark-compatible Parquet tables."
)

print(
    "The historical grid-year table is partitioned by epoch."
)

print(
    "Building height remains a separate 2018 "
    "cross-sectional table."
)

print(
    "Stable cell IDs were derived from the grid row and column."
)

print(
    "No geometry objects or repeated text coordinates were "
    "stored in every historical partition."
)

print("\nParquet root:")
print(PARQUET_ROOT)

print("\nManifest:")
print(manifest_path)
```

```python
# -------------------------------------------------------------------
# CELL 18B: REPAIR HIVE PARTITION STRUCTURE AND FINISH VALIDATION
# -------------------------------------------------------------------
#
# Cell 18 successfully created the Parquet outputs but failed during
# the final read because "epoch" existed in two places:
#
# 1. Inside each physical Parquet file.
# 2. In the Hive-style directory path, such as epoch=2015.
#
# Correct layout:
# - Keep epoch in the directory path.
# - Remove epoch from each physical Parquet file.
# - Let Spark/PyArrow reconstruct epoch from the directory path.
#
# This cell:
# 1. Creates a backup of each original grid-year Parquet file.
# 2. Removes the internal epoch column.
# 3. Validates Hive partition discovery.
# 4. Checks temporal cell overlap.
# 5. Checks duplicate cell-epoch keys.
# 6. Validates the supporting Parquet tables.
# 7. Writes the final Parquet manifest.
#
# Do not rerun Cell 18 before running this cell.

from pathlib import Path
import hashlib
import json
import shutil

import numpy as np
import pyarrow.dataset as ds
import pyarrow.parquet as pq


# -------------------------------------------------------------------
# PATHS
# -------------------------------------------------------------------

PARQUET_ROOT = Path(
    "/content/drive/MyDrive/"
    "GHSL_Bengaluru_Pilot/parquet"
)

GRID_YEAR_FOLDER = (
    PARQUET_ROOT
    / "grid_year"
)

CELL_REFERENCE_FOLDER = (
    PARQUET_ROOT
    / "cell_reference_100m"
)

HEIGHT_FOLDER = (
    PARQUET_ROOT
    / "building_height_2018"
)


GRID_YEAR_FILES = {
    2015: (
        GRID_YEAR_FOLDER
        / "epoch=2015"
        / "part-00000.parquet"
    ),
    2020: (
        GRID_YEAR_FOLDER
        / "epoch=2020"
        / "part-00000.parquet"
    ),
}


CELL_REFERENCE_FILE = (
    CELL_REFERENCE_FOLDER
    / "part-00000.parquet"
)

HEIGHT_FILE = (
    HEIGHT_FOLDER
    / "part-00000.parquet"
)


# Expected record count produced by Cell 18.
EXPECTED_CELL_COUNT_PER_EPOCH = 1_451_078


# -------------------------------------------------------------------
# HELPER FUNCTIONS
# -------------------------------------------------------------------

def calculate_sha256(
    file_path,
    chunk_size=1024 * 1024,
):
    """
    Calculate the SHA-256 checksum of a file.
    """

    sha256 = hashlib.sha256()

    with file_path.open("rb") as file_handle:
        while True:
            chunk = file_handle.read(
                chunk_size
            )

            if not chunk:
                break

            sha256.update(chunk)

    return sha256.hexdigest()


def get_physical_parquet_schema(
    file_path,
):
    """
    Read only the schema physically stored inside a Parquet file.

    Using ParquetFile avoids inferring the epoch value from the
    Hive-style parent directory.
    """

    return (
        pq.ParquetFile(file_path)
        .schema_arrow
    )


def get_physical_parquet_row_count(
    file_path,
):
    """
    Return the number of rows physically stored in one Parquet file.
    """

    return int(
        pq.ParquetFile(file_path)
        .metadata
        .num_rows
    )


# -------------------------------------------------------------------
# VERIFY REQUIRED INPUT FILES
# -------------------------------------------------------------------

required_files = [
    *GRID_YEAR_FILES.values(),
    CELL_REFERENCE_FILE,
    HEIGHT_FILE,
]


missing_files = [
    str(file_path)
    for file_path in required_files
    if not file_path.exists()
]


if missing_files:
    raise FileNotFoundError(
        "The following required Parquet files are missing:\n"
        + "\n".join(missing_files)
    )


# -------------------------------------------------------------------
# REPAIR GRID-YEAR PARTITION FILES
# -------------------------------------------------------------------

print("=" * 100)
print("REPAIRING GRID-YEAR HIVE PARTITION STRUCTURE")
print("=" * 100)


partition_repair_results = {}


for epoch, file_path in GRID_YEAR_FILES.items():
    print("\n" + "-" * 100)
    print("Epoch:", epoch)
    print("File:", file_path)

    original_schema = (
        get_physical_parquet_schema(
            file_path
        )
    )

    original_columns = list(
        original_schema.names
    )

    original_row_count = (
        get_physical_parquet_row_count(
            file_path
        )
    )

    print(
        "Original physical columns:",
        original_columns,
    )

    print(
        "Original row count:",
        f"{original_row_count:,}",
    )

    backup_path = Path(
        str(file_path)
        + ".pre_partition_fix_backup"
    )

    temporary_path = Path(
        str(file_path)
        + ".tmp"
    )

    if temporary_path.exists():
        temporary_path.unlink()

    if "epoch" in original_columns:
        physical_table = (
            pq.ParquetFile(file_path)
            .read()
        )

        repaired_table = (
            physical_table.drop(
                ["epoch"]
            )
        )

        if not backup_path.exists():
            shutil.copy2(
                file_path,
                backup_path,
            )

        pq.write_table(
            repaired_table,
            temporary_path,
            compression="snappy",
            use_dictionary=True,
            write_statistics=True,
        )

        temporary_schema = (
            get_physical_parquet_schema(
                temporary_path
            )
        )

        temporary_row_count = (
            get_physical_parquet_row_count(
                temporary_path
            )
        )

        if "epoch" in temporary_schema.names:
            raise RuntimeError(
                "The temporary repaired file still contains "
                f"an internal epoch column: {temporary_path}"
            )

        if temporary_row_count != original_row_count:
            raise RuntimeError(
                f"Row count changed while repairing epoch {epoch}. "
                f"Before: {original_row_count:,}; "
                f"after: {temporary_row_count:,}"
            )

        temporary_path.replace(
            file_path
        )

        action = (
            "Internal epoch column removed"
        )

    else:
        action = (
            "No repair required because the internal "
            "epoch column was already absent"
        )

    repaired_schema = (
        get_physical_parquet_schema(
            file_path
        )
    )

    repaired_row_count = (
        get_physical_parquet_row_count(
            file_path
        )
    )

    print(
        "Action:",
        action,
    )

    print(
        "Repaired physical columns:",
        list(repaired_schema.names),
    )

    print(
        "Repaired row count:",
        f"{repaired_row_count:,}",
    )

    print(
        "Backup:",
        str(backup_path)
        if backup_path.exists()
        else "Not required",
    )

    assert "epoch" not in repaired_schema.names, (
        f"Internal epoch column remains in epoch {epoch}."
    )

    assert repaired_row_count == original_row_count, (
        f"Row count changed for epoch {epoch}."
    )

    assert (
        repaired_row_count
        == EXPECTED_CELL_COUNT_PER_EPOCH
    ), (
        f"Unexpected row count for epoch {epoch}: "
        f"{repaired_row_count:,}"
    )

    partition_repair_results[
        str(epoch)
    ] = {
        "epoch": int(epoch),
        "file": str(file_path),
        "backup": (
            str(backup_path)
            if backup_path.exists()
            else None
        ),
        "action": action,
        "row_count": int(
            repaired_row_count
        ),
        "physical_columns": list(
            repaired_schema.names
        ),
        "file_size_bytes": int(
            file_path.stat().st_size
        ),
        "sha256": calculate_sha256(
            file_path
        ),
    }


# -------------------------------------------------------------------
# VALIDATE HIVE PARTITION DISCOVERY
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("HIVE PARTITION VALIDATION")
print("=" * 100)


grid_year_dataset = ds.dataset(
    GRID_YEAR_FOLDER,
    format="parquet",
    partitioning="hive",
    ignore_prefixes=[
        ".",
        "_",
    ],
)


dataset_schema = (
    grid_year_dataset.schema
)


print("\nCombined dataset schema:")
print(dataset_schema)


assert "epoch" in dataset_schema.names, (
    "Hive partition discovery did not reconstruct "
    "the epoch column."
)


epoch_field = (
    dataset_schema.field(
        "epoch"
    )
)


print("\nDiscovered epoch field:")
print(epoch_field)


grid_year_index_table = (
    grid_year_dataset.to_table(
        columns=[
            "cell_id",
            "epoch",
        ]
    )
)


combined_row_count = int(
    grid_year_index_table.num_rows
)


print("\nCombined grid-year rows:")
print(
    f"{combined_row_count:,}"
)


assert combined_row_count == (
    EXPECTED_CELL_COUNT_PER_EPOCH
    * 2
), (
    "Unexpected combined row count."
)


# -------------------------------------------------------------------
# VALIDATE PARTITION VALUES
# -------------------------------------------------------------------

epoch_array = (
    grid_year_index_table[
        "epoch"
    ].to_numpy(
        zero_copy_only=False
    )
)


observed_epochs = sorted(
    {
        int(value)
        for value in epoch_array
        if value is not None
    }
)


print("\nObserved epochs:")
print(observed_epochs)


assert observed_epochs == [
    2015,
    2020,
], (
    f"Unexpected epoch values: {observed_epochs}"
)


# -------------------------------------------------------------------
# TEMPORAL CELL OVERLAP
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("TEMPORAL CELL OVERLAP")
print("=" * 100)


cell_id_array = (
    grid_year_index_table[
        "cell_id"
    ].to_numpy(
        zero_copy_only=False
    )
)


cells_2015 = set(
    cell_id_array[
        epoch_array == 2015
    ].tolist()
)


cells_2020 = set(
    cell_id_array[
        epoch_array == 2020
    ].tolist()
)


common_cells = (
    cells_2015
    & cells_2020
)


only_2015_cells = (
    cells_2015
    - cells_2020
)


only_2020_cells = (
    cells_2020
    - cells_2015
)


print(
    "2015 cells:",
    f"{len(cells_2015):,}",
)

print(
    "2020 cells:",
    f"{len(cells_2020):,}",
)

print(
    "Cells in both epochs:",
    f"{len(common_cells):,}",
)

print(
    "Cells only in 2015:",
    f"{len(only_2015_cells):,}",
)

print(
    "Cells only in 2020:",
    f"{len(only_2020_cells):,}",
)


assert len(cells_2015) == (
    EXPECTED_CELL_COUNT_PER_EPOCH
)

assert len(cells_2020) == (
    EXPECTED_CELL_COUNT_PER_EPOCH
)

assert len(common_cells) == (
    EXPECTED_CELL_COUNT_PER_EPOCH
)

assert len(only_2015_cells) == 0

assert len(only_2020_cells) == 0


# -------------------------------------------------------------------
# VALIDATE DUPLICATE CELL-EPOCH KEYS
# -------------------------------------------------------------------

encoded_keys = np.core.records.fromarrays(
    [
        cell_id_array.astype(
            np.int64,
            copy=False,
        ),
        epoch_array.astype(
            np.int32,
            copy=False,
        ),
    ],
    names="cell_id,epoch",
)


unique_key_count = int(
    np.unique(encoded_keys).size
)


duplicate_cell_epoch_count = (
    combined_row_count
    - unique_key_count
)


print(
    "\nDuplicate cell-epoch keys:",
    f"{duplicate_cell_epoch_count:,}",
)


assert duplicate_cell_epoch_count == 0, (
    "Duplicate cell-epoch keys were found."
)


# -------------------------------------------------------------------
# VALIDATE SUPPORTING PARQUET TABLES
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("SUPPORTING PARQUET VALIDATION")
print("=" * 100)


cell_reference_parquet = (
    pq.ParquetFile(
        CELL_REFERENCE_FILE
    )
)


height_parquet = (
    pq.ParquetFile(
        HEIGHT_FILE
    )
)


cell_reference_row_count = int(
    cell_reference_parquet
    .metadata
    .num_rows
)


height_row_count = int(
    height_parquet
    .metadata
    .num_rows
)


print(
    "Cell-reference rows:",
    f"{cell_reference_row_count:,}",
)

print(
    "Building-height rows:",
    f"{height_row_count:,}",
)


print("\nCell-reference schema:")
print(
    cell_reference_parquet
    .schema_arrow
)


print("\nBuilding-height schema:")
print(
    height_parquet
    .schema_arrow
)


assert cell_reference_row_count == (
    EXPECTED_CELL_COUNT_PER_EPOCH
), (
    "Unexpected cell-reference row count."
)


assert height_row_count == (
    EXPECTED_CELL_COUNT_PER_EPOCH
), (
    "Unexpected building-height row count."
)


# -------------------------------------------------------------------
# VERIFY CELL-REFERENCE UNIQUENESS
# -------------------------------------------------------------------

cell_reference_ids = (
    cell_reference_parquet
    .read(
        columns=[
            "cell_id",
        ]
    )[
        "cell_id"
    ]
    .to_numpy(
        zero_copy_only=False
    )
)


cell_reference_duplicate_count = (
    len(cell_reference_ids)
    - len(np.unique(cell_reference_ids))
)


print(
    "\nDuplicate cell IDs in cell-reference table:",
    f"{cell_reference_duplicate_count:,}",
)


assert cell_reference_duplicate_count == 0


# -------------------------------------------------------------------
# VERIFY HEIGHT TABLE UNIQUENESS AND YEAR
# -------------------------------------------------------------------

height_index_table = (
    height_parquet.read(
        columns=[
            "cell_id",
            "reference_year",
        ]
    )
)


height_cell_ids = (
    height_index_table[
        "cell_id"
    ].to_numpy(
        zero_copy_only=False
    )
)


height_years = (
    height_index_table[
        "reference_year"
    ].to_numpy(
        zero_copy_only=False
    )
)


height_unique_years = sorted(
    {
        int(value)
        for value in height_years
    }
)


height_duplicate_count = (
    len(height_cell_ids)
    - len(np.unique(height_cell_ids))
)


print(
    "Building-height reference years:",
    height_unique_years,
)

print(
    "Duplicate building-height cell IDs:",
    f"{height_duplicate_count:,}",
)


assert height_unique_years == [
    2018,
]

assert height_duplicate_count == 0


# -------------------------------------------------------------------
# WRITE FINAL PARQUET MANIFEST
# -------------------------------------------------------------------

manifest_path = (
    PARQUET_ROOT
    / "bengaluru_pilot_parquet_manifest.json"
)


manifest = {
    "status": "success",
    "partitioning": {
        "format": "hive",
        "partition_column": "epoch",
        "partition_values": [
            2015,
            2020,
        ],
        "epoch_stored_inside_physical_files": False,
        "epoch_reconstructed_from_directory_path": True,
    },
    "grid_year": {
        "combined_row_count": (
            combined_row_count
        ),
        "observed_epochs": (
            observed_epochs
        ),
        "duplicate_cell_epoch_count": int(
            duplicate_cell_epoch_count
        ),
        "partitions": (
            partition_repair_results
        ),
    },
    "temporal_overlap": {
        "cells_2015": int(
            len(cells_2015)
        ),
        "cells_2020": int(
            len(cells_2020)
        ),
        "common_cells": int(
            len(common_cells)
        ),
        "only_2015": int(
            len(only_2015_cells)
        ),
        "only_2020": int(
            len(only_2020_cells)
        ),
    },
    "cell_reference": {
        "file": str(
            CELL_REFERENCE_FILE
        ),
        "row_count": int(
            cell_reference_row_count
        ),
        "duplicate_cell_id_count": int(
            cell_reference_duplicate_count
        ),
        "file_size_bytes": int(
            CELL_REFERENCE_FILE
            .stat()
            .st_size
        ),
        "sha256": calculate_sha256(
            CELL_REFERENCE_FILE
        ),
    },
    "building_height_2018": {
        "file": str(
            HEIGHT_FILE
        ),
        "row_count": int(
            height_row_count
        ),
        "reference_years": (
            height_unique_years
        ),
        "duplicate_cell_id_count": int(
            height_duplicate_count
        ),
        "file_size_bytes": int(
            HEIGHT_FILE
            .stat()
            .st_size
        ),
        "sha256": calculate_sha256(
            HEIGHT_FILE
        ),
        "temporal_policy": (
            "cross_sectional_only"
        ),
    },
}


with manifest_path.open(
    "w",
    encoding="utf-8",
) as manifest_file:
    json.dump(
        manifest,
        manifest_file,
        indent=2,
    )


# -------------------------------------------------------------------
# FINAL RESULT
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("FINAL PARQUET RESULT")
print("=" * 100)

print(
    "SUCCESS: The Hive partition-column conflict was corrected."
)

print(
    "The physical grid-year Parquet files no longer contain "
    "an internal epoch column."
)

print(
    "The epoch field is reconstructed correctly from the "
    "epoch=2015 and epoch=2020 directory paths."
)

print(
    "Both epochs contain the same 1,451,078 stable cells."
)

print(
    "There are no duplicate cell-epoch keys."
)

print(
    "The cell-reference and 2018 building-height tables "
    "passed their row-count and uniqueness checks."
)

print(
    "The Parquet dataset is ready for HDFS and Spark."
)

print("\nParquet root:")
print(PARQUET_ROOT)

print("\nManifest:")
print(manifest_path)

print(
    "\nImportant: Do not upload files ending in "
    "'.pre_partition_fix_backup' to HDFS."
)
```

```python
# -------------------------------------------------------------------
# CELL 18C: MOVE BACKUPS OUTSIDE GRID_YEAR AND FINISH VALIDATION
# -------------------------------------------------------------------

from pathlib import Path
import hashlib
import json
import shutil

import numpy as np
import pyarrow.dataset as ds
import pyarrow.parquet as pq


# -------------------------------------------------------------------
# PATHS
# -------------------------------------------------------------------

PARQUET_ROOT = Path(
    "/content/drive/MyDrive/GHSL_Bengaluru_Pilot/parquet"
)

GRID_YEAR_FOLDER = (
    PARQUET_ROOT
    / "grid_year"
)

CELL_REFERENCE_FILE = (
    PARQUET_ROOT
    / "cell_reference_100m"
    / "part-00000.parquet"
)

HEIGHT_FILE = (
    PARQUET_ROOT
    / "building_height_2018"
    / "part-00000.parquet"
)

BACKUP_FOLDER = (
    PARQUET_ROOT
    / "_repair_backups"
)

BACKUP_FOLDER.mkdir(
    parents=True,
    exist_ok=True,
)


GRID_YEAR_FILES = {
    2015: (
        GRID_YEAR_FOLDER
        / "epoch=2015"
        / "part-00000.parquet"
    ),
    2020: (
        GRID_YEAR_FOLDER
        / "epoch=2020"
        / "part-00000.parquet"
    ),
}


EXPECTED_CELL_COUNT_PER_EPOCH = 1_451_078

EXPECTED_COMBINED_ROW_COUNT = (
    EXPECTED_CELL_COUNT_PER_EPOCH * 2
)


# -------------------------------------------------------------------
# HELPER FUNCTIONS
# -------------------------------------------------------------------

def calculate_sha256(
    file_path,
    chunk_size=1024 * 1024,
):
    sha256 = hashlib.sha256()

    with file_path.open("rb") as file_handle:
        while True:
            chunk = file_handle.read(
                chunk_size
            )

            if not chunk:
                break

            sha256.update(chunk)

    return sha256.hexdigest()


def physical_schema(file_path):
    return (
        pq.ParquetFile(file_path)
        .schema_arrow
    )


def physical_row_count(file_path):
    return int(
        pq.ParquetFile(file_path)
        .metadata
        .num_rows
    )


# -------------------------------------------------------------------
# VERIFY REPAIRED PRODUCTION FILES
# -------------------------------------------------------------------

print("=" * 100)
print("VERIFYING REPAIRED PRODUCTION FILES")
print("=" * 100)


for epoch, file_path in GRID_YEAR_FILES.items():
    if not file_path.exists():
        raise FileNotFoundError(
            f"Missing production file for epoch {epoch}: "
            f"{file_path}"
        )

    schema = physical_schema(
        file_path
    )

    row_count = physical_row_count(
        file_path
    )

    print(f"\nEpoch {epoch}")
    print("File:", file_path)
    print("Rows:", f"{row_count:,}")
    print("Physical columns:", schema.names)

    assert "epoch" not in schema.names, (
        f"The production file for epoch {epoch} still "
        "contains an internal epoch column."
    )

    assert row_count == EXPECTED_CELL_COUNT_PER_EPOCH, (
        f"Unexpected row count for epoch {epoch}. "
        f"Expected {EXPECTED_CELL_COUNT_PER_EPOCH:,}, "
        f"found {row_count:,}."
    )


# -------------------------------------------------------------------
# FIND AND MOVE BACKUP FILES
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("MOVING BACKUPS OUTSIDE THE HIVE DATASET")
print("=" * 100)


backup_candidates = sorted(
    path
    for path in GRID_YEAR_FOLDER.rglob("*")
    if (
        path.is_file()
        and "pre_partition_fix_backup" in path.name
    )
)


print(
    "\nBackup files found inside grid_year:",
    len(backup_candidates),
)


moved_backups = []


for source_path in backup_candidates:
    epoch_directory_name = (
        source_path.parent.name
    )

    destination_name = (
        f"{epoch_directory_name}_"
        f"{source_path.name}"
    )

    destination_path = (
        BACKUP_FOLDER
        / destination_name
    )

    source_checksum = calculate_sha256(
        source_path
    )

    if destination_path.exists():
        destination_checksum = calculate_sha256(
            destination_path
        )

        if destination_checksum != source_checksum:
            raise FileExistsError(
                "A different backup file already exists at:\n"
                f"{destination_path}"
            )

        source_path.unlink()

        action = (
            "Identical backup already existed in the "
            "backup folder; source duplicate removed"
        )

    else:
        shutil.move(
            str(source_path),
            str(destination_path),
        )

        action = (
            "Backup moved outside grid_year"
        )

    print("\nSource:", source_path)
    print("Destination:", destination_path)
    print("Action:", action)

    moved_backups.append(
        {
            "source": str(source_path),
            "destination": str(
                destination_path
            ),
            "sha256": source_checksum,
            "action": action,
        }
    )


# -------------------------------------------------------------------
# CONFIRM NO BACKUPS REMAIN INSIDE GRID_YEAR
# -------------------------------------------------------------------

remaining_backup_files = sorted(
    path
    for path in GRID_YEAR_FOLDER.rglob("*")
    if (
        path.is_file()
        and "pre_partition_fix_backup" in path.name
    )
)


print(
    "\nBackup files remaining inside grid_year:",
    len(remaining_backup_files),
)


assert not remaining_backup_files, (
    "One or more backup files remain inside the "
    "Hive-partitioned grid_year dataset."
)


# -------------------------------------------------------------------
# VERIFY GRID_YEAR FILE INVENTORY
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("GRID-YEAR FILE INVENTORY")
print("=" * 100)


all_grid_year_files = sorted(
    path
    for path in GRID_YEAR_FOLDER.rglob("*")
    if path.is_file()
)


for file_path in all_grid_year_files:
    print(file_path)


production_parquet_files = [
    path
    for path in all_grid_year_files
    if path.suffix.lower() == ".parquet"
]


print(
    "\nProduction Parquet files:",
    len(production_parquet_files),
)


assert len(production_parquet_files) == 2, (
    "The grid_year dataset should contain exactly "
    "two production Parquet files."
)


expected_production_files = set(
    GRID_YEAR_FILES.values()
)

actual_production_files = set(
    production_parquet_files
)


assert actual_production_files == expected_production_files, (
    "Unexpected production files were found inside grid_year."
)


# -------------------------------------------------------------------
# READ GRID_YEAR AS A HIVE-PARTITIONED DATASET
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("HIVE PARTITION VALIDATION")
print("=" * 100)


grid_year_dataset = ds.dataset(
    GRID_YEAR_FOLDER,
    format="parquet",
    partitioning="hive",
    ignore_prefixes=[
        ".",
        "_",
    ],
)


print("\nDataset schema:")
print(grid_year_dataset.schema)


assert "epoch" in grid_year_dataset.schema.names, (
    "Hive partition discovery did not reconstruct "
    "the epoch column."
)


epoch_field = (
    grid_year_dataset.schema.field(
        "epoch"
    )
)


print("\nDiscovered epoch field:")
print(epoch_field)


grid_year_index_table = (
    grid_year_dataset.to_table(
        columns=[
            "cell_id",
            "epoch",
        ]
    )
)


combined_row_count = int(
    grid_year_index_table.num_rows
)


print("\nCombined grid-year rows:")
print(f"{combined_row_count:,}")


assert combined_row_count == (
    EXPECTED_COMBINED_ROW_COUNT
), (
    "Unexpected combined row count. "
    f"Expected {EXPECTED_COMBINED_ROW_COUNT:,}, "
    f"found {combined_row_count:,}."
)


# -------------------------------------------------------------------
# VALIDATE EPOCH VALUES
# -------------------------------------------------------------------

cell_id_array = (
    grid_year_index_table[
        "cell_id"
    ].to_numpy(
        zero_copy_only=False
    )
)


epoch_array = (
    grid_year_index_table[
        "epoch"
    ].to_numpy(
        zero_copy_only=False
    )
)


observed_epochs = sorted(
    {
        int(value)
        for value in epoch_array
        if value is not None
    }
)


print("\nObserved epochs:")
print(observed_epochs)


assert observed_epochs == [
    2015,
    2020,
], (
    f"Unexpected epoch values: {observed_epochs}"
)


# -------------------------------------------------------------------
# TEMPORAL CELL OVERLAP
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("TEMPORAL CELL OVERLAP")
print("=" * 100)


cells_2015 = np.unique(
    cell_id_array[
        epoch_array == 2015
    ]
)


cells_2020 = np.unique(
    cell_id_array[
        epoch_array == 2020
    ]
)


common_cells = np.intersect1d(
    cells_2015,
    cells_2020,
    assume_unique=True,
)


only_2015_cells = np.setdiff1d(
    cells_2015,
    cells_2020,
    assume_unique=True,
)


only_2020_cells = np.setdiff1d(
    cells_2020,
    cells_2015,
    assume_unique=True,
)


print(
    "2015 cells:",
    f"{cells_2015.size:,}",
)

print(
    "2020 cells:",
    f"{cells_2020.size:,}",
)

print(
    "Cells in both epochs:",
    f"{common_cells.size:,}",
)

print(
    "Cells only in 2015:",
    f"{only_2015_cells.size:,}",
)

print(
    "Cells only in 2020:",
    f"{only_2020_cells.size:,}",
)


assert cells_2015.size == (
    EXPECTED_CELL_COUNT_PER_EPOCH
)

assert cells_2020.size == (
    EXPECTED_CELL_COUNT_PER_EPOCH
)

assert common_cells.size == (
    EXPECTED_CELL_COUNT_PER_EPOCH
)

assert only_2015_cells.size == 0

assert only_2020_cells.size == 0


# -------------------------------------------------------------------
# VALIDATE CELL-EPOCH UNIQUENESS
# -------------------------------------------------------------------

structured_keys = np.empty(
    combined_row_count,
    dtype=[
        ("cell_id", np.int64),
        ("epoch", np.int32),
    ],
)


structured_keys[
    "cell_id"
] = cell_id_array.astype(
    np.int64,
    copy=False,
)


structured_keys[
    "epoch"
] = epoch_array.astype(
    np.int32,
    copy=False,
)


unique_key_count = int(
    np.unique(
        structured_keys
    ).size
)


duplicate_cell_epoch_count = (
    combined_row_count
    - unique_key_count
)


print(
    "\nDuplicate cell-epoch keys:",
    f"{duplicate_cell_epoch_count:,}",
)


assert duplicate_cell_epoch_count == 0, (
    "Duplicate cell-epoch records were found."
)


# -------------------------------------------------------------------
# VALIDATE SUPPORTING PARQUET TABLES
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("SUPPORTING PARQUET VALIDATION")
print("=" * 100)


if not CELL_REFERENCE_FILE.exists():
    raise FileNotFoundError(
        f"Missing cell-reference file: "
        f"{CELL_REFERENCE_FILE}"
    )


if not HEIGHT_FILE.exists():
    raise FileNotFoundError(
        f"Missing building-height file: "
        f"{HEIGHT_FILE}"
    )


cell_reference_parquet = (
    pq.ParquetFile(
        CELL_REFERENCE_FILE
    )
)


height_parquet = (
    pq.ParquetFile(
        HEIGHT_FILE
    )
)


cell_reference_row_count = int(
    cell_reference_parquet
    .metadata
    .num_rows
)


height_row_count = int(
    height_parquet
    .metadata
    .num_rows
)


print(
    "Cell-reference rows:",
    f"{cell_reference_row_count:,}",
)

print(
    "Building-height rows:",
    f"{height_row_count:,}",
)


assert cell_reference_row_count == (
    EXPECTED_CELL_COUNT_PER_EPOCH
)

assert height_row_count == (
    EXPECTED_CELL_COUNT_PER_EPOCH
)


# -------------------------------------------------------------------
# VALIDATE CELL-REFERENCE UNIQUENESS
# -------------------------------------------------------------------

cell_reference_ids = (
    cell_reference_parquet.read(
        columns=[
            "cell_id",
        ]
    )[
        "cell_id"
    ].to_numpy(
        zero_copy_only=False
    )
)


cell_reference_duplicate_count = (
    cell_reference_ids.size
    - np.unique(
        cell_reference_ids
    ).size
)


print(
    "Duplicate cell-reference IDs:",
    f"{cell_reference_duplicate_count:,}",
)


assert cell_reference_duplicate_count == 0


# -------------------------------------------------------------------
# VALIDATE BUILDING-HEIGHT TABLE
# -------------------------------------------------------------------

height_index_table = (
    height_parquet.read(
        columns=[
            "cell_id",
            "reference_year",
        ]
    )
)


height_cell_ids = (
    height_index_table[
        "cell_id"
    ].to_numpy(
        zero_copy_only=False
    )
)


height_years = (
    height_index_table[
        "reference_year"
    ].to_numpy(
        zero_copy_only=False
    )
)


height_duplicate_count = (
    height_cell_ids.size
    - np.unique(
        height_cell_ids
    ).size
)


height_reference_years = sorted(
    {
        int(value)
        for value in height_years
    }
)


print(
    "Duplicate building-height cell IDs:",
    f"{height_duplicate_count:,}",
)

print(
    "Building-height reference years:",
    height_reference_years,
)


assert height_duplicate_count == 0

assert height_reference_years == [
    2018,
]


# -------------------------------------------------------------------
# BUILD FINAL MANIFEST
# -------------------------------------------------------------------

partition_files = {}


for epoch, file_path in GRID_YEAR_FILES.items():
    partition_files[
        str(epoch)
    ] = {
        "epoch": int(epoch),
        "file": str(file_path),
        "row_count": int(
            physical_row_count(
                file_path
            )
        ),
        "physical_columns": list(
            physical_schema(
                file_path
            ).names
        ),
        "file_size_bytes": int(
            file_path.stat().st_size
        ),
        "sha256": calculate_sha256(
            file_path
        ),
    }


manifest_path = (
    PARQUET_ROOT
    / "bengaluru_pilot_parquet_manifest.json"
)


manifest = {
    "status": "success",
    "partitioning": {
        "format": "hive",
        "partition_column": "epoch",
        "partition_values": [
            2015,
            2020,
        ],
        "epoch_stored_inside_physical_files": False,
        "epoch_reconstructed_from_directory_path": True,
    },
    "backup_policy": {
        "backup_directory": str(
            BACKUP_FOLDER
        ),
        "backups_inside_grid_year": False,
        "moved_backups": (
            moved_backups
        ),
    },
    "grid_year": {
        "combined_row_count": int(
            combined_row_count
        ),
        "observed_epochs": (
            observed_epochs
        ),
        "duplicate_cell_epoch_count": int(
            duplicate_cell_epoch_count
        ),
        "partitions": (
            partition_files
        ),
    },
    "temporal_overlap": {
        "cells_2015": int(
            cells_2015.size
        ),
        "cells_2020": int(
            cells_2020.size
        ),
        "common_cells": int(
            common_cells.size
        ),
        "only_2015": int(
            only_2015_cells.size
        ),
        "only_2020": int(
            only_2020_cells.size
        ),
    },
    "cell_reference": {
        "file": str(
            CELL_REFERENCE_FILE
        ),
        "row_count": int(
            cell_reference_row_count
        ),
        "duplicate_cell_id_count": int(
            cell_reference_duplicate_count
        ),
        "file_size_bytes": int(
            CELL_REFERENCE_FILE
            .stat()
            .st_size
        ),
        "sha256": calculate_sha256(
            CELL_REFERENCE_FILE
        ),
    },
    "building_height_2018": {
        "file": str(
            HEIGHT_FILE
        ),
        "row_count": int(
            height_row_count
        ),
        "reference_years": (
            height_reference_years
        ),
        "duplicate_cell_id_count": int(
            height_duplicate_count
        ),
        "file_size_bytes": int(
            HEIGHT_FILE
            .stat()
            .st_size
        ),
        "sha256": calculate_sha256(
            HEIGHT_FILE
        ),
        "temporal_policy": (
            "cross_sectional_only"
        ),
    },
}


with manifest_path.open(
    "w",
    encoding="utf-8",
) as manifest_file:
    json.dump(
        manifest,
        manifest_file,
        indent=2,
    )


# -------------------------------------------------------------------
# FINAL RESULT
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("FINAL PARQUET RESULT")
print("=" * 100)

print(
    "SUCCESS: Backup Parquet files were moved outside "
    "the Hive-partitioned grid_year dataset."
)

print(
    "The grid_year root now contains exactly two "
    "production Parquet files."
)

print(
    "The combined dataset contains exactly "
    "2,902,156 rows."
)

print(
    "Both epochs contain the same 1,451,078 "
    "stable cells."
)

print(
    "There are no duplicate cell-epoch keys."
)

print(
    "The cell-reference and building-height tables "
    "passed validation."
)

print(
    "The Parquet dataset is ready for HDFS and Spark."
)

print("\nParquet root:")
print(PARQUET_ROOT)

print("\nBackup folder:")
print(BACKUP_FOLDER)

print("\nManifest:")
print(manifest_path)
```

```python
# -------------------------------------------------------------------
# COMPLETE INDIA GHSL EXPORT SCRIPT
# -------------------------------------------------------------------
#
# Products:
#
# 1. GHS-SMOD
#    Epochs: 1975-2020 at five-year intervals
#    Band: smod_code
#    Native resolution: 1 km
#
# 2. GHS-POP
#    Epochs: 1975-2020 at five-year intervals
#    Band: population_count
#    Native resolution: 100 m
#
# 3. GHS-BUILT-S
#    Epochs: 1975-2020 at five-year intervals
#    Bands:
#       - built_surface
#       - built_surface_nres
#    Native resolution: 100 m
#
# 4. GHS-BUILT-H
#    Reference year: 2018 only
#    Band: built_height
#    Native resolution: 100 m
#
# Total expected logical exports:
#
# 10 SMOD
# + 10 population
# + 10 built-surface
# + 1 building-height
# = 31 Earth Engine tasks
#
# Large outputs may be split by Earth Engine into multiple GeoTIFFs
# because fileDimensions is used.
#
# IMPORTANT:
#
# - This script exports raw source-resolution products.
# - SMOD is NOT expanded to 100 m during export.
# - Population and built surface remain at 100 m.
# - Building height remains a 2018-only product.
# - No historical building-height values are fabricated.
# - The India boundary is used as the output mask.
# - The files will be curated locally after download.
#
# The script is restart-safe for tasks still visible in the Earth
# Engine task history. Existing READY, RUNNING or COMPLETED tasks with
# the same description are skipped.
#
# First run with DRY_RUN = True.
# Review the 31 planned jobs.
# Then change DRY_RUN to False and run the cell once.

import csv
import json
import time
from datetime import datetime
from pathlib import Path

import ee


# -------------------------------------------------------------------
# USER CONFIGURATION
# -------------------------------------------------------------------

PROJECT_ID = "india-ghsl-urbanization"

DRIVE_FOLDER = "GHSL_India_Full_R2023A"

CONTROL_FOLDER = Path(
    "/content/drive/MyDrive/GHSL_India_Export_Control"
)

CONTROL_FOLDER.mkdir(
    parents=True,
    exist_ok=True,
)


# First run:
#
# DRY_RUN = True
#
# After reviewing the plan, change to:
#
# DRY_RUN = False

DRY_RUN = False


# Historical analysis epochs only.
#
# 2025 and 2030 are deliberately excluded from the core historical
# acquisition. They can be exported later as test/projection epochs.

HISTORICAL_EPOCHS = [
    1975,
    1980,
    1985,
    1990,
    1995,
    2000,
    2005,
    2010,
    2015,
    2020,
]


# The export files may be split into blocks of this many pixels.
#
# 8192 is:
# - large enough to avoid excessive tiny files
# - divisible by the 256-pixel shard size
# - manageable for later Rasterio/HDFS processing

FILE_DIMENSIONS = 8192

SHARD_SIZE = 256

MAX_PIXELS = 10_000_000_000_000


# Prevent accidental rapid duplicate submission.
SUBMISSION_DELAY_SECONDS = 1.0


# -------------------------------------------------------------------
# AUTHENTICATE AND INITIALIZE EARTH ENGINE
# -------------------------------------------------------------------

try:
    ee.Initialize(
        project=PROJECT_ID
    )

    connection_test = (
        ee.Number(1).getInfo()
    )

except Exception:
    ee.Authenticate(
        auth_mode="notebook"
    )

    ee.Initialize(
        project=PROJECT_ID
    )

    connection_test = (
        ee.Number(1).getInfo()
    )


if connection_test != 1:
    raise RuntimeError(
        "Earth Engine connection test failed."
    )


print("=" * 100)
print("EARTH ENGINE CONNECTION")
print("=" * 100)

print("Project:", PROJECT_ID)
print("Connection test:", connection_test)


# -------------------------------------------------------------------
# LOAD A DETAILED INDIA BOUNDARY
# -------------------------------------------------------------------
#
# The earlier pilot notebook used USDOS/LSIB_SIMPLE/2017.
#
# For national acquisition, use the GAUL level-0 India boundary
# because the simplified LSIB product can omit smaller islands.
#
# This boundary is an acquisition/clipping geometry.
# District-level analytical boundaries will be validated separately.

INDIA_BOUNDARY_ASSET = (
    "FAO/GAUL/2015/level0"
)

india_collection = (
    ee.FeatureCollection(
        INDIA_BOUNDARY_ASSET
    )
    .filter(
        ee.Filter.eq(
            "ADM0_NAME",
            "India",
        )
    )
)


india_feature_count = (
    india_collection.size().getInfo()
)


if india_feature_count != 1:
    raise RuntimeError(
        "Expected exactly one India boundary feature, "
        f"but found {india_feature_count}."
    )


india_geometry = (
    india_collection.geometry()
)


india_bounds = (
    india_geometry
    .bounds(
        maxError=1000
    )
    .getInfo()
)


india_area_km2 = (
    india_geometry
    .area(
        maxError=1000
    )
    .divide(
        1_000_000
    )
    .getInfo()
)


print("\n" + "=" * 100)
print("INDIA EXPORT BOUNDARY")
print("=" * 100)

print(
    "Boundary asset:",
    INDIA_BOUNDARY_ASSET,
)

print(
    "Matching features:",
    india_feature_count,
)

print(
    "Approximate boundary area:",
    f"{india_area_km2:,.2f}",
    "square kilometres",
)

print(
    "Boundary geometry type:",
    india_bounds.get(
        "type"
    ),
)


# -------------------------------------------------------------------
# DATASET DEFINITIONS
# -------------------------------------------------------------------

PRODUCTS = {
    "smod": {
        "asset_template": (
            "JRC/GHSL/P2023A/"
            "GHS_SMOD_V2-0/{year}"
        ),
        "bands": [
            "smod_code",
        ],
        "scale": 1000,
        "cast": "int16",
        "name_token": "SMOD",
        "source_resolution_m": 1000,
        "temporal_type": "historical",
    },
    "population": {
        "asset_template": (
            "JRC/GHSL/P2023A/"
            "GHS_POP/{year}"
        ),
        "bands": [
            "population_count",
        ],
        "scale": 100,
        "cast": "float32",
        "name_token": "POP",
        "source_resolution_m": 100,
        "temporal_type": "historical",
    },
    "built_surface": {
        "asset_template": (
            "JRC/GHSL/P2023A/"
            "GHS_BUILT_S/{year}"
        ),
        "bands": [
            "built_surface",
            "built_surface_nres",
        ],
        "scale": 100,
        "cast": "uint16",
        "name_token": "BUILT_S",
        "source_resolution_m": 100,
        "temporal_type": "historical",
    },
    "building_height": {
        "asset_template": (
            "JRC/GHSL/P2023A/"
            "GHS_BUILT_H/{year}"
        ),
        "bands": [
            "built_height",
        ],
        "scale": 100,
        "cast": "float32",
        "name_token": "BUILT_H",
        "source_resolution_m": 100,
        "temporal_type": "cross_sectional_only",
    },
}


# -------------------------------------------------------------------
# CAST HELPER
# -------------------------------------------------------------------

def cast_image(
    image,
    cast_name,
):
    if cast_name == "int16":
        return image.toInt16()

    if cast_name == "uint16":
        return image.toUint16()

    if cast_name == "float32":
        return image.toFloat()

    raise ValueError(
        f"Unsupported cast type: {cast_name}"
    )


# -------------------------------------------------------------------
# BUILD EXPORT PLAN
# -------------------------------------------------------------------

export_jobs = []


for epoch in HISTORICAL_EPOCHS:
    for product_key in [
        "smod",
        "population",
        "built_surface",
    ]:
        product = PRODUCTS[
            product_key
        ]

        asset_id = (
            product[
                "asset_template"
            ].format(
                year=epoch
            )
        )

        description = (
            f"GHSL_India_"
            f"{product['name_token']}_"
            f"{epoch}_"
            f"{product['scale']}m_"
            f"R2023A"
        )

        filename_prefix = (
            f"ghsl_india_"
            f"{product_key}_"
            f"{epoch}_"
            f"{product['scale']}m_"
            f"r2023a"
        )

        export_jobs.append(
            {
                "job_key": (
                    f"{product_key}_{epoch}"
                ),
                "product": product_key,
                "year": epoch,
                "asset_id": asset_id,
                "bands": product["bands"],
                "scale": product["scale"],
                "cast": product["cast"],
                "description": description,
                "filename_prefix": (
                    filename_prefix
                ),
                "source_resolution_m": (
                    product[
                        "source_resolution_m"
                    ]
                ),
                "temporal_type": (
                    product[
                        "temporal_type"
                    ]
                ),
            }
        )


# Building height is added once for 2018.
height_product = PRODUCTS[
    "building_height"
]

height_year = 2018

height_asset_id = (
    height_product[
        "asset_template"
    ].format(
        year=height_year
    )
)

height_description = (
    "GHSL_India_BUILT_H_"
    "2018_100m_R2023A"
)

height_filename_prefix = (
    "ghsl_india_"
    "building_height_"
    "2018_100m_r2023a"
)


export_jobs.append(
    {
        "job_key": (
            "building_height_2018"
        ),
        "product": (
            "building_height"
        ),
        "year": 2018,
        "asset_id": (
            height_asset_id
        ),
        "bands": (
            height_product["bands"]
        ),
        "scale": (
            height_product["scale"]
        ),
        "cast": (
            height_product["cast"]
        ),
        "description": (
            height_description
        ),
        "filename_prefix": (
            height_filename_prefix
        ),
        "source_resolution_m": (
            height_product[
                "source_resolution_m"
            ]
        ),
        "temporal_type": (
            height_product[
                "temporal_type"
            ]
        ),
    }
)


# -------------------------------------------------------------------
# VALIDATE PLAN COMPLETENESS
# -------------------------------------------------------------------

expected_job_count = 31


if len(export_jobs) != expected_job_count:
    raise RuntimeError(
        "Unexpected export-job count. "
        f"Expected {expected_job_count}, "
        f"found {len(export_jobs)}."
    )


job_keys = [
    job["job_key"]
    for job in export_jobs
]


if len(job_keys) != len(
    set(job_keys)
):
    raise RuntimeError(
        "Duplicate job keys exist in the export plan."
    )


descriptions = [
    job["description"]
    for job in export_jobs
]


if len(descriptions) != len(
    set(descriptions)
):
    raise RuntimeError(
        "Duplicate Earth Engine task descriptions exist."
    )


# -------------------------------------------------------------------
# VERIFY ASSETS AND BANDS BEFORE SUBMISSION
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("VERIFYING ALL SOURCE ASSETS")
print("=" * 100)


asset_validation_results = []


for index, job in enumerate(
    export_jobs,
    start=1,
):
    image = ee.Image(
        job["asset_id"]
    )

    actual_bands = (
        image.bandNames()
        .getInfo()
    )

    expected_bands = (
        job["bands"]
    )

    missing_bands = [
        band
        for band in expected_bands
        if band not in actual_bands
    ]

    asset_valid = (
        len(missing_bands) == 0
    )

    asset_validation_results.append(
        {
            "job_key": (
                job["job_key"]
            ),
            "asset_id": (
                job["asset_id"]
            ),
            "expected_bands": (
                expected_bands
            ),
            "actual_bands": (
                actual_bands
            ),
            "missing_bands": (
                missing_bands
            ),
            "valid": (
                asset_valid
            ),
        }
    )

    print(
        f"{index:>2}/{len(export_jobs)} "
        f"{job['job_key']:<32} "
        f"{'PASS' if asset_valid else 'FAIL'}"
    )

    if not asset_valid:
        print(
            "   Expected:",
            expected_bands,
        )

        print(
            "   Actual:",
            actual_bands,
        )

        print(
            "   Missing:",
            missing_bands,
        )


invalid_assets = [
    result
    for result in asset_validation_results
    if not result["valid"]
]


if invalid_assets:
    raise RuntimeError(
        "One or more GHSL assets failed band validation. "
        "No exports were submitted."
    )


# -------------------------------------------------------------------
# PRINT EXPORT PLAN
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("COMPLETE INDIA EXPORT PLAN")
print("=" * 100)


for index, job in enumerate(
    export_jobs,
    start=1,
):
    print(
        f"{index:>2}. "
        f"{job['job_key']:<32} | "
        f"{job['scale']:>4} m | "
        f"{','.join(job['bands'])}"
    )


print("\nSummary:")
print("SMOD jobs:", 10)
print("Population jobs:", 10)
print("Built-surface jobs:", 10)
print("Building-height jobs:", 1)
print("Total jobs:", len(export_jobs))
print("Drive folder:", DRIVE_FOLDER)
print("File dimensions:", FILE_DIMENSIONS)
print("Shard size:", SHARD_SIZE)
print("Dry run:", DRY_RUN)


# -------------------------------------------------------------------
# SAVE PRE-SUBMISSION PLAN
# -------------------------------------------------------------------

plan_timestamp = datetime.now().strftime(
    "%Y%m%d_%H%M%S"
)

latest_plan_path = (
    CONTROL_FOLDER
    / "india_export_plan_latest.json"
)

timestamped_plan_path = (
    CONTROL_FOLDER
    / (
        "india_export_plan_"
        f"{plan_timestamp}.json"
    )
)


plan_document = {
    "created_at": (
        datetime.now().isoformat()
    ),
    "project_id": (
        PROJECT_ID
    ),
    "drive_folder": (
        DRIVE_FOLDER
    ),
    "boundary_asset": (
        INDIA_BOUNDARY_ASSET
    ),
    "boundary_feature_count": (
        india_feature_count
    ),
    "boundary_area_km2": (
        india_area_km2
    ),
    "historical_epochs": (
        HISTORICAL_EPOCHS
    ),
    "building_height_year": 2018,
    "file_dimensions": (
        FILE_DIMENSIONS
    ),
    "shard_size": (
        SHARD_SIZE
    ),
    "max_pixels": (
        MAX_PIXELS
    ),
    "dry_run": (
        DRY_RUN
    ),
    "expected_job_count": (
        expected_job_count
    ),
    "jobs": (
        export_jobs
    ),
    "asset_validation": (
        asset_validation_results
    ),
}


for plan_path in [
    latest_plan_path,
    timestamped_plan_path,
]:
    with plan_path.open(
        "w",
        encoding="utf-8",
    ) as plan_file:
        json.dump(
            plan_document,
            plan_file,
            indent=2,
        )


print("\nExport plan saved:")
print(latest_plan_path)


# -------------------------------------------------------------------
# STOP AFTER DRY RUN
# -------------------------------------------------------------------

if DRY_RUN:
    print("\n" + "=" * 100)
    print("DRY RUN COMPLETE")
    print("=" * 100)

    print(
        "All 31 source assets and required bands were validated."
    )

    print(
        "No Earth Engine export tasks were submitted."
    )

    print(
        "Review the plan above and the saved JSON manifest."
    )

    print(
        "If the plan is correct, change DRY_RUN to False "
        "and rerun this cell once."
    )


# -------------------------------------------------------------------
# SUBMIT EXPORT TASKS
# -------------------------------------------------------------------

india_export_tasks = {}


if not DRY_RUN:
    print("\n" + "=" * 100)
    print("CHECKING EXISTING EARTH ENGINE TASKS")
    print("=" * 100)

    existing_tasks = (
        ee.batch.Task.list()
    )

    existing_tasks_by_description = {}


    for task in existing_tasks:
        status = task.status()

        description = status.get(
            "description"
        )

        if description:
            existing_tasks_by_description[
                description
            ] = {
                "task": task,
                "status": status,
            }


    print(
        "Existing Earth Engine tasks indexed:",
        len(
            existing_tasks_by_description
        ),
    )


    print("\n" + "=" * 100)
    print("SUBMITTING INDIA EXPORT TASKS")
    print("=" * 100)


    submission_records = []


    for index, job in enumerate(
        export_jobs,
        start=1,
    ):
        description = (
            job["description"]
        )

        existing_record = (
            existing_tasks_by_description
            .get(
                description
            )
        )

        if existing_record is not None:
            existing_state = (
                existing_record[
                    "status"
                ].get(
                    "state",
                    "UNKNOWN",
                )
            )

            if existing_state in {
                "READY",
                "RUNNING",
                "COMPLETED",
            }:
                existing_task = (
                    existing_record[
                        "task"
                    ]
                )

                india_export_tasks[
                    job["job_key"]
                ] = existing_task

                print(
                    f"{index:>2}/{len(export_jobs)} "
                    f"SKIPPED EXISTING | "
                    f"{job['job_key']:<32} | "
                    f"{existing_state}"
                )

                submission_records.append(
                    {
                        **job,
                        "action": (
                            "skipped_existing"
                        ),
                        "task_id": (
                            existing_task.id
                        ),
                        "initial_state": (
                            existing_state
                        ),
                    }
                )

                continue


        source_image = (
            ee.Image(
                job["asset_id"]
            )
            .select(
                job["bands"]
            )
        )

        casted_image = cast_image(
            source_image,
            job["cast"],
        )

        # Mask the exported raster to the India boundary.
        #
        # The export region uses the India bounding rectangle, while
        # clipToCollection preserves only pixels inside the boundary.

        export_image = (
            casted_image
            .clipToCollection(
                india_collection
            )
        )


        task = (
            ee.batch.Export.image.toDrive(
                image=export_image,
                description=(
                    job["description"]
                ),
                folder=DRIVE_FOLDER,
                fileNamePrefix=(
                    job[
                        "filename_prefix"
                    ]
                ),
                region=india_geometry,
                scale=job["scale"],
                maxPixels=MAX_PIXELS,
                shardSize=SHARD_SIZE,
                fileDimensions=[
                    FILE_DIMENSIONS,
                    FILE_DIMENSIONS,
                ],
                skipEmptyTiles=True,
                fileFormat="GeoTIFF",
                formatOptions={
                    "cloudOptimized": True,
                },
            )
        )

        task.start()

        india_export_tasks[
            job["job_key"]
        ] = task

        initial_status = (
            task.status()
        )

        initial_state = (
            initial_status.get(
                "state",
                "UNKNOWN",
            )
        )

        print(
            f"{index:>2}/{len(export_jobs)} "
            f"SUBMITTED | "
            f"{job['job_key']:<32} | "
            f"{initial_state} | "
            f"{task.id}"
        )

        submission_records.append(
            {
                **job,
                "action": "submitted",
                "task_id": task.id,
                "initial_state": (
                    initial_state
                ),
            }
        )

        time.sleep(
            SUBMISSION_DELAY_SECONDS
        )


    # ---------------------------------------------------------------
    # SAVE SUBMISSION MANIFEST
    # ---------------------------------------------------------------

    submission_manifest = {
        "submitted_at": (
            datetime.now().isoformat()
        ),
        "project_id": (
            PROJECT_ID
        ),
        "drive_folder": (
            DRIVE_FOLDER
        ),
        "task_count": (
            len(
                submission_records
            )
        ),
        "tasks": (
            submission_records
        ),
    }


    submission_manifest_path = (
        CONTROL_FOLDER
        / "india_export_submission_latest.json"
    )


    with submission_manifest_path.open(
        "w",
        encoding="utf-8",
    ) as submission_file:
        json.dump(
            submission_manifest,
            submission_file,
            indent=2,
        )


    # ---------------------------------------------------------------
    # SAVE CSV TASK MANIFEST
    # ---------------------------------------------------------------

    submission_csv_path = (
        CONTROL_FOLDER
        / "india_export_submission_latest.csv"
    )


    csv_fields = [
        "job_key",
        "product",
        "year",
        "asset_id",
        "bands",
        "scale",
        "source_resolution_m",
        "temporal_type",
        "description",
        "filename_prefix",
        "action",
        "task_id",
        "initial_state",
    ]


    with submission_csv_path.open(
        "w",
        newline="",
        encoding="utf-8",
    ) as csv_file:
        writer = csv.DictWriter(
            csv_file,
            fieldnames=csv_fields,
        )

        writer.writeheader()

        for record in submission_records:
            csv_record = {
                field: record.get(
                    field
                )
                for field in csv_fields
            }

            csv_record["bands"] = (
                ",".join(
                    record["bands"]
                )
            )

            writer.writerow(
                csv_record
            )


    # ---------------------------------------------------------------
    # INITIAL TASK STATUS SUMMARY
    # ---------------------------------------------------------------

    state_counts = {}


    for job_key, task in (
        india_export_tasks.items()
    ):
        status = task.status()

        state = status.get(
            "state",
            "UNKNOWN",
        )

        state_counts[state] = (
            state_counts.get(
                state,
                0,
            )
            + 1
        )


    print("\n" + "=" * 100)
    print("INITIAL TASK SUMMARY")
    print("=" * 100)

    print(
        "Tasks tracked:",
        len(india_export_tasks),
    )

    for state in sorted(
        state_counts
    ):
        print(
            f"{state}:",
            state_counts[state],
        )


    print("\nSubmission manifest:")
    print(submission_manifest_path)

    print("\nSubmission CSV:")
    print(submission_csv_path)

    print("\nImportant:")
    print(
        "Do not rerun the submission with different task "
        "descriptions while these tasks are active."
    )

    print(
        "Large exports may remain READY for a while before "
        "Earth Engine schedules them."
    )

    print(
        "Earth Engine may generate multiple GeoTIFF files "
        "for a single task because fileDimensions is enabled."
    )

    print(
        "SMOD remains at 1 km. It will be mapped to 100 m "
        "during local curation, not during raw export."
    )

    print(
        "Building height remains a 2018-only "
        "cross-sectional product."
    )
```

```python
# -------------------------------------------------------------------
# INDIA GHSL DRIVE INVENTORY AND GEOTIFF VALIDATION
# -------------------------------------------------------------------
#
# Read-only validation.
#
# This cell does NOT:
# - modify GeoTIFF files
# - delete or move files
# - submit Earth Engine tasks
# - upload files to HDFS
#
# It validates:
# 1. All 31 logical exports are present.
# 2. Every GeoTIFF shard is readable.
# 3. File size is greater than zero.
# 4. CRS is present.
# 5. Band count is correct.
# 6. Resolution matches the product.
# 7. Raster dimensions, transforms and bounds are valid.
# 8. Shards belonging to one logical export are consistent.
# 9. Validation reports are saved in Google Drive.

from pathlib import Path
from datetime import datetime
from collections import defaultdict
import csv
import json
import math
import subprocess
import sys


# -------------------------------------------------------------------
# IMPORT RASTERIO
# -------------------------------------------------------------------

try:
    import rasterio

except ImportError:
    print("Installing rasterio...")

    subprocess.check_call(
        [
            sys.executable,
            "-m",
            "pip",
            "install",
            "-q",
            "rasterio",
        ]
    )

    import rasterio


# -------------------------------------------------------------------
# PATHS
# -------------------------------------------------------------------

MY_DRIVE = Path(
    "/content/drive/MyDrive"
)

CONTROL_FOLDER = (
    MY_DRIVE
    / "GHSL_India_Export_Control"
)

SUBMISSION_MANIFEST_PATH = (
    CONTROL_FOLDER
    / "india_export_submission_latest.json"
)

VALIDATION_JSON_PATH = (
    CONTROL_FOLDER
    / "india_drive_geotiff_validation_latest.json"
)

VALIDATION_CSV_PATH = (
    CONTROL_FOLDER
    / "india_drive_geotiff_validation_latest.csv"
)

LOGICAL_EXPORT_CSV_PATH = (
    CONTROL_FOLDER
    / "india_logical_export_inventory_latest.csv"
)


# -------------------------------------------------------------------
# VERIFY REQUIRED PATHS
# -------------------------------------------------------------------

if not MY_DRIVE.exists():
    raise FileNotFoundError(
        "Google Drive is not mounted at:\n"
        f"{MY_DRIVE}"
    )


if not CONTROL_FOLDER.exists():
    raise FileNotFoundError(
        "Control folder not found:\n"
        f"{CONTROL_FOLDER}"
    )


if not SUBMISSION_MANIFEST_PATH.exists():
    raise FileNotFoundError(
        "Submission manifest not found:\n"
        f"{SUBMISSION_MANIFEST_PATH}"
    )


# -------------------------------------------------------------------
# LOAD SUBMISSION MANIFEST
# -------------------------------------------------------------------

with SUBMISSION_MANIFEST_PATH.open(
    "r",
    encoding="utf-8",
) as manifest_file:
    submission_manifest = json.load(
        manifest_file
    )


expected_jobs = submission_manifest.get(
    "tasks",
    []
)


if len(expected_jobs) != 31:
    raise RuntimeError(
        "Expected 31 export jobs in the submission manifest, "
        f"but found {len(expected_jobs)}."
    )


drive_folder_name = (
    submission_manifest.get(
        "drive_folder",
        "GHSL_India_Full_R2023A",
    )
)


# -------------------------------------------------------------------
# PRINT STARTING INFORMATION
# -------------------------------------------------------------------

print("=" * 100)
print("INDIA GHSL DRIVE INVENTORY")
print("=" * 100)

print(
    "Submission manifest:",
    SUBMISSION_MANIFEST_PATH,
)

print(
    "Expected logical exports:",
    len(expected_jobs),
)

print(
    "Expected Drive folder name:",
    drive_folder_name,
)


# -------------------------------------------------------------------
# FIND MATCHING DRIVE FOLDERS
# -------------------------------------------------------------------
#
# This also detects folders such as:
#
# GHSL_India_Full_R2023A
# GHSL_India_Full_R2023A (1)
# GHSL_India_Full_R2023A (2)

candidate_export_folders = sorted(
    path
    for path in MY_DRIVE.iterdir()
    if (
        path.is_dir()
        and path.name.startswith(
            drive_folder_name
        )
    )
)


print("\nMatching Drive folders found:")


if candidate_export_folders:
    for folder in candidate_export_folders:
        print(
            " -",
            folder,
        )

else:
    print(
        " - None"
    )


if not candidate_export_folders:
    raise FileNotFoundError(
        "No matching India export folder was found.\n"
        f"Expected a folder beginning with: {drive_folder_name}"
    )


# -------------------------------------------------------------------
# FIND ALL GEOTIFF FILES
# -------------------------------------------------------------------

all_tiff_files = []


for folder in candidate_export_folders:
    for path in folder.rglob("*"):
        if (
            path.is_file()
            and path.suffix.lower()
            in {
                ".tif",
                ".tiff",
            }
        ):
            all_tiff_files.append(
                path
            )


# Remove duplicate physical paths if any folder is referenced twice.

unique_tiff_files = {}


for file_path in all_tiff_files:
    resolved_path = str(
        file_path.resolve()
    )

    unique_tiff_files[
        resolved_path
    ] = file_path


all_tiff_files = sorted(
    unique_tiff_files.values()
)


total_tiff_size_bytes = sum(
    file_path.stat().st_size
    for file_path in all_tiff_files
)


print("\nPhysical GeoTIFF files found:")
print(
    f"{len(all_tiff_files):,}"
)


print("\nTotal GeoTIFF size:")
print(
    f"{total_tiff_size_bytes / (1024 ** 3):,.3f}",
    "GiB",
)


if not all_tiff_files:
    raise RuntimeError(
        "No GeoTIFF files were found in the export folder."
    )


# -------------------------------------------------------------------
# EXPECTED PRODUCT RULES
# -------------------------------------------------------------------

PRODUCT_RULES = {
    "smod": {
        "expected_band_count": 1,
        "expected_resolution_m": 1000.0,
        "expected_bands": [
            "smod_code",
        ],
    },
    "population": {
        "expected_band_count": 1,
        "expected_resolution_m": 100.0,
        "expected_bands": [
            "population_count",
        ],
    },
    "built_surface": {
        "expected_band_count": 2,
        "expected_resolution_m": 100.0,
        "expected_bands": [
            "built_surface",
            "built_surface_nres",
        ],
    },
    "building_height": {
        "expected_band_count": 1,
        "expected_resolution_m": 100.0,
        "expected_bands": [
            "built_height",
        ],
    },
}


RESOLUTION_RELATIVE_TOLERANCE = 0.001


def resolution_matches(
    observed,
    expected,
):
    tolerance = max(
        0.01,
        expected
        * RESOLUTION_RELATIVE_TOLERANCE,
    )

    return (
        abs(
            abs(observed)
            - expected
        )
        <= tolerance
    )


# -------------------------------------------------------------------
# GROUP FILES BY EXPECTED FILENAME PREFIX
# -------------------------------------------------------------------

files_by_job = defaultdict(
    list
)

unmatched_tiff_files = []

multiple_match_files = []


for file_path in all_tiff_files:
    file_name_lower = (
        file_path.name.lower()
    )

    matched_job_keys = []

    for job in expected_jobs:
        filename_prefix = str(
            job["filename_prefix"]
        ).lower()

        if file_name_lower.startswith(
            filename_prefix
        ):
            matched_job_keys.append(
                job["job_key"]
            )

    if len(matched_job_keys) == 1:
        files_by_job[
            matched_job_keys[0]
        ].append(
            file_path
        )

    elif len(matched_job_keys) == 0:
        unmatched_tiff_files.append(
            file_path
        )

    else:
        multiple_match_files.append(
            {
                "file": str(
                    file_path
                ),
                "matched_job_keys": (
                    matched_job_keys
                ),
            }
        )


if multiple_match_files:
    print(
        "\nFiles matching multiple logical exports:"
    )

    for record in multiple_match_files:
        print(
            " -",
            record,
        )

    raise RuntimeError(
        "One or more GeoTIFF files matched multiple "
        "expected filename prefixes."
    )


# -------------------------------------------------------------------
# LOGICAL EXPORT INVENTORY
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("LOGICAL EXPORT INVENTORY")
print("=" * 100)


logical_export_records = []

missing_job_keys = []


for index, job in enumerate(
    expected_jobs,
    start=1,
):
    job_key = job["job_key"]

    matched_files = sorted(
        files_by_job.get(
            job_key,
            [],
        )
    )

    matched_size_bytes = sum(
        file_path.stat().st_size
        for file_path in matched_files
    )

    if matched_files:
        logical_status = "FOUND"

    else:
        logical_status = "MISSING"

        missing_job_keys.append(
            job_key
        )

    logical_record = {
        "job_key": job_key,
        "product": job["product"],
        "year": int(
            job["year"]
        ),
        "filename_prefix": (
            job["filename_prefix"]
        ),
        "logical_status": (
            logical_status
        ),
        "shard_count": int(
            len(matched_files)
        ),
        "total_size_bytes": int(
            matched_size_bytes
        ),
        "total_size_gib": round(
            matched_size_bytes
            / (1024 ** 3),
            6,
        ),
    }

    logical_export_records.append(
        logical_record
    )

    print(
        f"{index:>2}/31 "
        f"{job_key:<32} | "
        f"{logical_status:<7} | "
        f"shards={len(matched_files):>4} | "
        f"size="
        f"{matched_size_bytes / (1024 ** 2):>10,.2f} MiB"
    )


found_logical_export_count = (
    len(expected_jobs)
    - len(missing_job_keys)
)


print("\nLogical exports found:")
print(
    f"{found_logical_export_count}/31"
)


print("\nLogical exports missing:")
print(
    len(missing_job_keys)
)


if missing_job_keys:
    for job_key in missing_job_keys:
        print(
            " -",
            job_key,
        )


print("\nUnmatched GeoTIFF files:")
print(
    len(unmatched_tiff_files)
)


if unmatched_tiff_files:
    for file_path in unmatched_tiff_files:
        print(
            " -",
            file_path,
        )


# -------------------------------------------------------------------
# COUNT MATCHED PHYSICAL SHARDS
# -------------------------------------------------------------------

total_matched_physical_files = sum(
    len(
        files_by_job.get(
            job["job_key"],
            [],
        )
    )
    for job in expected_jobs
)


print("\nMatched physical GeoTIFF shards:")
print(
    total_matched_physical_files
)


# -------------------------------------------------------------------
# VALIDATE EVERY PHYSICAL GEOTIFF SHARD
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("PHYSICAL GEOTIFF VALIDATION")
print("=" * 100)


physical_validation_records = []

invalid_file_records = []

physical_index = 0


for job in expected_jobs:
    job_key = job["job_key"]
    product = job["product"]
    year = int(
        job["year"]
    )

    rules = PRODUCT_RULES.get(
        product
    )

    if rules is None:
        raise KeyError(
            "No validation rules found for product: "
            f"{product}"
        )

    expected_band_count = int(
        rules[
            "expected_band_count"
        ]
    )

    expected_resolution_m = float(
        rules[
            "expected_resolution_m"
        ]
    )

    matched_files = sorted(
        files_by_job.get(
            job_key,
            [],
        )
    )

    for file_path in matched_files:
        physical_index += 1

        file_size_bytes = int(
            file_path.stat().st_size
        )

        errors = []

        warnings = []

        record = {
            "job_key": job_key,
            "product": product,
            "year": year,
            "file": str(
                file_path
            ),
            "folder": str(
                file_path.parent
            ),
            "file_name": (
                file_path.name
            ),
            "file_size_bytes": (
                file_size_bytes
            ),
            "file_size_mib": round(
                file_size_bytes
                / (1024 ** 2),
                6,
            ),
            "expected_band_count": (
                expected_band_count
            ),
            "expected_resolution_m": (
                expected_resolution_m
            ),
        }

        if file_size_bytes <= 0:
            errors.append(
                "File is empty"
            )

        try:
            with rasterio.open(
                file_path
            ) as dataset:
                width = int(
                    dataset.width
                )

                height = int(
                    dataset.height
                )

                band_count = int(
                    dataset.count
                )

                crs_text = (
                    dataset.crs.to_string()
                    if dataset.crs
                    else None
                )

                x_resolution = float(
                    dataset.transform.a
                )

                y_resolution = float(
                    dataset.transform.e
                )

                transform_values = [
                    float(value)
                    for value
                    in tuple(
                        dataset.transform
                    )
                ]

                bounds = dataset.bounds

                bounds_values = {
                    "left": float(
                        bounds.left
                    ),
                    "bottom": float(
                        bounds.bottom
                    ),
                    "right": float(
                        bounds.right
                    ),
                    "top": float(
                        bounds.top
                    ),
                }

                dtype_values = list(
                    dataset.dtypes
                )

                nodata_values = list(
                    dataset.nodatavals
                )

                band_descriptions = list(
                    dataset.descriptions
                )

                driver = dataset.driver

                tiled = bool(
                    dataset.is_tiled
                )

                compression = (
                    dataset.compression.value
                    if dataset.compression
                    else None
                )

                block_shapes = [
                    list(shape)
                    for shape
                    in dataset.block_shapes
                ]

                # ---------------------------------------------------
                # SMALL READABILITY TEST
                # ---------------------------------------------------
                #
                # Read one pixel from the centre of every band.
                # This confirms the raster can be read without
                # loading the complete national shard into memory.

                if (
                    width > 0
                    and height > 0
                    and band_count > 0
                ):
                    centre_column = (
                        width // 2
                    )

                    centre_row = (
                        height // 2
                    )

                    sample_data = dataset.read(
                        indexes=list(
                            range(
                                1,
                                band_count + 1,
                            )
                        ),
                        window=(
                            (
                                centre_row,
                                centre_row + 1,
                            ),
                            (
                                centre_column,
                                centre_column + 1,
                            ),
                        ),
                    )

                    if sample_data.shape != (
                        band_count,
                        1,
                        1,
                    ):
                        errors.append(
                            "Unexpected sample-read shape: "
                            f"{sample_data.shape}"
                        )

                # ---------------------------------------------------
                # DIMENSION CHECKS
                # ---------------------------------------------------

                if width <= 0:
                    errors.append(
                        "Width is not positive"
                    )

                if height <= 0:
                    errors.append(
                        "Height is not positive"
                    )

                # ---------------------------------------------------
                # BAND CHECK
                # ---------------------------------------------------

                if band_count != expected_band_count:
                    errors.append(
                        "Unexpected band count: "
                        f"expected {expected_band_count}, "
                        f"found {band_count}"
                    )

                # ---------------------------------------------------
                # CRS CHECK
                # ---------------------------------------------------

                if dataset.crs is None:
                    errors.append(
                        "CRS is missing"
                    )

                # ---------------------------------------------------
                # RESOLUTION CHECKS
                # ---------------------------------------------------

                if not resolution_matches(
                    x_resolution,
                    expected_resolution_m,
                ):
                    errors.append(
                        "Unexpected X resolution: "
                        f"expected approximately "
                        f"{expected_resolution_m}, "
                        f"found {x_resolution}"
                    )

                if not resolution_matches(
                    y_resolution,
                    expected_resolution_m,
                ):
                    errors.append(
                        "Unexpected Y resolution: "
                        f"expected approximately "
                        f"{expected_resolution_m}, "
                        f"found {y_resolution}"
                    )

                # ---------------------------------------------------
                # TRANSFORM CHECK
                # ---------------------------------------------------

                if not all(
                    math.isfinite(value)
                    for value
                    in transform_values
                ):
                    errors.append(
                        "Transform contains a non-finite value"
                    )

                # ---------------------------------------------------
                # BOUNDS CHECKS
                # ---------------------------------------------------

                if not all(
                    math.isfinite(value)
                    for value
                    in bounds_values.values()
                ):
                    errors.append(
                        "Bounds contain a non-finite value"
                    )

                if (
                    bounds.left
                    >= bounds.right
                ):
                    errors.append(
                        "Invalid horizontal bounds"
                    )

                if (
                    bounds.bottom
                    >= bounds.top
                ):
                    errors.append(
                        "Invalid vertical bounds"
                    )

                # Earth Engine GeoTIFFs may not preserve
                # human-readable band descriptions.

                if all(
                    description is None
                    for description
                    in band_descriptions
                ):
                    warnings.append(
                        "Band descriptions are absent"
                    )

                record.update(
                    {
                        "driver": driver,
                        "width": width,
                        "height": height,
                        "band_count": (
                            band_count
                        ),
                        "crs": crs_text,
                        "x_resolution": (
                            x_resolution
                        ),
                        "y_resolution": (
                            y_resolution
                        ),
                        "transform": (
                            transform_values
                        ),
                        "bounds": (
                            bounds_values
                        ),
                        "dtypes": (
                            dtype_values
                        ),
                        "nodata_values": (
                            nodata_values
                        ),
                        "band_descriptions": (
                            band_descriptions
                        ),
                        "tiled": tiled,
                        "compression": (
                            compression
                        ),
                        "block_shapes": (
                            block_shapes
                        ),
                    }
                )

        except Exception as error:
            errors.append(
                "Rasterio open/read failed: "
                f"{type(error).__name__}: "
                f"{str(error)}"
            )

        if errors:
            status = "FAIL"

        else:
            status = "PASS"

        record[
            "status"
        ] = status

        record[
            "errors"
        ] = errors

        record[
            "warnings"
        ] = warnings

        physical_validation_records.append(
            record
        )

        if errors:
            invalid_file_records.append(
                record
            )

        print(
            f"{physical_index:>4}/"
            f"{total_matched_physical_files:<4} "
            f"{status:<4} | "
            f"{job_key:<32} | "
            f"{file_path.name}"
        )

        if errors:
            for error in errors:
                print(
                    "      ERROR:",
                    error,
                )


# -------------------------------------------------------------------
# GROUP VALIDATION RECORDS BY LOGICAL EXPORT
# -------------------------------------------------------------------

records_by_job = defaultdict(
    list
)


for record in physical_validation_records:
    records_by_job[
        record["job_key"]
    ].append(
        record
    )


# -------------------------------------------------------------------
# CROSS-SHARD CONSISTENCY CHECKS
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("LOGICAL EXPORT CONSISTENCY")
print("=" * 100)


consistency_records = []

consistency_failures = []


for job in expected_jobs:
    job_key = job["job_key"]
    product = job["product"]
    year = int(
        job["year"]
    )

    records = records_by_job.get(
        job_key,
        [],
    )

    consistency_errors = []

    observed_crs_values = sorted(
        {
            record.get(
                "crs"
            )
            for record in records
            if record.get(
                "crs"
            )
        }
    )

    observed_band_counts = sorted(
        {
            int(
                record.get(
                    "band_count"
                )
            )
            for record in records
            if record.get(
                "band_count"
            )
            is not None
        }
    )

    observed_x_resolutions = sorted(
        {
            round(
                abs(
                    float(
                        record.get(
                            "x_resolution"
                        )
                    )
                ),
                6,
            )
            for record in records
            if record.get(
                "x_resolution"
            )
            is not None
        }
    )

    observed_y_resolutions = sorted(
        {
            round(
                abs(
                    float(
                        record.get(
                            "y_resolution"
                        )
                    )
                ),
                6,
            )
            for record in records
            if record.get(
                "y_resolution"
            )
            is not None
        }
    )

    observed_dtypes = sorted(
        {
            tuple(
                record.get(
                    "dtypes",
                    [],
                )
            )
            for record in records
        }
    )

    failed_shard_count = sum(
        record["status"] == "FAIL"
        for record in records
    )

    if not records:
        consistency_errors.append(
            "No GeoTIFF shards found"
        )

    if len(
        observed_crs_values
    ) > 1:
        consistency_errors.append(
            "Multiple CRS values found within "
            "the same logical export"
        )

    if len(
        observed_band_counts
    ) > 1:
        consistency_errors.append(
            "Multiple band counts found within "
            "the same logical export"
        )

    if len(
        observed_x_resolutions
    ) > 1:
        consistency_errors.append(
            "Multiple X resolutions found within "
            "the same logical export"
        )

    if len(
        observed_y_resolutions
    ) > 1:
        consistency_errors.append(
            "Multiple Y resolutions found within "
            "the same logical export"
        )

    if len(
        observed_dtypes
    ) > 1:
        consistency_errors.append(
            "Multiple dtype combinations found within "
            "the same logical export"
        )

    if failed_shard_count > 0:
        consistency_errors.append(
            f"{failed_shard_count} physical shard(s) failed validation"
        )

    if consistency_errors:
        consistency_status = "FAIL"

    else:
        consistency_status = "PASS"

    consistency_record = {
        "job_key": job_key,
        "product": product,
        "year": year,
        "status": (
            consistency_status
        ),
        "shard_count": int(
            len(records)
        ),
        "failed_shard_count": int(
            failed_shard_count
        ),
        "crs_values": (
            observed_crs_values
        ),
        "band_counts": (
            observed_band_counts
        ),
        "x_resolutions": (
            observed_x_resolutions
        ),
        "y_resolutions": (
            observed_y_resolutions
        ),
        "dtypes": [
            list(dtype_tuple)
            for dtype_tuple
            in observed_dtypes
        ],
        "errors": (
            consistency_errors
        ),
    }

    consistency_records.append(
        consistency_record
    )

    if consistency_errors:
        consistency_failures.append(
            consistency_record
        )

    print(
        f"{job_key:<32} | "
        f"{consistency_status:<4} | "
        f"shards={len(records):>4} | "
        f"CRS={observed_crs_values} | "
        f"bands={observed_band_counts} | "
        f"res_x={observed_x_resolutions}"
    )

    if consistency_errors:
        for error in consistency_errors:
            print(
                "   ERROR:",
                error,
            )


# -------------------------------------------------------------------
# SUMMARY COUNTS
# -------------------------------------------------------------------

passed_physical_file_count = sum(
    record["status"] == "PASS"
    for record
    in physical_validation_records
)


failed_physical_file_count = (
    len(
        physical_validation_records
    )
    - passed_physical_file_count
)


logical_consistency_pass_count = sum(
    record["status"] == "PASS"
    for record
    in consistency_records
)


overall_success = (
    found_logical_export_count == 31
    and len(
        missing_job_keys
    ) == 0
    and total_matched_physical_files > 0
    and failed_physical_file_count == 0
    and logical_consistency_pass_count == 31
)


# -------------------------------------------------------------------
# PRINT SUMMARY
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("VALIDATION SUMMARY")
print("=" * 100)

print(
    "Expected logical exports:",
    31,
)

print(
    "Logical exports found:",
    found_logical_export_count,
)

print(
    "Logical exports missing:",
    len(
        missing_job_keys
    ),
)

print(
    "Total GeoTIFF files discovered:",
    len(
        all_tiff_files
    ),
)

print(
    "Physical GeoTIFF shards matched:",
    total_matched_physical_files,
)

print(
    "Physical shards passed:",
    passed_physical_file_count,
)

print(
    "Physical shards failed:",
    failed_physical_file_count,
)

print(
    "Logical consistency checks passed:",
    f"{logical_consistency_pass_count}/31",
)

print(
    "Unmatched GeoTIFF files:",
    len(
        unmatched_tiff_files
    ),
)

print(
    "Total GeoTIFF size:",
    f"{total_tiff_size_bytes / (1024 ** 3):,.3f} GiB",
)


# -------------------------------------------------------------------
# BUILD VALIDATION DOCUMENT
# -------------------------------------------------------------------

checked_at = datetime.now().isoformat(
    timespec="seconds"
)


validation_document = {
    "checked_at": checked_at,
    "status": (
        "success"
        if overall_success
        else "failed"
    ),
    "drive_folder_prefix": (
        drive_folder_name
    ),
    "candidate_export_folders": [
        str(folder)
        for folder
        in candidate_export_folders
    ],
    "expected_logical_export_count": 31,
    "found_logical_export_count": int(
        found_logical_export_count
    ),
    "missing_job_keys": (
        missing_job_keys
    ),
    "physical_geotiff_discovered_count": int(
        len(
            all_tiff_files
        )
    ),
    "physical_geotiff_matched_count": int(
        total_matched_physical_files
    ),
    "physical_pass_count": int(
        passed_physical_file_count
    ),
    "physical_fail_count": int(
        failed_physical_file_count
    ),
    "logical_consistency_pass_count": int(
        logical_consistency_pass_count
    ),
    "unmatched_tiff_files": [
        str(path)
        for path
        in unmatched_tiff_files
    ],
    "multiple_match_files": (
        multiple_match_files
    ),
    "total_tiff_size_bytes": int(
        total_tiff_size_bytes
    ),
    "total_tiff_size_gib": round(
        total_tiff_size_bytes
        / (1024 ** 3),
        6,
    ),
    "logical_exports": (
        logical_export_records
    ),
    "logical_consistency": (
        consistency_records
    ),
    "physical_geotiffs": (
        physical_validation_records
    ),
}


# -------------------------------------------------------------------
# SAVE JSON REPORT
# -------------------------------------------------------------------

with VALIDATION_JSON_PATH.open(
    "w",
    encoding="utf-8",
) as json_file:
    json.dump(
        validation_document,
        json_file,
        indent=2,
    )


# -------------------------------------------------------------------
# SAVE PHYSICAL GEOTIFF CSV
# -------------------------------------------------------------------

physical_csv_fields = [
    "job_key",
    "product",
    "year",
    "status",
    "file",
    "file_name",
    "file_size_bytes",
    "file_size_mib",
    "driver",
    "width",
    "height",
    "band_count",
    "expected_band_count",
    "crs",
    "x_resolution",
    "y_resolution",
    "expected_resolution_m",
    "dtypes",
    "nodata_values",
    "band_descriptions",
    "tiled",
    "compression",
    "errors",
    "warnings",
]


with VALIDATION_CSV_PATH.open(
    "w",
    newline="",
    encoding="utf-8",
) as csv_file:
    writer = csv.DictWriter(
        csv_file,
        fieldnames=physical_csv_fields,
        extrasaction="ignore",
    )

    writer.writeheader()

    for record in physical_validation_records:
        csv_record = dict(
            record
        )

        for field in [
            "dtypes",
            "nodata_values",
            "band_descriptions",
            "errors",
            "warnings",
        ]:
            csv_record[field] = json.dumps(
                csv_record.get(
                    field,
                    [],
                )
            )

        writer.writerow(
            csv_record
        )


# -------------------------------------------------------------------
# SAVE LOGICAL EXPORT CSV
# -------------------------------------------------------------------

logical_csv_fields = [
    "job_key",
    "product",
    "year",
    "filename_prefix",
    "logical_status",
    "shard_count",
    "total_size_bytes",
    "total_size_gib",
]


with LOGICAL_EXPORT_CSV_PATH.open(
    "w",
    newline="",
    encoding="utf-8",
) as csv_file:
    writer = csv.DictWriter(
        csv_file,
        fieldnames=logical_csv_fields,
    )

    writer.writeheader()

    writer.writerows(
        logical_export_records
    )


# -------------------------------------------------------------------
# FINAL RESULT
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("FINAL VALIDATION RESULT")
print("=" * 100)


if overall_success:
    print(
        "SUCCESS: All 31 logical India exports were found."
    )

    print(
        "Every matched GeoTIFF shard passed metadata "
        "and read-access validation."
    )

    print(
        "CRS, resolution, band count and dtype were "
        "consistent within every logical export."
    )

    print(
        "The acquisition is ready for the final "
        "content-range and checksum validation."
    )

else:
    print(
        "ATTENTION: One or more inventory or GeoTIFF "
        "validation checks failed."
    )

    print(
        "Do not upload the India files to HDFS yet."
    )

    if missing_job_keys:
        print(
            "\nMissing logical exports:"
        )

        for job_key in missing_job_keys:
            print(
                " -",
                job_key,
            )

    if invalid_file_records:
        print(
            "\nFailed physical GeoTIFF shards:"
        )

        for record in invalid_file_records:
            print(
                " -",
                record["file"],
            )

            for error in record["errors"]:
                print(
                    "    ERROR:",
                    error,
                )

    if consistency_failures:
        print(
            "\nFailed logical consistency checks:"
        )

        for record in consistency_failures:
            print(
                " -",
                record["job_key"],
            )

            for error in record["errors"]:
                print(
                    "    ERROR:",
                    error,
                )


print("\nValidation JSON:")
print(
    VALIDATION_JSON_PATH
)

print("\nPhysical GeoTIFF CSV:")
print(
    VALIDATION_CSV_PATH
)

print("\nLogical export inventory CSV:")
print(
    LOGICAL_EXPORT_CSV_PATH
)
```

```python
# -------------------------------------------------------------------
# MONITOR COMPLETE INDIA EXPORT TASKS
# -------------------------------------------------------------------
#
# Safe to rerun.
# This cell does not create or resubmit export tasks.
#
# It reads task IDs from the saved submission manifest, so it can
# recover status even if the notebook runtime restarts.

from pathlib import Path
from datetime import datetime
import csv
import json

import ee


PROJECT_ID = "india-ghsl-urbanization"

CONTROL_FOLDER = Path(
    "/content/drive/MyDrive/GHSL_India_Export_Control"
)

SUBMISSION_MANIFEST_PATH = (
    CONTROL_FOLDER
    / "india_export_submission_latest.json"
)

STATUS_JSON_PATH = (
    CONTROL_FOLDER
    / "india_export_status_latest.json"
)

STATUS_CSV_PATH = (
    CONTROL_FOLDER
    / "india_export_status_latest.csv"
)


# -------------------------------------------------------------------
# INITIALIZE EARTH ENGINE
# -------------------------------------------------------------------

try:
    ee.Initialize(project=PROJECT_ID)
    assert ee.Number(1).getInfo() == 1

except Exception:
    ee.Authenticate(auth_mode="notebook")
    ee.Initialize(project=PROJECT_ID)

    assert ee.Number(1).getInfo() == 1


# -------------------------------------------------------------------
# LOAD SUBMISSION MANIFEST
# -------------------------------------------------------------------

if not SUBMISSION_MANIFEST_PATH.exists():
    raise FileNotFoundError(
        "Submission manifest not found:\n"
        f"{SUBMISSION_MANIFEST_PATH}"
    )


with SUBMISSION_MANIFEST_PATH.open(
    "r",
    encoding="utf-8",
) as manifest_file:
    submission_manifest = json.load(
        manifest_file
    )


submission_records = (
    submission_manifest["tasks"]
)


if len(submission_records) != 31:
    raise RuntimeError(
        "Expected 31 task records, but found "
        f"{len(submission_records)}."
    )


# -------------------------------------------------------------------
# INDEX CURRENT EARTH ENGINE TASKS BY TASK ID
# -------------------------------------------------------------------

current_tasks = ee.batch.Task.list()

tasks_by_id = {
    task.id: task
    for task in current_tasks
}


# -------------------------------------------------------------------
# CHECK TASK STATUS
# -------------------------------------------------------------------

status_records = []

state_counts = {}


print("=" * 100)
print("COMPLETE INDIA EXPORT STATUS")
print("=" * 100)

checked_at = datetime.now().isoformat(
    timespec="seconds"
)

print("Checked at:", checked_at)


for index, record in enumerate(
    submission_records,
    start=1,
):
    task_id = record["task_id"]
    job_key = record["job_key"]

    task = tasks_by_id.get(
        task_id
    )

    if task is None:
        state = "NOT_FOUND"
        error_message = (
            "Task is not visible in the current "
            "Earth Engine task listing."
        )
        description = record["description"]

    else:
        status = task.status()

        state = status.get(
            "state",
            "UNKNOWN",
        )

        error_message = status.get(
            "error_message",
            None,
        )

        description = status.get(
            "description",
            record["description"],
        )

    state_counts[state] = (
        state_counts.get(
            state,
            0,
        )
        + 1
    )

    status_record = {
        "job_key": job_key,
        "product": record["product"],
        "year": record["year"],
        "task_id": task_id,
        "description": description,
        "state": state,
        "error_message": error_message,
        "asset_id": record["asset_id"],
        "scale": record["scale"],
        "filename_prefix": (
            record["filename_prefix"]
        ),
        "drive_folder": (
            submission_manifest[
                "drive_folder"
            ]
        ),
    }

    status_records.append(
        status_record
    )

    error_display = (
        error_message
        if error_message
        else "None"
    )

    print(
        f"{index:>2}/31 "
        f"{job_key:<32} | "
        f"{state:<10} | "
        f"Error: {error_display}"
    )


# -------------------------------------------------------------------
# SUMMARY
# -------------------------------------------------------------------

print("\n" + "=" * 100)
print("EXPORT STATUS SUMMARY")
print("=" * 100)


for state in [
    "COMPLETED",
    "RUNNING",
    "READY",
    "FAILED",
    "CANCELLED",
    "NOT_FOUND",
    "UNKNOWN",
]:
    count = state_counts.get(
        state,
        0,
    )

    if count > 0:
        print(
            f"{state:<12}: {count}"
        )


completed_count = state_counts.get(
    "COMPLETED",
    0,
)

failed_count = state_counts.get(
    "FAILED",
    0,
)

cancelled_count = state_counts.get(
    "CANCELLED",
    0,
)

not_found_count = state_counts.get(
    "NOT_FOUND",
    0,
)

all_completed = (
    completed_count == 31
)

has_problem = (
    failed_count > 0
    or cancelled_count > 0
    or not_found_count > 0
)


print("\n" + "=" * 100)
print("OVERALL STATUS")
print("=" * 100)


if all_completed:
    print(
        "SUCCESS: All 31 India export tasks completed."
    )

    print(
        "The next step is Google Drive inventory and "
        "GeoTIFF-shard validation."
    )

elif has_problem:
    print(
        "ATTENTION: One or more tasks failed, were cancelled, "
        "or are no longer visible."
    )

    print(
        "Do not rerun all exports. Inspect and resubmit only "
        "the affected job keys."
    )

else:
    print(
        "IN PROGRESS: Tasks are still running or queued."
    )

    print(
        "Wait and rerun only this monitoring cell."
    )


# -------------------------------------------------------------------
# SAVE STATUS REPORTS
# -------------------------------------------------------------------

status_document = {
    "checked_at": checked_at,
    "project_id": PROJECT_ID,
    "task_count": len(
        status_records
    ),
    "state_counts": state_counts,
    "all_completed": all_completed,
    "has_problem": has_problem,
    "tasks": status_records,
}


with STATUS_JSON_PATH.open(
    "w",
    encoding="utf-8",
) as status_file:
    json.dump(
        status_document,
        status_file,
        indent=2,
    )


csv_fields = [
    "job_key",
    "product",
    "year",
    "task_id",
    "description",
    "state",
    "error_message",
    "asset_id",
    "scale",
    "filename_prefix",
    "drive_folder",
]


with STATUS_CSV_PATH.open(
    "w",
    newline="",
    encoding="utf-8",
) as csv_file:
    writer = csv.DictWriter(
        csv_file,
        fieldnames=csv_fields,
    )

    writer.writeheader()

    writer.writerows(
        status_records
    )


print("\nStatus JSON:")
print(STATUS_JSON_PATH)

print("\nStatus CSV:")
print(STATUS_CSV_PATH)

print(
    "\nThis cell is safe to rerun and does not submit exports."
)
```
